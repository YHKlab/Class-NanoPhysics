## Basic Theory

### Continuity equations (module `continuity`, Neamen Ch. 6)

$$\frac{1}{q}\frac{dJ_p}{dx}=G-R,\qquad \frac{1}{q}\frac{dJ_n}{dx}=R-G\qquad(\text{steady state, Sec. 6.2.1}),\qquad R=\frac{np-n_i^2}{\tau_{p0}(n+n')+\tau_{n0}(p+p')}\ \ (6.99),\ \ n'=p'=n_i$$

The currents are the drift–diffusion currents of Lab 4 (module `drift_diffusion`, (5.35)); the densities and quasi-Fermi levels follow the Boltzmann relations of the `statistics` module, $n=n_ie^{(E_{Fn}-E_{Fi})/kT}$, $p=n_ie^{(E_{Fi}-E_{Fp})/kT}$ (6.85a), (6.85b).

### Ends of the bar

- **Ohmic contact:** the contact holds the equilibrium (or injected) densities, $\delta=0$ (or $\delta(0)$); it is a perfect sink of excess carriers (infinite surface recombination velocity).
- **Free end:** no contact and no surface recombination (Sec. 6.6 with zero surface recombination velocity): no current through the surface, $d\delta/dx=0$. Without a current path the bias must be $V=0$.

### Textbook assumptions

| Assumption | Where Neamen uses it | This simulation |

| **Complete ionization** (Sec. 4.4.2) and charge neutrality (4.56) | $n_0$, $p_0$ (4.60), (4.62) | Kept |

| **Extrinsic doping**, $p_0\gg n_0$ (Sec. 6.3.2) | (6.45)–(6.48) | Not assumed |
| **Charge neutrality** $\delta n\approx\delta p$ and **small internal field** $\lvert\mathcal{E}_{int}\rvert\ll\lvert\mathcal{E}_{app}\rvert$ (6.33) | Derivation of the ambipolar transport equation (6.39) | Imposed: $\delta n=\delta p$ (the Poisson equation is not solved in this lab; see Lab 6). The internal (Dember) field is kept |
| **SRH recombination**: single trap level, trap density not too large (Sec. 6.5.1, (6.99)); midgap trap $n'\approx p'\approx n_i$ for (6.101)–(6.103) | $R=\delta n/\tau_{n0}$ or $\delta p/\tau_{p0}$ at low injection | Kept: $n'=p'=n_i$, constant $\tau_{n0}$, $\tau_{p0}$; no Auger or radiative recombination; **SRH = Off** sets $R=0$ |
| **Infinite sample** / surfaces (Sec. 6.6) | Examples of Sec. 6.3.3 assume an infinite crystal | Free ends: zero surface recombination velocity; ohmic contacts: infinite |
| **Steady state** and the simplifications of Table 6.2 | Applications of the ambipolar equation | Steady state only |

**Simulation model.** One dimension; one lattice temperature; constant $E_g$, $\varepsilon_r$ and mobilities; $n_i(T)=n_i(300)(T/300)^{3/2}e^{-E_g/2k(1/T-1/300)}$; prescribed generation $G(x)$ (uniform or $G_0e^{-s/d}$, no reflection); uniform grid with Scharfetter–Gummel currents.

### Analytic models used for comparison

| Structure / condition | Quantity | Analytic model [1] |
| --- | --- | --- |
| Uniform bar, $G>0$ | Excess carriers, quasi-Fermi levels | No contacts (infinite sample): $\delta n=\delta p=G\tau_{minority}$ and the exact SRH steady state $R_{SRH}(n_0+\delta,p_0+\delta)=G$; ohmic contacts at both ends: $\delta(x)=G\tau\left[1-\frac{\cosh(x/L)}{\cosh(L_{bar}/2L)}\right]$ ($x$ from the center); SRH off: $\delta=Gs(L_{bar}-s)/2D$; $E_F-E_{Fi}=kT\ln(n_0/n_i)$ (6.84a); $E_{Fn}-E_{Fi}$, $E_{Fi}-E_{Fp}$ (6.85a), (6.85b) |

<h3>Grid and units</h3>

<p>The bar is sampled on a uniform grid $x_i=i\,\Delta x$ (1001 points by default, or the case preset). Internally lengths are in cm, potentials in V, times in s and concentrations in cm⁻³.</p>

<h3>Continuity equation (module <code>continuity</code>)</h3>

<p>With the potential fixed, the Scharfetter–Gummel currents of the <code>drift_diffusion</code> module make the continuity equation of each carrier a linear tridiagonal system (<code>drift_diffusion.flux_operator</code> plus $G-R$). SRH is linearized with its denominator fixed, $R\approx n_{other}\,n_{own}/\mathrm{den}-n_i^2/\mathrm{den}$, and the system is solved with <code>scipy.linalg.solve_banded</code>. Ohmic end nodes are fixed and removed from the system; at a free end the first node owns a half cell and no current crosses the surface.</p>

<h3>Quasi-neutral iteration</h3>

<p>(1) From the present $\delta$, the total current is set by the bias, $V=\int\mathcal{E}\,dx$ with $\mathcal{E}=[J/e-(D_n-D_p)\delta']/(\mu_nn+\mu_pp)$, or $J=0$ with a free end; (2) the potential $\psi=\psi_{eq}+V-\int_0^x\mathcal{E}\,dx'$ is built from that field and the continuity equation of the minority carrier is solved; (3) $n=n_0+\delta$, $p=p_0+\delta$. The loop stops when $\max\lvert\Delta\delta\rvert/(n_{own}+n_i)<10^{-7}$; it needs 4–40 iterations.</p>

<h3>Ambipolar sweep (5-C)</h3>

<p><code>run_ambipolar_sweep</code> solves each p-type bar at $V=0$ and at $V$. In each solution it fits $\ln\delta n$ between $0.6\,\delta(0)$ and $0.2\,\delta(0)$ (within $0.8L$ of the injecting contact) to get $\lambda$, and takes the medians of $\tau'=\delta n/R$, of the current-driven field $\mathcal{E}_J=J/\sigma$ and of $\delta n$ in that window. The two equations $D'-\mu'\mathcal{E}_J\lambda=\lambda^2/\tau'$ give $D'$ and $\mu'$; $\mu'$ is reported only if the drift changes $\lambda^2/\tau'$ by more than 30 % (<code>drift_min</code>). The formulas are evaluated at $n=n_0+\delta n$, $p=N_a+\delta n$ with the median $\delta n$ of each bar or with $\delta n=\sqrt{0.6\times0.2}\,\delta(0)$ for the continuous curves.</p>

# (1) Calculation

Modules of `semi1d` used by this lab, then the parts written in this notebook: units, materials and device; the lab model (how the modules are coupled) and its analysis; runners and case presets.

```python
###############################################################################
# Imports: modules of semi1d (folder one level above this notebook)
###############################################################################

import os
import re
import sys
sys.path.insert(0, os.path.abspath('..'))

import numpy as np
from dataclasses import dataclass, field
from types import SimpleNamespace
from scipy import constants as sc_const

from semi1d import statistics          # Ch. 3-4: n_i(T), equilibrium densities, quasi-Fermi levels
from semi1d import drift_diffusion     # Ch. 5:   drift and diffusion currents, Einstein relation
from semi1d import continuity          # Ch. 6:   continuity equations, SRH recombination, ambipolar coefficients
```

```python
###############################################################################
# Units, Materials and Device
###############################################################################
# Internal units: length [cm], potential [V], energy [eV], time [s], density [cm^-3]; positions shown in [um].

Q = sc_const.e                          # Elementary charge [C]
EPS0 = sc_const.epsilon_0 * 1e-2        # Vacuum permittivity [F/cm]
UM2CM, US2S = 1e-4, 1e-6


@dataclass(frozen=True)
class Material:
    name: str
    eps_r: float        # Relative permittivity
    Eg: float           # Band gap [eV]
    ni300: float        # Intrinsic carrier concentration at 300 K [cm^-3]
    mun: float          # Electron mobility [cm^2/V s]
    mup: float          # Hole mobility [cm^2/V s]


# 300 K values: Neamen (2012), Table 5.1 (mobility) and Appendix B.4
MATERIALS = {
    'Si':   Material('Si',   11.7, 1.12, 1.5e10, 1350.0,  480.0),
    'Ge':   Material('Ge',   16.0, 0.66, 2.4e13, 3900.0, 1900.0),
    'GaAs': Material('GaAs', 13.1, 1.42, 1.8e6,  8500.0,  400.0),
}


@dataclass
class GridConfig:
    points_per_debye: float = 8.0       # Auto grid: dx <= L_D(min) / points_per_debye
    npts_min: int = 401                 # Auto grid: lower limit of grid points
    npts_max: int = 20001               # Auto grid: upper limit of grid points
    npts_uniform: int = 1001            # Auto grid for a uniform bar (no built-in space charge)


GRID_CFG = GridConfig()


class Device:
    """
    1D device: grid x [um] (measured from the left contact), doping N_a(x), N_d(x) [cm^-3], material and
    transport parameters (n_i(T) from semi1d.statistics, D from the Einstein relation in semi1d.drift_diffusion).
    x0_um is the plotting origin (junction or device center); srh=False removes SRH recombination (R = 0).
    bc_left / bc_right: 'ohmic' contact (equilibrium concentrations) or 'free' surface (no carrier flux).
    """

    def __init__(self, x_um, Na, Nd, material='Si', T=300.0, ni300=None, mun=None, mup=None,
                 taun=1e-7, taup=1e-7, srh=True, x0_um=None, bc_left='ohmic', bc_right='ohmic'):
        mat = MATERIALS[material] if isinstance(material, str) else material
        self.material = mat
        self.x_um = np.asarray(x_um, dtype=float)
        self.x = self.x_um * UM2CM                                  # [cm]
        self.N = len(self.x)
        self.dx = self.x[1] - self.x[0]
        self.Na = np.broadcast_to(np.asarray(Na, dtype=float), self.x.shape).copy()
        self.Nd = np.broadcast_to(np.asarray(Nd, dtype=float), self.x.shape).copy()
        self.C = self.Nd - self.Na                                  # Net doping
        self.T = float(T)
        self.Vt = statistics.thermal_voltage(self.T)
        self.Eg = mat.Eg
        self.ni = statistics.intrinsic_density(self.T, mat.Eg, mat.ni300 if ni300 is None else float(ni300))
        self.mun = mat.mun if mun is None else float(mun)
        self.mup = mat.mup if mup is None else float(mup)
        self.Dn, self.Dp = drift_diffusion.einstein(self.mun, self.Vt), drift_diffusion.einstein(self.mup, self.Vt)
        self.taun, self.taup = float(taun), float(taup)
        self.srh = bool(srh)
        self.tau_eff_n = self.taun if self.srh else np.inf      # Used by the analytic models
        self.tau_eff_p = self.taup if self.srh else np.inf
        self.eps = mat.eps_r * EPS0
        self.x0_um = 0.5 * self.x_um[-1] if x0_um is None else float(x0_um)
        if bc_left not in ('ohmic', 'free') or bc_right not in ('ohmic', 'free'):
            raise ValueError("Ends must be 'ohmic' (contact) or 'free' (no surface recombination).")
        self.bc_left, self.bc_right = bc_left, bc_right


def debye_length_um(N, eps, Vt):
    """Extrinsic Debye length (eps kT / e^2 N)^{1/2} [um] (used to choose the grid)."""
    return np.sqrt(eps * Vt / (Q * N)) / UM2CM


def grid_points(uniform, L_um, Nmax, eps, Vt, npts=None):
    """Grid points: npts, or automatic (1001 for a uniform bar; otherwise dx <= L_D(Nmax)/8, 401-20001)."""
    if npts is None:
        if uniform:
            return GRID_CFG.npts_uniform
        LD = debye_length_um(Nmax, eps, Vt)
        npts = int(np.clip(int(np.ceil(GRID_CFG.points_per_debye * L_um / LD)) + 1, GRID_CFG.npts_min,
                           GRID_CFG.npts_max))
    if int(npts) < 11:
        raise ValueError("Use at least 11 grid points.")
    return int(npts)


def check_positive(**values):
    bad = [k for k, v in values.items() if not np.isfinite(v) or v <= 0]
    if bad:
        raise ValueError(f"Lengths, temperature, lifetimes and doping concentrations must be positive: {bad}")


def print_device(dev, structure_name, say, origin='device center'):
    """Device and grid block of the log."""
    LD_min = debye_length_um(float(np.max(np.maximum(dev.Na, dev.Nd))), dev.eps, dev.Vt)
    say('(0) Device and grid')
    say(f'structure \t = {structure_name}, {dev.material.name}, T = {dev.T:g} K')
    say(f'device length \t = {dev.x_um[-1]:.3f} µm, number of grid = {dev.N}, dx = {dev.dx / UM2CM * 1e3:.3f} nm')
    say(f'x range \t = {-dev.x0_um:.3f} to {dev.x_um[-1] - dev.x0_um:.3f} µm (x = 0 at the {origin})')
    say(f'Eg, ni, kT/e \t = {dev.Eg:.3g} eV, {dev.ni:.4e} cm^-3, {dev.Vt:.5f} V')
    say(f'μn, μp \t = {dev.mun:g}, {dev.mup:g} cm^2/V s  (Dn, Dp = {dev.Dn:.4g}, {dev.Dp:.4g} cm^2/s)')
    return LD_min


###############################################################################
# Comparison Table and Results
###############################################################################

ITER_ROWS = ('iters',)        # Printed first in every table


def _is_num(v):
    return isinstance(v, (int, float, np.integer, np.floating)) and not isinstance(v, bool) and np.isfinite(v)


def plain_sub(s):
    """Subscripts in printed text, where they cannot be typeset: 'E_g', 'μ_n', 'N_{a}' -> 'Eg', 'μn', 'Na'."""
    return re.sub(r'(?<=[^\W\d_])_(\{[^}]*\}|[A-Za-z0-9]+)', lambda m: m.group(1).strip('{}'), s) if isinstance(s, str) else s


def row(key, name, sim, ref=None, model=''):
    """One line of the comparison table. diff = (sim - ref)/|ref| when both are numbers. The key is the name used
    by res.value(key); subscripts in the quantity and the model are written without underscores (plain_sub)."""
    sim = float(sim) if _is_num(sim) else sim
    ref = float(ref) if _is_num(ref) else ref
    diff = (sim - ref) / abs(ref) if _is_num(sim) and _is_num(ref) and ref != 0 else None
    return dict(key=key, name=plain_sub(name), sim=sim, ref=ref, diff=diff, model=plain_sub(model))


def _fmt(v, width=12):
    if isinstance(v, str):
        s = v
    elif not _is_num(v):                          # None or NaN
        s = '—'
    elif isinstance(v, (int, np.integer)):
        s = f'{v:d}'
    else:
        s = f'{v:.4g}'
    return f'{s:>{width}s}'


def select_rows(rows, keys=None):
    """Rows to print: iteration rows first, then the requested keys in the given order (None = all rows)."""
    if keys is None:
        return rows
    by_key = {r['key']: r for r in rows}
    out = [by_key[k] for k in ITER_ROWS if k in by_key]
    out += [by_key[k] for k in keys if k in by_key and k not in ITER_ROWS]
    missing = [k for k in keys if k not in by_key]
    if missing:
        print(f'(not applicable for these parameters: {", ".join(missing)})')
    return out


def print_table(rows):
    """Print the comparison table: Key | Quantity | Simulation | Analytic | Diff. | Model"""
    print(f"{'Key':<14s} {'Quantity':<46s} {'Simulation':>12s} {'Analytic':>12s} {'Diff.':>8s}  Model")
    print('-' * 118)
    for r in rows:
        diff = f"{100 * r['diff']:+.1f}%" if r['diff'] is not None else '—'
        print(f"{r['key']:<14s} {r['name']:<46s} {_fmt(r['sim'])} {_fmt(r['ref'])} {diff:>8s}  {r['model']}")


@dataclass
class SimulationResult:
    """Result of a case: res.table (comparison rows), res.data (arrays), res.value(key)."""
    params: dict
    device: object
    states: list
    table: list
    data: dict = field(default_factory=dict)
    case: str = None
    panels: list = None

    def row(self, key):
        for r in self.table:
            if r['key'] == key:
                return r
        raise KeyError(f"No table row '{key}'. Available keys: {[r['key'] for r in self.table]}")

    def value(self, key, which='sim'):
        """Table value by key; which = 'sim', 'ref' or 'diff'."""
        return self.row(key)[which]


###############################################################################
# Case Runner
###############################################################################
# CASES = {case_id: dict(title, summary, params, ranges, fixed, runner, rows, cols, panels, schematic)}

def to_float(text, name='value'):
    try:
        return float(str(text).strip())
    except ValueError:
        raise ValueError(f"'{name}' must be a number such as 1e16; got {text!r}.") from None


def run_case(case_id, verbose=True, **overrides):
    """Run a case, e.g. run_case(list(CASES)[0], T=350.0). Parameters not given keep their presets."""
    case = CASES[case_id]
    bad = set(overrides) - set(case['params'])
    if bad:
        raise KeyError(f"Case {case_id} has no parameter {sorted(bad)}; available: {list(case['params'])}")
    kw = {**case.get('fixed', {}), **case_kwargs({**case['params'], **overrides})}
    if verbose:
        print(f"Case {case_id}: {case['title']}\n")
    if 'cols' in case:
        kw['cols'] = case['cols']
    res = case['runner'](verbose=verbose, rows=case.get('rows'), **kw)
    res.case, res.panels = case_id, case.get('panels')
    return res


def list_cases():
    """Print the case IDs, titles and parameter presets."""
    for cid, case in CASES.items():
        ps = ', '.join(f'{k}={v:g}' if isinstance(v, float) else f'{k}={v!r}' for k, v in case['params'].items())
        print(f"{cid}  {case['title']}\n      run_case('{cid}', {ps})")
```

```python
###############################################################################
# Lab 5 Device: uniformly doped bar
###############################################################################

STRUCTURES = {0: 'Uniform bar'}


def build_bar(doping_type='n', N=1e16, length_um=100.0, material='Si', T=300.0, ni300=None, mun=None, mup=None,
              taun_us=0.1, taup_us=0.1, srh=True, bc_left='ohmic', bc_right='ohmic', npts=None):
    """Uniformly doped bar (n- or p-type) on a uniform grid; x = 0 of the plots at the center."""
    check_positive(length_um=length_um, T=T, taun_us=taun_us, taup_us=taup_us, N=N)
    mat = MATERIALS[material]
    npts = grid_points(True, length_um, N, mat.eps_r * EPS0, statistics.thermal_voltage(T), npts)
    x = np.linspace(0.0, length_um, npts)
    zero, dop = np.zeros_like(x), np.full_like(x, N)
    Na, Nd = (dop, zero) if doping_type == 'p' else (zero, dop)
    return Device(x, Na, Nd, material=mat, T=T, ni300=ni300, mun=mun, mup=mup, taun=taun_us * US2S,
                  taup=taup_us * US2S, srh=srh, bc_left=bc_left, bc_right=bc_right)


###############################################################################
# Coupling of the Modules (quasi-neutral model, Neamen Sec. 6.3)
###############################################################################
# Charge neutrality delta_n = delta_p = delta and a small internal field (6.33), as in the derivation of the
# ambipolar transport equation:
#   (1) total current J from the bias:  V = integral E dx with  E = (J/q - (D_n - D_p) d(delta)/dx) / (mu_n n + mu_p p)
#       (drift_diffusion.quasi_neutral_field; J = 0 with a free end)
#   (2) continuity.solve_continuity for the minority carrier with that field                       -> delta
#   (3) n = n0 + delta, p = p0 + delta; repeat until delta stops changing

@dataclass
class SolverConfig:
    maxit: int = 3000                   # Maximum number of iterations
    tol_carrier: float = 1e-7           # Convergence: max|d delta|/(n_own + n_i)


SIM_CFG = SolverConfig()


class ExcessSolver:
    """Excess carriers in a uniform bar, quasi-neutral model (statistics, drift_diffusion and continuity modules);
    the equilibrium densities are the charge-neutral densities of semi1d.statistics."""

    def __init__(self, dev):
        self.d = dev
        self.psi_eq = statistics.neutral_potential(dev.C, dev.ni, dev.Vt)
        self.n_eq, self.p_eq = statistics.boltzmann_densities(self.psi_eq, dev.ni, dev.Vt)

    def solve(self, V=0.0, G=0.0, inj=0.0, maxit=None):
        """Steady state at the bias V of the left contact (right contact grounded), generation G, injection inj."""
        G = np.broadcast_to(np.asarray(G, dtype=float), (self.d.N,)).copy()
        return self._quasi_neutral(V, G, inj, maxit or SIM_CFG.maxit)

    def _quasi_neutral(self, V, G, inj, maxit):
        d = self.d
        minority = 'p' if d.C[0] > 0 else 'n'            # Carrier solved with the continuity module
        own_eq = self.p_eq if minority == 'p' else self.n_eq
        delta = np.zeros(d.N)
        if d.bc_left == 'ohmic':
            delta[0] = inj
        free = 'free' in (d.bc_left, d.bc_right)
        hist, converged = [], False
        for it in range(1, maxit + 1):
            n, p = self.n_eq + delta, self.p_eq + delta
            nh, ph = 0.5 * (n[1:] + n[:-1]), 0.5 * (p[1:] + p[:-1])         # Midpoints
            ddelta = np.diff(delta) / d.dx
            sig = d.mun * nh + d.mup * ph                                  # sigma/q
            if free:
                J = 0.0                                                    # No current through a free surface
            else:                                                          # J from V = integral E dx
                J = Q * (V + (d.Dn - d.Dp) * np.sum(ddelta / sig) * d.dx) / (np.sum(1.0 / sig) * d.dx)
            Eh = drift_diffusion.quasi_neutral_field(d, nh, ph, ddelta, ddelta, J)        # (1)
            psi = self.psi_eq + V - np.concatenate([[0.0], np.cumsum(Eh) * d.dx])
            own = continuity.solve_continuity(d, psi, n, p, G, minority)                  # (2)
            new = own - own_eq
            change = float(np.max(np.abs(new - delta) / (own + d.ni)))
            delta = new                                                                   # (3)
            hist.append((it, 0.0, change, change))
            if change < SIM_CFG.tol_carrier:
                converged = True
                break
        n, p = self.n_eq + delta, self.p_eq + delta
        nh, ph = 0.5 * (n[1:] + n[:-1]), 0.5 * (p[1:] + p[:-1])
        ddelta = np.diff(delta) / d.dx
        Eh = drift_diffusion.quasi_neutral_field(d, nh, ph, ddelta, ddelta, J)
        psi = self.psi_eq + V - np.concatenate([[0.0], np.cumsum(Eh) * d.dx])
        return self.pack(V, psi, n, p, G, inj, hist, converged)

    def pack(self, V, psi, n, p, G, inj, hist, converged):
        d = self.d
        J = drift_diffusion.currents(d, psi, n, p)
        bands = statistics.band_diagram(psi, n, p, d.ni, d.Vt, d.Eg)
        xh = 0.5 * (d.x_um[1:] + d.x_um[:-1])
        return SimpleNamespace(
            V=V, x=d.x_um, xh=xh, xc=d.x_um - d.x0_um, xhc=xh - d.x0_um, psi=psi, n=n, p=p,
            dn=n - self.n_eq, dp=p - self.p_eq, E=np.gradient(-psi, d.x), rho=p - n + d.C,
            R=continuity.srh_rate(d, n, p), G=G, **bands, **vars(J), J=float(np.mean(J.Jt)), hist=hist,
            iters=len(hist), converged=converged, inj=inj, dev=d)


###############################################################################
# Lab 5 Analysis
###############################################################################

def decay_length(x, y):
    """Distance at which y(x) falls to y(0)/e (linear interpolation)."""
    y0 = y[0]
    idx = np.where(y < y0 / np.e)[0]
    if y0 <= 0 or len(idx) == 0 or idx[0] == 0:
        return np.nan
    i = idx[0]
    return float(np.interp(y0 / np.e, [y[i], y[i - 1]], [x[i], x[i - 1]]))


def semilog_length(x, y, lo=0.05, hi=0.8):
    """Decay length from the slope of ln y, using hi > y/y(0) > lo."""
    y0 = y[0]
    m = (y > lo * y0) & (y < hi * y0)
    if y0 <= 0 or m.sum() < 3:
        return np.nan
    return float(-1.0 / np.polyfit(x[m], np.log(y[m]), 1)[0])


def analyze_excess(st, s, p):
    """
    Lab 5: compare the excess carriers of a uniform bar (generation or injection) with Ch. 6 models.

    input:
        SimpleNamespace st      Solved state (Solver.solve)
        Solver          s       Solver (equilibrium solution)
        dict            p       Simulation parameters (run_bar)
    output:
        list            rows    Comparison-table rows
        dict            ov      Analytic overlays for plotting {panel: [(x, y, label, dash), ...]}
    """
    d = st.dev
    x, L = st.x, st.x[-1]
    G, inj = float(np.max(st.G)), st.inj
    rows, ov = [], {k: [] for k in ('E', 'rho', 'psi', 'excess')}
    i = d.N // 2
    n0, p0 = s.n_eq[i], s.p_eq[i]
    ntype = p['doping_type'] == 'n'
    Ndop = p['N']
    if G > 0:
        depth = p.get('gen_depth_um') or 0.0
        D_min, tau_min, Lname = (d.Dp, d.tau_eff_p, 'L_p') if ntype else (d.Dn, d.tau_eff_n, 'L_n')
        dmin = st.dp if ntype else st.dn
        Lm = continuity.diffusion_length(D_min, tau_min) / UM2CM
        ends = (d.bc_left, d.bc_right)
        model = continuity.excess_profile(x * UM2CM, L * UM2CM, G, D_min, tau_min, depth * UM2CM, *ends)
        if depth == 0:          # Uniform generation
            gt = G * tau_min
            if ends == ('free', 'free'):
                exact = continuity.delta_uniform_srh(G, n0, p0, d.ni, d.taun, d.taup)
                rows += [row('dp_center', 'Excess holes δp at the center [cm⁻³]', st.dp[i], gt,
                             'δ = Gτ_minority (low injection)'),
                         row('dn_center', 'Excess electrons δn at the center [cm⁻³]', st.dn[i], gt, 'δn = δp'),
                         row('dp_srh', 'Excess δ at the center, exact SRH [cm⁻³]', dmin[i], exact,
                             'R_SRH(n0 + δ, p0 + δ) = G, any injection'),
                         row('dp_ends', 'δ at the ends / δ at the center', min(dmin[0], dmin[-1]) / dmin[i], 1.0,
                             'no contacts: no sink at the ends')]
                ov['excess'] += [(x, np.full(d.N, gt), 'G τ', 'dot'), (x, np.full(d.N, exact), 'exact SRH', 'dash')]
            else:
                if ends == ('ohmic', 'ohmic'):
                    bar_model = 'Gτ[1 − cosh(x/L)/cosh(L_bar/2L)]' if d.srh else 'G s(L_bar − s)/2D (SRH off)'
                else:
                    bar_model = 'D δ″ − δ/τ = −G; contact δ = 0, free end δ′ = 0'
                rows += [row('dp_center', 'Excess holes δp at the center [cm⁻³]', st.dp[i], gt if d.srh else None,
                             'δ = Gτ_minority (steady state)' if d.srh else 'SRH off'),
                         row('dn_center', 'Excess electrons δn at the center [cm⁻³]', st.dn[i],
                             gt if d.srh else None, 'δn = δp'),
                         row('dp_bar', 'δ at the center, bar with contacts [cm⁻³]', dmin[i], model[i], bar_model)]
                if d.srh:
                    ov['excess'].append((x, np.full(d.N, gt), 'G τ', 'dot'))
                ov['excess'].append((x, model, 'bar with contacts', 'dashdot'))
            rows += [row('R_center', 'Recombination rate R at the center [cm⁻³s⁻¹]', st.R[i], G if d.srh else None,
                         'R = G (steady state)' if d.srh else 'SRH off'),
                     row('EF_Ei', 'Equilibrium E_F − E_Fi [eV]', s.psi_eq[i], d.Vt * np.log(n0 / d.ni),
                         'kT ln(n0/ni) (6.84a)'),
                     row('EFn_Ei', 'E_Fn − E_Fi at the center [eV]', st.Fn[i] - st.Ei[i],
                         d.Vt * np.log((n0 + model[i]) / d.ni), 'kT ln((n0+δn)/ni) (6.85a)'),
                     row('Ei_EFp', 'E_Fi − E_Fp at the center [eV]', st.Ei[i] - st.Fp[i],
                         d.Vt * np.log((p0 + model[i]) / d.ni), 'kT ln((p0+δp)/ni) (6.85b)'),
                     row('injection_G', 'Injection level δ/N at the center', dmin[i] / Ndop, '≪ 1',
                         'low injection, Sec. 6.3.2')]
        else:                   # Absorption near the left surface, G = G0 exp(-s/d)
            d_cm, L_cm = depth * UM2CM, continuity.diffusion_length(D_min, tau_min)
            k0 = int(np.searchsorted(x, 3 * depth))
            m = slice(k0, int(np.searchsorted(x, 0.7 * L)))
            xc_cm = x * UM2CM
            rec = np.trapezoid(st.R, xc_cm) / np.trapezoid(st.G, xc_cm)
            rec_model = np.trapezoid(model / tau_min, xc_cm) / np.trapezoid(st.G, xc_cm)
            rows += [row('d0', 'Excess δ(0) at the illuminated surface [cm⁻³]', dmin[0], model[0],
                         'D δ″ − δ/τ = −G₀e^{−s/d}, δ′(0) = 0'),
                     row('d0_semi', 'δ(0), semi-infinite bar [cm⁻³]', dmin[0],
                         G * tau_min * d_cm / (d_cm + L_cm) if d.srh else None, 'G₀τ d/(d + L)')]
            if d.srh:           # Same fit on the model profile (includes the right contact)
                rows.append(row('L_tail', 'Decay length beyond 3d [µm]', semilog_length(x[m], dmin[m], 0.1, 0.9),
                                semilog_length(x[m], model[m], 0.1, 0.9),
                                f'model, same fit; {Lname} = (Dτ)^1/2 = {Lm:.3g} µm'))
            rows += [row('R_total', 'Recombined fraction ∫R dx / ∫G dx', rec, rec_model,
                         'the rest leaves through the right contact'),
                     row('injection_G', 'Injection level δ(0)/N', dmin[0] / Ndop, '≪ 1', 'low injection, Sec. 6.3.2')]
            ov['excess'].append((x, model, 'diffusion model', 'dash'))
    if inj > 0:
        if ntype:
            dmin, dmaj, Dm, tm, Lname = st.dp, st.dn, d.Dp, d.tau_eff_p, 'L_p'
        else:
            dmin, dmaj, Dm, tm, Lname = st.dn, st.dp, d.Dn, d.tau_eff_n, 'L_n'
        Lm = continuity.diffusion_length(Dm, tm) / UM2CM
        j = int(np.argmin(np.abs(x - (Lm if np.isfinite(Lm) else 0.5 * L))))
        rows += [row('L_decay', 'Minority excess 1/e decay length [µm]', decay_length(x, dmin), Lm,
                     f'{Lname} = (Dτ)^1/2'),
                 row('L_semilog', 'Minority excess semilog-slope length [µm]', semilog_length(x, dmin), Lm,
                     f'{Lname} = (Dτ)^1/2'),
                 row('maj_min_ratio', f'Majority/minority excess at {Lname} from the contact'
                     if np.isfinite(Lm) else 'Majority/minority excess at the center',
                     dmaj[j] / dmin[j] if dmin[j] != 0 else np.nan, 1.0, 'charge neutrality'),
                 row('injection_inj', 'Injection level δ(0)/N', inj / Ndop, '≪ 1', 'low injection, Sec. 6.3.2')]
        if np.isfinite(Lm):
            ov['excess'] += [(x, inj * np.exp(-x / Lm), f'δ(0) exp(−x/{Lname})', 'dot'),
                             (x, inj * continuity.finite_decay(x, L, Lm), 'bar with contacts (sinh)', 'dashdot')]
        else:
            ov['excess'] += [(x, inj * continuity.finite_decay(x, L, Lm), 'no recombination (linear)', 'dashdot')]

    return rows, ov


def _tail(st, carrier, window=(0.6, 0.2), x_max=0.8):
    """Tail of the excess carriers injected at the left contact: decay length λ [cm] from the slope of ln δ
    between window[0] and window[1] of δ(0), effective lifetime τ' = δ/R, current field E_J = J/σ [V/cm]
    and the median δ in the window."""
    d = st.dev
    y = st.dn if carrier == 'n' else st.dp
    m = (y < window[0] * y[0]) & (y > window[1] * y[0]) & (st.x < x_max * st.x[-1])
    if m.sum() < 5:
        return dict(lam=np.nan, tau=np.nan, EJ=np.nan, delta=np.nan)
    lam = -1.0 / np.polyfit(d.x[m], np.log(y[m]), 1)[0]
    sigma = Q * (d.mun * st.n[m] + d.mup * st.p[m])
    return dict(lam=float(lam), tau=float(np.median(y[m] / st.R[m])), EJ=float(np.median(st.J / sigma)),
                delta=float(np.median(y[m])))


def run_ambipolar_sweep(N_min=1e11, N_max=1e17, per_decade=1, inj=1e14, V=-1.5, length_um=300.0, material='Si',
                        mun=None, mup=None, taun_us=0.1, taup_us=0.1, T=300.0, ni300=None, npts=751,
                        window=(0.6, 0.2), drift_min=0.3, verbose=True, rows=None, cols=None,
                        doping_type='p'):
    """
    Ambipolar transport: the same injection δ(0) = inj into p-type bars with N_a = N_min ... N_max.
    Each bar is solved at V = 0 and at V; the tails give D' and mu' of the ambipolar transport equation
    D' δ'' + mu' E δ' - δ/τ' = 0, with δ ∝ exp(-s/λ):  D' - mu' E λ = λ²/τ'  (one equation per bias).

    input:
        float   N_min, N_max    Range of the acceptor doping [cm^-3]; per_decade = points per decade
        float   inj             Excess carriers at the left contact [cm^-3]
        float   V               Bias of the left contact for the drift run [V] (right contact grounded)
        tuple   window          Fit window of δ/δ(0) for the tails
        float   drift_min       Minimum relative change of λ²/τ' by the drift for reporting mu'
        (other parameters as in run_bar; SRH is on)
    output:
        SimulationResult        res, res.data: arrays per doping (see keys) and theory curves
    """
    if doping_type != 'p':
        raise ValueError("The ambipolar sweep uses p-type bars (minority electrons).")
    if not 0 < N_min < N_max or inj <= 0 or V == 0 or per_decade < 1:
        raise ValueError("Use 0 < N_min < N_max, inj > 0, V != 0 and per_decade >= 1.")
    say = print if verbose else (lambda *a, **k: None)
    nd = int(round(np.log10(N_max / N_min) * per_decade))
    if nd > 24:
        raise ValueError("Use at most 25 doping values (N_max/N_min and per_decade).")
    Ns = np.logspace(np.log10(N_min), np.log10(N_max), nd + 1)
    common = dict(doping_type='p', length_um=length_um, material=material, mun=mun, mup=mup, taun_us=taun_us,
                  taup_us=taup_us, T=T, ni300=ni300, npts=npts, inj=inj, srh=True, verbose=False)
    say('(0)-(1) Injection into p-type bars, solved at V = 0 and V = {:+.2f} V ({}) ...'.format(
        V, 'quasi-neutral model'))
    keys = ('lam0', 'lamE', 'tau0', 'tauE', 'EJ0', 'EJ', 'delta', 'D_sim', 'mu_sim', 'D_th', 'mu_th', 'tau_th',
            'lam0_th', 'lamE_th')
    data = {k: np.full(Ns.size, np.nan) for k in keys}
    data.update(N=Ns, iters0=np.zeros(Ns.size, int), itersE=np.zeros(Ns.size, int), profiles=[], states=[])
    for k, N in enumerate(Ns):
        r0 = run_bar(V=0.0, N=N, **common)
        r1 = run_bar(V=V, N=N, **common)
        s0, s1 = r0.data['state'], r1.data['state']
        dev = s0.dev
        say(f'\tNa = {N:.3g} cm^-3: V = 0 {"converged" if s0.converged else "NOT converged"} in {s0.iters}, '
            f'V = {V:+.2f} V {"converged" if s1.converged else "NOT converged"} in {s1.iters} iterations')
        t0, t1 = _tail(s0, 'n', window), _tail(s1, 'n', window)
        data['iters0'][k], data['itersE'][k] = s0.iters, s1.iters
        for name, t in (('0', t0), ('E', t1)):
            data['lam' + name][k], data['tau' + name][k] = t['lam'] / UM2CM, t['tau'] / US2S
        data['EJ0'][k], data['EJ'][k], data['delta'][k] = t0['EJ'], t1['EJ'], t0['delta']
        # D' - mu' E_J λ = λ²/τ' at both biases. mu' is reported only if the drift changes λ²/τ' by more than
        # drift_min (relative); a smaller change is within the accuracy of the linear ambipolar equation.
        A = np.array([[1.0, -t0['EJ'] * t0['lam']], [1.0, -t1['EJ'] * t1['lam']]])
        b = np.array([t0['lam'] ** 2 / t0['tau'], t1['lam'] ** 2 / t1['tau']])
        if np.all(np.isfinite(A)) and np.all(np.isfinite(b)) and abs(b[1] - b[0]) > drift_min * b[0]:
            data['D_sim'][k], data['mu_sim'][k] = np.linalg.solve(A, b)
        else:
            data['D_sim'][k] = b[0]                               # No current at V = 0: D' = λ²/τ'
        # Theory at the densities of the fit window
        n_eq, p_eq = dev.ni ** 2 / N, N
        for name, t in (('0', t0), ('E', t1)):
            n, p = n_eq + t['delta'], p_eq + t['delta']
            D_th = continuity.D_ambipolar(n, p, dev.Dn, dev.Dp)
            mu_th = continuity.mu_ambipolar(n, p, dev.mun, dev.mup)
            tau_th = t['delta'] / ((n * p - dev.ni ** 2) / (dev.taup * (n + dev.ni) + dev.taun * (p + dev.ni)))
            a = mu_th * t['EJ'] * tau_th
            lam_th = (np.sqrt(a * a + 4 * D_th * tau_th) - a) / 2
            data['lam' + name + '_th'][k] = lam_th / UM2CM
            if name == '0':
                data['D_th'][k], data['mu_th'][k], data['tau_th'][k] = D_th, mu_th, tau_th / US2S
        data['profiles'].append((s0.x, s0.dn / s0.dn[0], s1.dn / s1.dn[0]))
        data['states'] += [s0, s1]
    say('Done!!')
    # Theory curves at the typical density of the fit window
    Nc = np.logspace(np.log10(N_min) - 0.5, np.log10(N_max) + 0.5, 200)
    dref = np.sqrt(window[0] * window[1]) * inj
    nc, pc = dev.ni ** 2 / Nc + dref, Nc + dref
    data['curve'] = dict(N=Nc, D=continuity.D_ambipolar(nc, pc, dev.Dn, dev.Dp),
                         mu=continuity.mu_ambipolar(nc, pc, dev.mun, dev.mup), delta=dref)
    data['limits'] = dict(D_low=dev.Dn, D_high=2 * dev.Dn * dev.Dp / (dev.Dn + dev.Dp), mu_low=dev.mun)
    data['V'] = V
    D, i0, i1 = data, 0, Ns.size - 1
    table = [row('D_lowN', f"D' at N_a = {Ns[i0]:.0e} [cm²/s]", D['D_sim'][i0], D['D_th'][i0], "(6.43), δ ≫ N_a"),
             row('D_highN', f"D' at N_a = {Ns[i1]:.0e} [cm²/s]", D['D_sim'][i1], D['D_th'][i1], "(6.43), δ ≪ N_a"),
             row('mu_highN', f"μ' at N_a = {Ns[i1]:.0e} [cm²/V s]", D['mu_sim'][i1], D['mu_th'][i1], 'Sec. 6.3'),
             row('tau_lowN', f"τ'/τ at N_a = {Ns[i0]:.0e}", D['tau0'][i0] / taun_us, D['tau_th'][i0] / taun_us,
                 'SRH, high injection → 2'),
             row('D_low', "Low-injection limit D' = D_n [cm²/s]", None, dev.Dn, '(6.45)'),
             row('D_high', "High-injection limit D' = 2D_nD_p/(D_n + D_p) [cm²/s]", None,
                 data['limits']['D_high'], '(6.43), n = p'),
             row('mu_low', "Low-injection limit μ' = μ_n [cm²/V s]", None, dev.mun, "μ' → 0 for n = p"),
             row('L_min', 'Minority diffusion length L_n = (D_nτ_n)^1/2 [µm]', None,
                 continuity.diffusion_length(dev.Dn, dev.taun) / UM2CM, 'low injection, V = 0')]
    params = dict(analysis='ambipolar', N_min=N_min, N_max=N_max, per_decade=per_decade, inj=inj, V=V,
                  length_um=length_um, material=material, mun=mun, mup=mup, taun_us=taun_us, taup_us=taup_us,
                  T=T, ni300=ni300, npts=npts, window=window, drift_min=drift_min, structure=0, doping_type='p')
    res = SimulationResult(params, dev, data['states'], table, data)
    if verbose:
        print('\n(2) Comparison with analytic models')
        print_table(select_rows(res.table, rows))
        print()
        print_ambipolar_table(res)
    return res


def print_ambipolar_table(res):
    """Per-doping results of run_ambipolar_sweep."""
    d = res.data
    head = ['Na [cm⁻³]', 'δ/Na', 'λ(V=0) [µm]', "λ(V) [µm]", 'λ(V) theory', 'EJ [V/cm]', "τ'/τ",
            "D' sim", "D' (6.43)", "μ' sim", "μ' theory"]
    fmt = ['{:.2e}', '{:.3g}', '{:.4g}', '{:.4g}', '{:.4g}', '{:+.3g}', '{:.3f}', '{:.4g}', '{:.4g}', '{:.4g}',
           '{:.4g}']
    tau = res.params['taun_us']
    cols = [d['N'], d['delta'] / d['N'], d['lam0'], d['lamE'], d['lamE_th'], d['EJ'], d['tau0'] / tau, d['D_sim'],
            d['D_th'], d['mu_sim'], d['mu_th']]
    print(f"Ambipolar sweep, V = {d['V']:+.2f} V (λ: tail length, EJ = J/σ in the tail; —: drift changes the tail "
          f"too little to resolve μ')")
    print('  '.join(f'{h:>11s}' for h in head))
    for k in range(d['N'].size):
        print('  '.join(f'{f.format(c[k]):>11s}' if np.isfinite(c[k]) else f"{'—':>11s}" for f, c in zip(fmt, cols)))
```

```python
###############################################################################
# Lab 5 Runners
###############################################################################

def run_bar(doping_type='n', N=1e16, length_um=100.0, material='Si', T=300.0, ni300=None, mun=None, mup=None,
            taun_us=0.1, taup_us=0.1, srh=True, V=0.0, G=0.0, gen_depth_um=0.0, inj=0.0, bc_left='ohmic',
            bc_right='ohmic', npts=None, maxit=None, verbose=True, rows=None):
    """
    Excess carriers in a uniformly doped bar.

    input:
        float   G, gen_depth_um     Generation rate [cm^-3 s^-1]: uniform, or G exp(-s/d) with d = gen_depth_um [um]
        float   inj                 Excess carriers at the left contact [cm^-3]
        float   V                   Bias of the left contact [V] (right contact grounded)
        str     bc_left, bc_right   'ohmic' contact or 'free' end (no contact, no surface recombination)
        bool    srh                 SRH recombination on (True) or off (False, R = 0)
    output:
        SimulationResult            res (res.data['state'], res.data['overlays'])
    """
    params = dict(structure=0, doping_type=doping_type, N=N, length_um=length_um, material=material, T=T,
                  ni300=ni300, mun=mun, mup=mup, taun_us=taun_us, taup_us=taup_us, srh=srh, V=V, G=G,
                  gen_depth_um=gen_depth_um, inj=inj, bc_left=bc_left, bc_right=bc_right, npts=npts, analysis='point')
    if G < 0 or inj < 0 or gen_depth_um < 0:
        raise ValueError("Use G >= 0, inj >= 0 and gen_depth_um >= 0.")
    free = [side for side, bc in (('left', bc_left), ('right', bc_right)) if bc == 'free']
    if free and V != 0:
        raise ValueError("A free end carries no current: use V = 0, or ohmic contacts at both ends.")
    if inj > 0 and bc_left != 'ohmic':
        raise ValueError("Injection needs an ohmic left contact (bc_left='ohmic').")
    if len(free) == 2 and not srh and G > 0:
        raise ValueError("Free ends without recombination: nothing removes the generated carriers, so there is no "
                         "steady state. Turn SRH on or use ohmic contacts.")
    say = print if verbose else (lambda *a, **k: None)
    dev = build_bar(doping_type, N, length_um, material, T, ni300, mun, mup, taun_us, taup_us, srh, bc_left, bc_right,
                    npts)
    G_x = continuity.generation_profile(dev.x_um, G, gen_depth_um) if gen_depth_um > 0 else G
    print_device(dev, 'Uniform bar', say)
    if srh:
        say(f'τn, τp \t = {taun_us:g}, {taup_us:g} us  (Ln, Lp = '
            f'{continuity.diffusion_length(dev.Dn, dev.taun) / UM2CM:.4g}, '
            f'{continuity.diffusion_length(dev.Dp, dev.taup) / UM2CM:.4g} µm)')
    else:
        say('recombination \t = SRH off (R = 0)')
    if free:
        say('ends \t\t = ' + ('no contacts (infinite sample: free ends, no surface recombination)' if len(free) == 2
                               else f"{free[0]} end free (no surface recombination), "
                                    f"{'right' if free[0] == 'left' else 'left'} end ohmic"))
    if G > 0:
        say(f'generation \t = {G:.3g} cm^-3 s^-1' + (f' x exp(-s/d), d = {gen_depth_um:g} µm' if gen_depth_um > 0
                                                      else ' (uniform)'))
    say('model \t\t = quasi-neutral, δn = δp (Neamen Sec. 6.3)')
    say('\n(1) Solving ...')
    s = ExcessSolver(dev)
    st = s.solve(V, G=G_x, inj=inj, maxit=maxit)
    say(f'\tV = {V:+.3f} V    : {"converged" if st.converged else "NOT converged"} in {st.iters} iterations, '
        f'J = {st.J:+.4e} A/cm^2')
    iters = row('iters', 'Iterations', st.iters, None,
                'converged' if st.converged else 'NOT converged (increase SIM_CFG.maxit)')
    table, ov = analyze_excess(st, s, params)
    res = SimulationResult(params, dev, [st], [iters] + table, dict(state=st, overlays=ov))
    say('Done!!')
    if verbose:
        print('\n(2) Comparison with analytic models')
        print_table(select_rows(res.table, rows))
    return res


###############################################################################
# Lab 5 Cases
###############################################################################

def case_kwargs(p):
    """Translate case parameters into runner arguments (ends, lifetimes, automatic grid)."""
    kw = dict(p)
    if 'ends' in kw:                                          # 'free' (no contacts) or 'ohmic' at both ends
        kw['bc_left'] = kw['bc_right'] = kw.pop('ends')
    if 'tau_us' in kw:                                        # tau_n = tau_p
        kw['taun_us'] = kw['taup_us'] = kw.pop('tau_us')
    if kw.get('npts') == 0:
        kw['npts'] = None                                     # Automatic grid
    return kw


_MU = dict(mun=1350.0, mup=480.0)

CASES = {
    '5-A': dict(title='Uniform generation', runner=run_bar, fixed=dict(V=0.0),
                summary='No contacts (infinite sample): δ = Gτ everywhere. Ohmic contacts: sinks, δ → 0 at both ends.',
                params=dict(doping_type='n', N=1e15, ni300=1e10, tau_us=0.1, G=1e20, ends='free', length_um=100.0,
                            **_MU, srh=True, T=300.0, npts=1001),
                rows=['dp_center', 'dn_center', 'dp_srh', 'dp_ends', 'dp_bar', 'R_center', 'EF_Ei', 'EFn_Ei',
                      'Ei_EFp', 'injection_G'],
                panels=['excess', 'band', 'rate', 'density'],
                schematic=lambda v: schematic_uniform_generation(v.get('doping_type', 'n'), v.get('ends', 'free'))),
    '5-B': dict(title='Surface absorption', runner=run_bar, fixed=dict(V=0.0, bc_left='free', bc_right='ohmic'),
                summary='Light absorbed near the left surface, G = G₀e^{−s/d}: carriers diffuse inward and decay with '
                        'L = (Dτ)^1/2.',
                params=dict(doping_type='p', N=1e16, tau_us=0.1, G=2e21, gen_depth_um=1.0, length_um=100.0, **_MU,
                            srh=True, T=300.0, npts=2001),
                ranges=dict(gen_depth_um=(0.1, 10.0, 0.1)),
                rows=['d0', 'd0_semi', 'L_tail', 'R_total', 'injection_G'],
                panels=['excess', 'rate', 'band', 'density'],
                schematic=lambda v: schematic_absorption(v.get('doping_type', 'p'))),
    '5-C': dict(title='Ambipolar transport vs doping', runner=run_ambipolar_sweep, fixed=dict(doping_type='p'),
                summary="Same injection into p-type bars from Na = Nmin to Nmax: D′ and μ′ change from the "
                        "high-injection values to Dn and μn.",
                params=dict(N_min=1e11, N_max=1e17, inj=1e14, V=-1.5, tau_us=0.1, length_um=300.0, **_MU,
                            T=300.0, npts=751),
                ranges=dict(V=(-2.0, 2.0, 0.1), length_um=(100.0, 500.0, 10.0)),
                rows=['D_lowN', 'D_highN', 'mu_highN', 'tau_lowN', 'D_low', 'D_high', 'mu_low', 'L_min'],
                panels=None, schematic=lambda v: schematic_ambipolar()),
}
```

```python
import contextlib
import html
import io
import time
import traceback

import ipywidgets as widgets
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from IPython.display import HTML, display

###############################################################################
# Plot Settings and Figure Builder
###############################################################################

C_E, C_H, C_T, C_A = '#1f77b4', '#d62728', '#222222', '#7f7f7f'   # Electrons, holes, total, analytic
FIG_WIDTH = 1150
PANEL_HEIGHT = 250         # Height of one panel [px]
ROW_GAP = 105              # Vertical gap between panel rows (axis title + panel title) [px]
NMAX_PLOT = 600            # Maximum points per curve sent to the browser
NMAX_FRAME = 300           # Maximum points per curve in each bias-sweep frame


def _sub(s):
    """Subscripts in Plotly text: 'E_g', 'μ_n', 'N_{a}' -> 'E<sub>g</sub>', 'μ<sub>n</sub>', 'N<sub>a</sub>'."""
    if not isinstance(s, str):
        return s
    return re.sub(r'(?<=[^\W\d_])_(\{[^}]*\}|[A-Za-z0-9]+)', lambda m: f"<sub>{m.group(1).strip('{}')}</sub>", s)


def _trace(x, y, name, color=None, dash='solid', mode='lines', width=2):
    """Trace specification."""
    return dict(x=np.asarray(x, dtype=float), y=np.asarray(y, dtype=float), name=name, color=color,
                dash=dash, mode=mode, width=width)


def _panel(title, xlabel, ylabel, traces, ylog=False, xrange=None, yrange=None, vlines=(), xlog=False, decades=None):
    """Panel specification; decades limits a log axis to that many decades below the maximum."""
    return dict(title=title, xlabel=xlabel, ylabel=ylabel, traces=traces, ylog=ylog,
                xrange=xrange, yrange=yrange, vlines=list(vlines), xlog=xlog, decades=decades)


def _thin(x, y, ylog, nmax=NMAX_PLOT):
    """Reduce points along the arc length of the curve, keeping steep regions dense."""
    if len(x) <= nmax:
        return x, y
    yy = np.log10(np.where(y > 0, y, np.nan)) if ylog else y.astype(float)
    ok = np.isfinite(yy)
    if ok.sum() < 2:
        idx = np.linspace(0, len(x) - 1, nmax).astype(int)
        return x[idx], y[idx]
    yy = np.where(ok, yy, np.nanmin(yy[ok]))
    xs = (x - x.min()) / max(np.ptp(x), 1e-30)
    ys = (yy - yy.min()) / max(np.ptp(yy), 1e-30)
    s = np.concatenate([[0.0], np.cumsum(np.hypot(np.diff(xs), np.diff(ys)))])
    idx = np.unique(np.concatenate([np.searchsorted(s, np.linspace(0, s[-1], nmax)),
                                    np.linspace(0, len(x) - 1, nmax // 4).astype(int), [len(x) - 1]]))
    idx = idx[idx < len(x)]
    return x[idx], y[idx]


def _scatter(t, ylog, legend, nmax=NMAX_PLOT, xrange=None):
    ylog = ylog or t.get('thin_log', False)
    x, y = t['x'], (np.where(t['y'] > 0, t['y'], np.nan) if ylog else t['y'])
    if xrange is not None and len(x) > 2:
        # Keep the zoomed window (plus one point on each side) before thinning.
        inside = np.flatnonzero((x >= xrange[0]) & (x <= xrange[1]))
        if inside.size:
            sl = slice(max(inside[0] - 1, 0), min(inside[-1] + 2, len(x)))
            x, y = x[sl], y[sl]
    x, y = _thin(x, y, ylog, t.get('nmax', nmax))
    extra = {k: t[k] for k in ('visible', 'legendgroup', 'showlegend', 'fill', 'fillcolor', 'opacity') if k in t}
    line = dict(color=t['color'], dash=t['dash'], width=t['width'])
    if 'shape' in t:
        line['shape'] = t['shape']
    return go.Scatter(x=x.astype(np.float32), y=y.astype(np.float32), name=_sub(t['name']), mode=t['mode'], line=line,
                      marker=dict(size=t.get('marker_size', 7), color=t['color']), legend=legend, **extra)


def _log_range(values, decades=30):
    v = np.concatenate([np.asarray(a, dtype=float).ravel() for a in values])
    v = v[np.isfinite(v) & (v > 0)]
    if v.size == 0:
        return None
    hi = np.log10(v.max())
    lo = max(np.log10(v.min()), hi - decades)
    if hi - lo < 1.0:                                   # Nearly constant: two decades around the values
        c = 0.5 * (hi + lo)
        return [c - 1.0, c + 1.0]
    pad = 0.05 * max(hi - lo, 1.0)
    return [lo - pad, hi + pad]


def _lin_range(values, symmetric=False):
    v = np.concatenate([np.asarray(a, dtype=float).ravel() for a in values])
    v = v[np.isfinite(v)]
    if v.size == 0:
        return None
    lo, hi = (-np.max(np.abs(v)), np.max(np.abs(v))) if symmetric else (v.min(), v.max())
    if hi - lo <= 1e-6 * max(abs(lo), abs(hi)):          # Nearly constant: include zero
        lo, hi = min(lo, 0.0), max(hi, 0.0)
        if hi == lo:
            return [-1.0, 1.0]
    pad = 0.05 * (hi - lo)
    return [lo - pad, hi + pad]


def build_figure(panels, title, ncols=2, bottom=60):
    """
    Build one Plotly figure with one legend per panel (placed to the right of the panel).

    output:
        go.Figure   fig
        list        trace_ids   Trace indices of each panel
    """
    nrows = (len(panels) + ncols - 1) // ncols
    plot_height = nrows * PANEL_HEIGHT + (nrows - 1) * ROW_GAP
    fig = make_subplots(rows=nrows, cols=ncols, subplot_titles=[_sub(pn['title']) if pn else ' ' for pn in panels],
                        horizontal_spacing=0.24, vertical_spacing=ROW_GAP / plot_height if nrows > 1 else 0.0)
    trace_ids = []
    for k, pn in enumerate(panels):
        r, c = k // ncols + 1, k % ncols + 1
        if pn is None:                              # Blank position (keeps the following panels aligned)
            fig.update_xaxes(visible=False, row=r, col=c)
            fig.update_yaxes(visible=False, row=r, col=c)
            trace_ids.append([])
            continue
        legend = 'legend' if k == 0 else f'legend{k + 1}'
        ids = []
        for t in pn['traces']:
            fig.add_trace(_scatter(t, pn['ylog'], legend, xrange=pn['xrange']), row=r, col=c)
            ids.append(len(fig.data) - 1)
        trace_ids.append(ids)
        sub = fig.get_subplot(r, c)
        fig.update_layout({legend: dict(x=sub.xaxis.domain[1] + 0.01, y=sub.yaxis.domain[1], xanchor='left',
                                        yanchor='top', font=dict(size=10), bgcolor='rgba(255,255,255,0.85)')})
        yrange = pn['yrange']
        if yrange is None:                          # Limit log axes; avoid zoomed-in axes on constant curves
            vals = [t['y'] for t in pn['traces']]
            if pn['ylog']:
                auto = _log_range(vals, decades=pn.get('decades') or 30)
                if auto is not None and (pn.get('decades') or auto[1] - auto[0] <= 2.0):
                    yrange = auto
            elif vals:
                auto = _lin_range(vals)
                v = np.concatenate([np.asarray(a, dtype=float).ravel() for a in vals])
                v = v[np.isfinite(v)]
                if v.size and np.ptp(v) <= 1e-6 * np.max(np.abs(v)):
                    yrange = auto
        fig.update_xaxes(title_text=_sub(pn['xlabel']), range=pn['xrange'], row=r, col=c)
        fig.update_yaxes(title_text=_sub(pn['ylabel']), range=yrange, row=r, col=c)
        if 'cm⁻³' in pn['ylabel'] or 'A/cm²' in pn['ylabel']:   # 1e16 -> 10^16 instead of SI prefixes (T, p)
            fig.update_yaxes(exponentformat='power', row=r, col=c)
        if pn['ylog']:
            fig.update_yaxes(type='log', exponentformat='power', row=r, col=c)
        if pn.get('xlog'):
            fig.update_xaxes(type='log', exponentformat='power', row=r, col=c)
        for xv in pn['vlines']:
            fig.add_vline(x=xv, line=dict(color='gray', dash='dot', width=1), row=r, col=c)
    fig.update_layout(width=FIG_WIDTH, height=plot_height + 90 + bottom, template='plotly_white', title=_sub(title),
                      margin=dict(l=80, r=200, t=90, b=bottom))
    fig.update_annotations(font_size=13)
    return fig, trace_ids


FIGURE_SINK = None          # Set by the launcher to collect figures instead of displaying them


def show_figure(fig):
    if FIGURE_SINK is not None:
        FIGURE_SINK(fig)
    else:
        display(HTML(fig.to_html(full_html=False, include_plotlyjs='inline', auto_play=False)))


PLOTLY_JS_SENT = {'done': False}     # The launcher embeds plotly.js in its first figure only


def figure_light_html(fig):
    """Figure HTML without plotly.js: draws with the plotly.js already in the page, or loads it from the
    plotly CDN if the page does not have it (for example after reloading the window)."""
    html_fig = fig.to_html(full_html=False, include_plotlyjs=False, auto_play=False)
    cdn = re.search(r'src="([^"]+)"', go.Figure().to_html(full_html=False, include_plotlyjs='cdn')).group(1)
    i0 = html_fig.index('<script>') + len('<script>')
    i1 = html_fig.rindex('</script>')
    wrapped = ("(function () { var draw = function () {" + html_fig[i0:i1] + "};"
               " if (window.Plotly) { draw(); return; }"
               " var s = document.createElement('script'); s.src = '" + cdn + "'; s.onload = draw;"
               " document.head.appendChild(s); })();")
    return html_fig[:i0] + wrapped + html_fig[i1:]


def launcher_figure_html(fig):
    """
    Figure HTML for the launcher: (html shown now, html kept for showing again).
    The first figure after launch() embeds plotly.js (about 4.7 MB); later figures carry only their data.
    """
    light = figure_light_html(fig)
    if not PLOTLY_JS_SENT['done']:
        PLOTLY_JS_SENT['done'] = True
        return fig.to_html(full_html=False, include_plotlyjs='inline', auto_play=False), light
    return light, light


def _overlay_traces(ov, key, x0=0.0):
    """Analytic overlays (computed on the device grid) shifted to the plotting origin x0."""
    return [_trace(np.asarray(x) - x0, y, name, C_A, dash, width=1.5) for x, y, name, dash in ov.get(key, [])]


def _rgba(hex_color, alpha):
    h = hex_color.lstrip('#')
    return f'rgba({int(h[0:2], 16)},{int(h[2:4], 16)},{int(h[4:6], 16)},{alpha})'


###############################################################################
# Device Schematics (inline SVG shown next to the case parameters)
###############################################################################

# Drawings with symbols only: the left contact is the bias terminal V, the right contact is grounded.

SCH_W, SCH_H = 480, 225
SCH_FONT = "font-family='Helvetica, Arial, sans-serif'"
SCH_COLORS = dict(p='#d6e4fb', n='#fbd5d5', p_dark='#3b6fd8', n_dark='#d24b4b', edge='#555555', wire='#333333',
                  contact='#444444', light='#e08a00', text='#222222', note='#666666')     # p: blue, n: red
_X0, _X1, _Y0, _Y1 = 110, 370, 95, 140          # Device block [px]


def _markup(s, size):
    """'N_{a}' and 'e^{-s/λ}' -> SVG tspans with subscripts and superscripts."""
    small = round(0.72 * size)
    s = re.sub(r'_\{([^}]*)\}', lambda m: f"<tspan baseline-shift='sub' font-size='{small}'>{m.group(1)}</tspan>", s)
    return re.sub(r'\^\{([^}]*)\}', lambda m: f"<tspan baseline-shift='super' font-size='{small}'>{m.group(1)}</tspan>", s)


def _text(x, y, s, size=13, anchor='middle', color=None, weight='normal', italic=False, bg=False):
    color = color or SCH_COLORS['text']
    style = " font-style='italic'" if italic else ''
    out = ''
    if bg:                                       # White box behind the label (hides lines underneath)
        w = 0.62 * size * len(re.sub(r'[_^]\{([^}]*)\}', r'\1', s)) + 6
        x_left = x - w / 2 if anchor == 'middle' else (x - 3 if anchor == 'start' else x - w + 3)
        out += f"<rect x='{x_left:.1f}' y='{y - size + 1:.1f}' width='{w:.1f}' height='{size + 4}' fill='white'/>"
    return out + (f"<text x='{x}' y='{y}' text-anchor='{anchor}' font-size='{size}' fill='{color}' "
                  f"font-weight='{weight}'{style} {SCH_FONT}>{_markup(s, size)}</text>")


def _line(x0, y0, x1, y1, color=None, width=1.5, dash=None):
    d = f" stroke-dasharray='{dash}'" if dash else ''
    return (f"<line x1='{x0}' y1='{y0}' x2='{x1}' y2='{y1}' stroke='{color or SCH_COLORS['wire']}' "
            f"stroke-width='{width}'{d}/>")


def _arrowhead(x, y, angle, size=7, color=None):
    """Filled arrowhead with its tip at (x, y), pointing along angle [rad]."""
    c, s = np.cos(angle), np.sin(angle)
    pts = [(x, y), (x - size * c + 0.5 * size * s, y - size * s - 0.5 * size * c),
           (x - size * c - 0.5 * size * s, y - size * s + 0.5 * size * c)]
    return (f"<polygon points='{' '.join(f'{px:.1f},{py:.1f}' for px, py in pts)}' "
            f"fill='{color or SCH_COLORS['wire']}'/>")


def _dimension(x0, x1, y, label):
    """Double-headed length arrow with a label underneath."""
    return (_line(x0 + 1, y, x1 - 1, y, width=1) + _arrowhead(x0, y, np.pi) + _arrowhead(x1, y, 0.0)
            + _line(x0, y - 6, x0, y + 6, width=1) + _line(x1, y - 6, x1, y + 6, width=1)
            + _text((x0 + x1) / 2, y + 18, label, bg=True))


def _wavy_arrow(x0, y0, x1, y1, color=None, waves=3, amp=4.0):
    """Light ray: sine-wave path from (x0, y0) to (x1, y1) with an arrowhead."""
    color = color or SCH_COLORS['light']
    L = np.hypot(x1 - x0, y1 - y0)
    ux, uy = (x1 - x0) / L, (y1 - y0) / L
    t = np.linspace(0, L - 8, 60)
    off = amp * np.sin(2 * np.pi * waves * t / (L - 8))
    xs, ys = x0 + ux * t - uy * off, y0 + uy * t + ux * off
    path = 'M ' + ' L '.join(f'{a:.1f},{b:.1f}' for a, b in zip(xs, ys))
    return (f"<path d='{path}' fill='none' stroke='{color}' stroke-width='1.8'/>"
            + _arrowhead(x1, y1, np.arctan2(uy, ux), size=8, color=color))


def _ground(x, y):
    return (_line(x, y, x + 16, y) + _line(x + 16, y, x + 16, y + 26)
            + _line(x + 4, y + 26, x + 28, y + 26, width=2) + _line(x + 8, y + 31, x + 24, y + 31, width=2)
            + _line(x + 12, y + 36, x + 20, y + 36, width=2) + _text(x + 16, y + 52, 'ground', size=11,
                                                                       color=SCH_COLORS['note']))


def _contacts():
    """Thin ohmic-contact plates at both ends of the device."""
    return (f"<rect x='{_X0 - 6}' y='{_Y0 - 4}' width='6' height='{_Y1 - _Y0 + 8}' fill='{SCH_COLORS['contact']}'/>"
            f"<rect x='{_X1}' y='{_Y0 - 4}' width='6' height='{_Y1 - _Y0 + 8}' fill='{SCH_COLORS['contact']}'/>")


def _circuit():
    """Bias terminal V on the left contact and ground on the right contact."""
    ym = (_Y0 + _Y1) / 2
    return (_line(46, ym, _X0 - 6, ym)
            + f"<circle cx='40' cy='{ym}' r='5.5' fill='white' stroke='{SCH_COLORS['wire']}' stroke-width='1.5'/>"
            + _text(40, ym - 14, 'V', weight='bold') + _ground(_X1 + 6, ym))


def _block(x0, x1, fill, label_lines, y0=_Y0, y1=_Y1):
    out = f"<rect x='{x0}' y='{y0}' width='{x1 - x0}' height='{y1 - y0}' fill='{fill}' stroke='{SCH_COLORS['edge']}'/>"
    yc = (y0 + y1) / 2
    for k, s in enumerate(label_lines):
        out += _text((x0 + x1) / 2, yc + 5 + (k - (len(label_lines) - 1) / 2) * 16, s,
                     weight='bold' if k == 0 else 'normal')
    return out


def _origin(x, label='x = 0', through=False):
    """Dashed mark of x = 0; through=True crosses the device (junction), otherwise it starts below the block."""
    y_top = _Y0 - 8 if through else _Y1
    return _line(x, y_top, x, 196, color='#888888', width=1, dash='4,3') + _text(x, 210, label, size=12, bg=True)


def _svg(body, gradient='', top=0):
    """White card with the drawing, shifted up by 'top' px when nothing is drawn above the device."""
    h = SCH_H - top
    return (f"<svg xmlns='http://www.w3.org/2000/svg' width='{SCH_W}' height='{h}' viewBox='0 0 {SCH_W} {h}'>"
            f"<defs>{gradient}</defs>"
            f"<rect x='0.5' y='0.5' width='{SCH_W - 1}' height='{h - 1}' rx='8' fill='white' stroke='#d0d0d0'/>"
            f"<g transform='translate(0,{-top})'>{body}</g></svg>")


def _bar(dtype, label2='N'):
    """Uniformly doped bar: n-type red, p-type blue."""
    return _block(_X0, _X1, SCH_COLORS[dtype], [f'{dtype}-type', label2])


def _pn(biased):
    xm = (_X0 + _X1) / 2
    body = (_block(_X0, xm, SCH_COLORS['p'], ['p-type', 'N_{a}']) + _block(xm, _X1, SCH_COLORS['n'], ['n-type', 'N_{d}'])
            + _contacts() + (_circuit() if biased else '') + _origin(xm, through=True)
            + _dimension(_X0, xm, 160, 'L_{p}') + _dimension(xm, _X1, 160, 'L_{n}'))
    return _svg(body, top=45)


def _glow(gid, width_frac=0.35):
    """Orange shading that fades from the left end of the bar (excess carriers near the left end)."""
    grad = (f"<linearGradient id='{gid}' x1='0' x2='1' y1='0' y2='0'>"
            f"<stop offset='0' stop-color='{SCH_COLORS['light']}' stop-opacity='0.75'/>"
            f"<stop offset='1' stop-color='{SCH_COLORS['light']}' stop-opacity='0'/></linearGradient>")
    rect = (f"<rect x='{_X0}' y='{_Y0}' width='{width_frac * (_X1 - _X0):.0f}' height='{_Y1 - _Y0}' "
            f"fill='url(#{gid})'/>")
    return grad, rect


def _open_bar(dtype, label2='N'):
    """Bar cut at both ends (the sample continues: no contacts)."""
    fill, edge = SCH_COLORS[dtype], SCH_COLORS['edge']
    zig = lambda x, s: ' '.join(f'{x + s * (5 if k % 2 else 0)},{_Y0 + k * (_Y1 - _Y0) / 6:.1f}' for k in range(7))
    body = (f"<rect x='{_X0}' y='{_Y0}' width='{_X1 - _X0}' height='{_Y1 - _Y0}' fill='{fill}'/>"
            + _line(_X0, _Y0, _X1, _Y0, edge, 1) + _line(_X0, _Y1, _X1, _Y1, edge, 1)
            + f"<polyline points='{zig(_X0, -1)}' fill='none' stroke='{edge}' stroke-width='1.2'/>"
            + f"<polyline points='{zig(_X1, 1)}' fill='none' stroke='{edge}' stroke-width='1.2'/>")
    ym = (_Y0 + _Y1) / 2 + 5
    body += (_text(_X0 - 26, ym, '⋯', size=18, color=SCH_COLORS['note'])
             + _text(_X1 + 26, ym, '⋯', size=18, color=SCH_COLORS['note']))
    yc = (_Y0 + _Y1) / 2
    for k, s in enumerate([f'{dtype}-type', label2]):
        body += _text((_X0 + _X1) / 2, yc + 5 + (k - 0.5) * 16, s, weight='bold' if k == 0 else 'normal')
    return body


_XM = (_X0 + _X1) / 2


###############################################################################
# Lab 5 Panels and Figures
###############################################################################

def band_panel(st, title=None):
    title = title or f'Energy band diagram (E_g = {st.dev.Eg:g} eV)'
    return _panel(title, 'x [µm]', 'Energy [eV]', [
        _trace(st.xc, st.Ec, 'E_c', C_T), _trace(st.xc, st.Ev, 'E_v', C_T),
        _trace(st.xc, st.Ei, 'E_Fi', C_A, 'dot', width=1.5),
        _trace(st.xc, st.Fn, 'E_Fn', C_E, 'dash'), _trace(st.xc, st.Fp, 'E_Fp', C_H, 'dash')])


def density_panel(st, title='Carrier and doping concentrations'):
    d = st.dev
    return _panel(title, 'x [µm]', 'Concentration [cm⁻³]', [
        _trace(st.xc, st.n, 'n', C_E), _trace(st.xc, st.p, 'p', C_H),
        _trace(st.xc, d.Nd, 'N_d', C_E, 'dot', width=1.5), _trace(st.xc, d.Na, 'N_a', C_H, 'dot', width=1.5)], ylog=True)


def excess_panel(st, ov, title='Excess carriers |δn|, |δp|'):
    # Hide values below the Gummel tolerance (numerical noise at or near equilibrium).
    ni = st.dev.ni
    mask = lambda v, eq: np.where(np.abs(v) > 10 * SIM_CFG.tol_carrier * (eq + ni), np.abs(v), np.nan)
    return _panel(title, 'x [µm]', '|δ| [cm⁻³]', [_trace(st.xc, mask(st.dn, st.n - st.dn), '|δn|', C_E),
                                             _trace(st.xc, mask(st.dp, st.p - st.dp), '|δp|', C_H)]
             + _overlay_traces(ov, 'excess', st.dev.x0_um), ylog=True, decades=8)


def rate_panel(st, title='Generation and recombination rates'):
    G = np.asarray(st.G)
    ylog = G.max() > 10 * max(G.min(), 1e-300)                 # Absorption profile: log axis
    return _panel(title, 'x [µm]', 'Rate [cm⁻³ s⁻¹]', [_trace(st.xc, st.R, 'R (SRH)', C_T),
                                                      _trace(st.xc, G, 'G', C_A, 'dash')], ylog=ylog, decades=6)


def _run_title(res):
    p, d = res.params, res.device
    title = f"[{res.case}] " if res.case else ''
    title += f"{STRUCTURES[p['structure']]} | {d.material.name}, T = {d.T:g} K"
    if p['structure'] == 3:
        title += f" | N_a = {p['Na']:.3g}, N_d = {p['Nd']:.3g} cm⁻³"
    return title


def plot_ambipolar(res, panels=None):
    """Ambipolar sweep: tail lengths, D' and mu' against the doping, and the normalized excess profiles."""
    from plotly.colors import sample_colorscale
    d, p = res.data, res.params
    N, Nc, lim = d['N'], d['curve']['N'], d['limits']
    span = np.array([Nc[0], Nc[-1]])
    hline = lambda v, name, dash: _trace(span, [v, v], name, C_A, dash, width=1.5)
    V = f"V = {d['V']:+.2g} V"
    colors = sample_colorscale('Viridis', list(np.linspace(0, 0.9, N.size)))
    prof = []
    for k, (x, y0, y1) in enumerate(d['profiles']):
        prof += [dict(_trace(x, y1, f'N_a = {N[k]:.0e}', colors[k]), legendgroup=f'N{k}'),
                 dict(_trace(x, y0, f'{N[k]:.0e}', colors[k], 'dot', width=1.5), legendgroup=f'N{k}', showlegend=False)]
    builders = {
        'lam': lambda: _panel('Tail length of δn (markers: simulation)', 'N_a [cm⁻³]', 'λ [µm]', [
            _trace(N, d['lam0'], 'sim, V = 0', C_T, mode='markers'),
            _trace(N, d['lamE'], f'sim, {V}', C_E, mode='markers'),
            _trace(N, d['lam0_th'], "ambipolar eq., V = 0", C_A, 'dash', width=1.5),
            _trace(N, d['lamE_th'], f"ambipolar eq., {V}", C_E, 'dot', width=1.5)], xlog=True),
        'D': lambda: _panel("Ambipolar diffusion coefficient D′", 'N_a [cm⁻³]', "D′ [cm²/s]", [
            _trace(N, d['D_sim'], 'sim: λ²/τ′ (V = 0)', C_T, mode='markers'),
            _trace(Nc, d['curve']['D'], f"(6.43) at δ = {d['curve']['delta']:.2g}", C_A),
            hline(lim['D_low'], 'D_n (low injection)', 'dash'),
            hline(lim['D_high'], '2D_{n}D_{p}/(D_{n}+D_{p}) (n = p)', 'dot')], xlog=True),
        'mu': lambda: _panel("Ambipolar mobility μ′", 'N_a [cm⁻³]', "μ′ [cm²/V s]", [
            _trace(N, d['mu_sim'], 'sim (V = 0 and V)', C_T, mode='markers'),
            _trace(Nc, d['curve']['mu'], "μ_{n}μ_{p}(p−n)/(μ_{n}n+μ_{p}p)", C_A),
            hline(lim['mu_low'], 'μ_n (low injection)', 'dash'), hline(0.0, '0 (n = p)', 'dot')], xlog=True),
        'profiles': lambda: _panel(f'δn/δn(0) (solid: {V}, dotted: V = 0)', 'Distance from the injecting contact [µm]',
                                   'δn/δn(0)', prof, ylog=True, yrange=[-3, 0.1])}
    keys = ['lam', 'D', 'mu', 'profiles'] if panels is None else panels
    title = (f"[{res.case}] " if res.case else '') + \
        (f"Ambipolar transport | p-type {res.device.material.name}, T = {res.device.T:g} K | "
         f"δ(0) = {p['inj']:.2g} cm⁻³, τ = {p['taun_us']:g} µs, L_bar = {p['length_um']:g} µm")
    fig, _ = build_figure([builders[k]() if k else None for k in keys], title)
    show_figure(fig)


def plot_operating_point(res, panels=None):
    st, ov = res.data['state'], res.data['overlays']
    builders = {'band': lambda: band_panel(st), 'density': lambda: density_panel(st),
                'excess': lambda: excess_panel(st, ov), 'rate': lambda: rate_panel(st)}
    keys = panels or ['excess', 'band', 'rate', 'density']
    fig, _ = build_figure([builders[k]() if k else None for k in keys], f'{_run_title(res)} | V = {st.V:+.3f} V')
    show_figure(fig)


def plot_result(res, panels=None):
    """Plot a result; panels = list of panel keys (default: the case's panels)."""
    panels = panels if panels is not None else res.panels
    print('\n(3) Visualizing ...')
    if res.params['analysis'] == 'ambipolar':
        print('\tMarkers = simulation, gray = ambipolar transport equation (Sec. 6.3); one color per doping')
        plot_ambipolar(res, panels)
    else:
        print('\tBlue = electrons, Red = holes, Black = total / simulation, Gray = analytic model')
        plot_operating_point(res, panels)
    print('Done!!')


###############################################################################
# Lab 5 Device Schematics
###############################################################################

def schematic_uniform_generation(dtype='n', ends='free'):
    """Uniform generation (light over the whole bar); ends = 'free' (infinite sample) or 'ohmic' (contacts)."""
    rays = ''.join(_wavy_arrow(x, 18, x, _Y0 - 8) for x in np.linspace(_X0 + 25, _X1 - 25, 5))
    light = (rays + _text(_X0 + 6, 30, 'hν', size=14, anchor='end', color=SCH_COLORS['light'], weight='bold')
             + _text(_X1 + 14, 52, 'G (uniform)', size=12, anchor='start'))
    if ends == 'free':
        return _svg(_open_bar(dtype) + _origin(_XM) + _text(_XM, 175, 'L → ∞', size=13, bg=True) + light)
    return _svg(_bar(dtype) + _contacts() + _origin(_XM) + _dimension(_X0, _X1, 160, 'L') + light)


def schematic_absorption(dtype='p'):
    """Light absorbed near the free left surface, G = G0 exp(-s/d); ohmic contact on the right."""
    grad, glow = _glow('absorb', 0.25)
    rays = ''.join(_wavy_arrow(28, y, _X0 - 4, y) for y in (_Y0 + 9, (_Y0 + _Y1) / 2, _Y1 - 9))
    body = (_bar(dtype) + glow
            + f"<rect x='{_X1}' y='{_Y0 - 4}' width='6' height='{_Y1 - _Y0 + 8}' fill='{SCH_COLORS['contact']}'/>"
            + rays + _text(52, _Y0 - 10, 'hν', size=14, color=SCH_COLORS['light'], weight='bold')
            + _text(_X0 + 4, _Y0 - 10, 'G = G_{0} e^{−s/d}', size=12, anchor='start')
            + _origin(_XM) + _dimension(_X0, _X1, 160, 'L'))
    return _svg(body, grad, top=45)


def schematic_ambipolar():
    """Injection delta(0) at the left contact of p-type bars with N_a = N_min ... N_max, biased on the left."""
    grad, glow = _glow('inject')
    body = (_bar('p', 'N_{a}: N_{min} … N_{max}') + glow + _contacts() + _circuit() + _origin(_XM)
            + _dimension(_X0, _X1, 160, 'L') + _text(_X0 + 6, _Y0 - 10, 'δ(0)', size=13, anchor='start'))
    return _svg(body, grad, top=45)
```

# (3) Lab 5 — Excess carriers: continuity and generation–recombination

Excess carriers created by generation or injection diffuse while they recombine. The continuity equations (Sec. 6.2.1) balance the divergence of each drift–diffusion current against generation and recombination [1, Ch. 6]:

$$\frac{1}{q}\frac{dJ_n}{dx}=R-G,\qquad \frac{1}{q}\frac{dJ_p}{dx}=G-R,\qquad R\approx\frac{\delta n}{\tau_n}\ \ (\text{p-type, low injection})$$

- Uniform generation in a sample without contacts gives the steady-state excess $\delta=G\tau$ everywhere and separates the quasi-Fermi levels $E_{Fn}$ and $E_{Fp}$ (6.85a), (6.85b).
- Light absorbed near a surface, $G(s)=G_0e^{-s/d}$, creates carriers within the absorption depth $d$; they diffuse into the bar, and beyond a few $d$ the excess decays with the diffusion length $L=\sqrt{D\tau}$.
- Excess electrons and holes move together (ambipolar transport). How they move, with $D'$ and $\mu'$, depends on the doping and on the injection level.
- **Recombination (SRH) = Off** sets $R=0$. Excess carriers then disappear only at ohmic contacts, and the profiles become straight lines or parabolas instead of exponentials.

### Ends of the bar: ohmic contacts and free surfaces

- **Ohmic contact:** the contact holds the equilibrium concentrations, so $\delta n=\delta p=0$; it is a perfect sink of excess carriers (infinite surface recombination velocity). Carriers generated within about one diffusion length of a contact diffuse into it instead of recombining in the bulk.
- **Free end:** no contact and no surface recombination. No carriers cross the surface ($J_n=J_p=0$ there) and the field outside is zero ($d\psi/dx=0$). Without a current path the bias must be $V=0$.

**Uniform generation (5-A).** With **Ends = No contacts** both ends are free, so the bar represents a piece of an infinite sample: $\delta$ is uniform and equal to the bulk value. The table compares it with $G\tau$ (low injection) and with the exact SRH solution $R_{SRH}(n_0+\delta,\,p_0+\delta)=G$,

$$\delta^2+\left[n_0+p_0-G(\tau_n+\tau_p)\right]\delta-G\left[\tau_p(n_0+n_i)+\tau_n(p_0+n_i)\right]=0,$$

which is about $\delta/N$ larger than $G\tau$ (1 % at the preset injection level $\delta/N=0.01$). With **Ends = Ohmic contacts** the excess falls to zero at both ends over about $L_p=11$ µm (holes in the n-type bar):

$$\delta(x)=G\tau\left[1-\frac{\cosh(x/L)}{\cosh(L_{bar}/2L)}\right]\qquad(x\text{ from the center}),$$

and only the center of a bar much longer than $2L$ reaches $G\tau$. With SRH off and no contacts nothing removes the generated carriers, so there is no steady state; the run stops with an error.

**Surface absorption (5-B).** Light enters through the free left surface and is absorbed as $G_0e^{-s/d}$ ($s$ from the surface); the right end is an ohmic contact. At low injection the minority excess obeys

$$D\frac{d^2\delta}{ds^2}-\frac{\delta}{\tau}=-G_0e^{-s/d},\qquad \frac{d\delta}{ds}\Big|_{s=0}=0,$$

with the semi-infinite solution

$$\delta(s)=\frac{G_0\tau d^2}{d^2-L^2}\left[e^{-s/d}-\frac{L}{d}\,e^{-s/L}\right],\qquad \delta(0)=\frac{G_0\tau d}{d+L}.$$

For $d\ll L$ the profile is $\delta(0)e^{-s/L}$ with $\delta(0)\approx G_0d\,\tau/L$: the absorbed flux $G_0d$ spreads over one diffusion length. For $d\gtrsim L$ the excess follows the absorption profile. The gray curve is the same equation solved with the contact at the right end (finite bar); the table also fits the decay length beyond $3d$ and compares the recombined fraction $\int R\,dx/\int G\,dx$ (the rest leaves through the contact).

### Ambipolar transport

The ambipolar transport equation is derived from the electron and hole continuity equations (6.29), (6.30) and the Poisson equation (6.32) under two assumptions: the internal field is small compared with the applied field (6.33), and $\delta n\approx\delta p$ (charge neutrality). The result is a single equation for the excess carriers [1, Sec. 6.3]:

$$D'\frac{\partial^2(\delta n)}{\partial x^2}+\mu'\mathcal{E}\frac{\partial(\delta n)}{\partial x}+g-R=\frac{\partial(\delta n)}{\partial t},\qquad D'=\frac{D_nD_p(n+p)}{D_nn+D_pp}\ \ (6.43),\qquad \mu'=\frac{\mu_n\mu_p(p-n)}{\mu_nn+\mu_pp}\ \ (6.41)$$

| p-type bar | $D'$ | $\mu'$ | $\tau'=\delta/R$ (SRH, $\tau_n=\tau_p=\tau$) |
| --- | --- | --- | --- |
| Low injection, $\delta\ll N_a$ | $D_n$ (6.45) | $\mu_n$ (6.46) | $\tau$ |
| High injection, $\delta\gg N_a$ ($n\approx p$) | $2D_nD_p/(D_n+D_p)$ | $0$ | $2\tau$ |

In low injection the excess carriers move like the minority carriers. In high injection they move as a neutral electron–hole cloud that diffuses with a $D'$ between $D_p$ and $D_n$ and does not drift.

**Case 5-C** injects the same $\delta(0)$ into p-type bars with $N_a=N_{min}\ldots N_{max}$ (one bar per decade), from high injection ($\delta\gg N_a$) to low injection ($\delta\ll N_a$). Each bar is solved at $V=0$ and at the bias $V$. In steady state ($g=0$) the tail is $\delta\propto e^{-s/\lambda}$ with

$$\lambda=\frac{\sqrt{(\mu'\mathcal{E}\tau')^2+4D'\tau'}-\mu'\mathcal{E}\tau'}{2}\ \xrightarrow{\ \mathcal{E}=0\ }\ \sqrt{D'\tau'}.$$

The simulation gives $\lambda$ (fit of $\ln\delta n$ between $0.6\,\delta(0)$ and $0.2\,\delta(0)$), $\tau'=\delta n/R$ and the field in the tail. Then $D'=\lambda_0^2/\tau'$ from the run at $V=0$, and $D'-\mu'\mathcal{E}\lambda=\lambda^2/\tau'$ at both biases gives $\mu'$. The figure shows $\lambda$, $D'$ and $\mu'$ against $N_a$ with the formulas above (gray) and their limits, and the normalized profiles. The case solves 14 operating points (about 0.1 s).

### Why the doping dependence is hidden in a plain diffusion experiment

- **$D'$ at $V=0$:** the tail length is $\sqrt{D'\tau'}$. From low to high injection $D'$ falls from $D_n=35$ to $18.3$ cm²/s (×0.52) while $\tau'$ doubles, so $\lambda$ changes by only $\sqrt{0.52\times2}\approx1.02$. The dotted profiles ($V=0$) almost coincide for all dopings; $D'$ appears only when $\tau'$ is measured separately (column $\tau'/\tau$ of the sweep table).
- **$\mu'$ needs a field:** at low injection the excess drifts with $\mu_n$, and $V=-1.5$ V ($\mathcal{E}\approx-50$ V/cm) stretches the tail from 18.7 to about 72 µm. At high injection $\mu'\to0$; in addition, the excess carriers raise the conductivity of the tail region, so the field there, $\mathcal{E}_J=J/\sigma$, is screened (below 1 V/cm for $N_a\le10^{12}$ cm⁻³) and the tail does not change. The table shows "—" for $\mu'$ where the drift changes $\lambda^2/\tau'$ by less than 30 %.
- **Transition, $\delta\approx N_a$:** $D'$, $\mu'$ and $\tau'$ change along the tail, so the ambipolar equation is no longer linear. The fitted values are effective values and deviate from the gray curves, which are evaluated at the typical density of the fit window ($10^{13}$–$10^{14}$ cm⁻³ here).
- The field in the ambipolar equation is the current-driven part $\mathcal{E}_J=J/\sigma$. The Dember field, which appears because $D_n\ne D_p$ even without current, is already contained in $D'$.

### How the simulation couples the carriers: quasi-neutrality

The electron and hole continuity equations each contain the field, and the field depends on both carriers. Neamen derives the ambipolar equation from the two continuity equations and the Poisson equation (6.32), then removes the Poisson equation with two assumptions (Sec. 6.3.1): charge neutrality, $\delta n\approx\delta p$, and an internal field small compared with the applied field (6.33). The simulation follows the same route and does not solve the Poisson equation (Lab 6 does):

- $\delta n=\delta p=\delta$ is imposed. The field is the one that carries the total current through the bar,
$$\mathcal{E}=\frac{J/e-(D_n-D_p)\,d\delta/dx}{\mu_nn+\mu_pp}\qquad(\texttt{drift\_diffusion.quasi\_neutral\_field}),$$
  with $J=0$ when an end is free and $J$ set by the bias otherwise. The term $(D_n-D_p)\,d\delta/dx$ is the Dember field that keeps the faster carrier from running ahead.
- The `continuity` module then solves for the minority carrier with this field; both steps repeat until $\delta$ stops changing. Only the modules of Chs. 3–6 are used.

Quasi-neutrality holds because any charge imbalance decays within the dielectric relaxation time $\varepsilon_s/\sigma$ (Sec. 6.3.4), far shorter than the lifetimes here.
