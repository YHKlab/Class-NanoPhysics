## Basic Theory

### Fermi–Dirac occupation and the energy reference

At thermal equilibrium, the probability that an electron state of energy $E$ is occupied is

$$
f(E)=\frac{1}{1+\exp[(E-E_F)/(k_BT)]}.
$$

Here $E_F$ is the Fermi level and $k_B$ is Boltzmann's constant. The probability that a state is empty is $1-f(E)$; an empty valence-band state corresponds to a hole.

Energies are referenced to midgap:

$$
E_C=\frac{E_g(T)}{2},\qquad E_V=-\frac{E_g(T)}{2},\qquad E_C-E_V=E_g(T).
$$

This reference describes the gap and relative levels, not absolute band-edge shifts relative to vacuum.

### Density of states

For three-dimensional parabolic bands, the effective densities of states are

$$
N_C(T)=2\left(\frac{2\pi m_{n,\mathrm{DOS}}^* k_BT}{h^2}\right)^{3/2},\qquad
N_V(T)=2\left(\frac{2\pi m_{p,\mathrm{DOS}}^* k_BT}{h^2}\right)^{3/2}.
$$

The factor 2 includes spin degeneracy. With SI quantities these expressions give m⁻³; dividing by $10^6$ gives cm⁻³. With fixed DOS masses, $N_C,N_V\propto T^{3/2}$.

The energy-dependent densities of states are

$$
g_c(E)=\frac{2N_C}{\sqrt{\pi}(k_BT)^{3/2}}\sqrt{E-E_C},\qquad E\ge E_C,
$$

$$
g_v(E)=\frac{2N_V}{\sqrt{\pi}(k_BT)^{3/2}}\sqrt{E_V-E},\qquad E\le E_V.
$$

Each DOS vanishes at its band edge and is zero outside that band. With energies in eV, the plotted DOS has units cm⁻³ eV⁻¹.

### Carrier concentrations and analytic distributions

The energy-resolved electron and hole densities are

$$
n(E)=g_c(E)f(E),\qquad p(E)=g_v(E)[1-f(E)].
$$

Their integrals give the total concentrations:

$$
n=\int_{E_C}^{\infty}g_c(E)f(E)\,dE,\qquad
p=\int_{-\infty}^{E_V}g_v(E)[1-f(E)]\,dE.
$$

Equivalently, with the normalized Fermi–Dirac integral,

$$
F_{1/2}(\eta)=\frac{2}{\sqrt{\pi}}\int_0^\infty\frac{\sqrt{x}}{1+\exp(x-\eta)}\,dx,
$$

$$
n=N_CF_{1/2}\!\left(\frac{E_F-E_C}{k_BT}\right),\qquad
p=N_VF_{1/2}\!\left(\frac{E_V-E_F}{k_BT}\right).
$$

The dashed analytic spectra are these continuous energy distributions. Bins only controls the sampled histogram drawn over them.

### Nondegenerate limit and intrinsic carriers

When the Fermi level is sufficiently far from either band, the Boltzmann approximation gives

$$
n\simeq N_C\exp\!\left(\frac{E_F-E_C}{k_BT}\right),\qquad
p\simeq N_V\exp\!\left(\frac{E_V-E_F}{k_BT}\right).
$$

Consequently,

$$
np=n_i^2,\qquad n_i\simeq\sqrt{N_CN_V}\exp\!\left(-\frac{E_g(T)}{2k_BT}\right).
$$

For an intrinsic semiconductor, $n=p$. In the nondegenerate limit its Fermi level is

$$
E_i=\frac{E_C+E_V}{2}+\frac{k_BT}{2}\ln\!\left(\frac{N_V}{N_C}\right).
$$

Thus $E_i$ need not coincide with midgap. These Boltzmann relations are limiting expressions; the plotted concentrations use Fermi–Dirac statistics.

### Dopant levels and incomplete ionization

Donor and acceptor energies are defined by their depths from the band edges:

$$
E_D=E_C-\Delta_D,\qquad E_A=E_V+\Delta_A.
$$

With the donor and acceptor statistical factors used here, the ionized concentrations are

$$
N_D^+=\frac{N_D}{1+2\exp[(E_F-E_D)/(k_BT)]},\qquad
N_A^-=\frac{N_A}{1+4\exp[(E_A-E_F)/(k_BT)]}.
$$

An ionized donor is positively charged; an ionized acceptor is negatively charged. At zero dopant concentration, a displayed level is only a reference marker and contributes no charge.

### Charge neutrality and temperature regions

Thermal equilibrium requires

$$
p+N_D^+=n+N_A^-.
$$

Together with the carrier and ionization expressions, this determines $E_F$.

| Region | Physical behavior |
| --- | --- |
| Freeze-out / partial ionization | At low temperature, many dopants remain neutral and the free-carrier concentration is small |
| Extrinsic | Ionized dopants dominate. For fully ionized, uncompensated n-type material, $n\simeq N_D$; for p-type material, $p\simeq N_A$ |
| Intrinsic | Thermal electron–hole generation dominates and $n\simeq p\simeq n_i$; $E_F$ approaches $E_i$ |

With the default $N_D=N_A=0$, the system is intrinsic and has no doping-induced extrinsic plateau.

### Material DOS masses

| Material | $m_{n,\mathrm{DOS}}^*/m_0$ | $m_{p,\mathrm{DOS}}^*/m_0$ | Definition used |
| --- | --- | --- | --- |
| Si | 1.0758 | 0.5559 | Chosen to reproduce $N_C(300\,\mathrm K)=2.8\times10^{19}$ and $N_V(300\,\mathrm K)=1.04\times10^{19}$ cm⁻³ |
| Ge | 0.55 | 0.37 | Electron and hole DOS masses |
| GaAs | 0.067 | 0.4730 | Gamma-valley electron mass; combined heavy- and light-hole DOS mass |
| Equal effective masses | 1 | 1 | Hypothetical symmetric DOS, giving $N_C=N_V$ |

Here $m_0$ is the free-electron mass. For GaAs, the heavy- and light-hole masses are combined as

$$
m_{p,\mathrm{DOS}}^*=\left(m_{hh}^{3/2}+m_{lh}^{3/2}\right)^{2/3},
\qquad m_{hh}=0.45m_0,\quad m_{lh}=0.082m_0,
$$

which gives $m_{p,\mathrm{DOS}}^*=0.4730433m_0$. The Si electron DOS mass already includes valley multiplicity.

### Temperature-dependent band gap

For Si, Ge, and GaAs, the Varshni relation gives

$$
E_g(T)=E_g(0)-\frac{\alpha T^2}{T+\beta}.
$$

| Material | $E_g(0)$ (eV) | $\alpha$ (eV/K) | $\beta$ (K) | $E_g(300\,\mathrm K)$ (eV) |
| --- | --- | --- | --- | --- |
| Si | 1.1695 | $4.73\times10^{-4}$ | 636 | 1.1240 |
| Ge | 0.7437 | $4.774\times10^{-4}$ | 235 | 0.6634 |
| GaAs | 1.521 | $5.58\times10^{-4}$ | 220 | 1.4244 |

The temperature-dependent gap enters both the carrier concentrations and the energy-level plots. Equal effective masses instead keeps the selected gap fixed. [Varshni parameter source](https://iue.tuwien.ac.at/phd/palankovski/node37.html)

The physical model assumes uniform thermal equilibrium, parabolic bands, fixed DOS masses, and isolated impurity levels. Impurity-band formation and doping-induced band-gap narrowing are not included.

```python
"""Uniform thermal equilibrium; energy eV (midgap=0), density cm^-3, T kelvin."""

KB = 8.617333262145e-5

# Explicit formats prevent Plotly's automatic k/M/B/T density abbreviations.
SCIENTIFIC_AXIS = dict(tickformat='.2e', hoverformat='.3e',
                       exponentformat='e', showexponent='all')


def effective_dos(mass_ratio, T=300):
    """Spin-inclusive 3-D DOS; SI calculation converted from m^-3 to cm^-3."""
    return 2*(2*np.pi*mass_ratio*m_e*KB_SI*T/h**2)**1.5 / 1e6

# Si DOS masses reproduce the original 300 K Nc/Nv values exactly.
# Combine these masses for the hole DOS: (0.45**1.5 + 0.082**1.5)**(2/3) = 0.4730433.
MATERIALS={
    'Equal masses': {'mn': 1.0, 'mp': 1.0, 'Eg': 1.12},
    'Si': {'mn': (2.8e19/effective_dos(1.0))**(2/3),
           'mp': (1.04e19/effective_dos(1.0))**(2/3), 'Eg': 1.12},
    'GaAs': {'mn': .067, 'mp': (.45**1.5+.082**1.5)**(2/3), 'Eg': 1.424},
    'Ge': {'mn': .55, 'mp': .37, 'Eg': .66},
}

# Varshni parameters (Eg0 in eV, alpha in eV/K, beta in K).
# TU Wien, Palankovski, Table 3.8:
# https://iue.tuwien.ac.at/phd/palankovski/node37.html
VARSHNI = {
    'Si': (1.1695, 4.73e-4, 636.),
    'Ge': (0.7437, 4.774e-4, 235.),
    'GaAs': (1.521, 5.58e-4, 220.),
}

def band_gap(material, T, custom_gap=1.12):
    if material == 'Equal masses':
        return custom_gap
    eg0, alpha, beta = VARSHNI[material]
    return eg0 - alpha*T*T/(T+beta)


def log_fermi_half(eta):
    """Log of normalized F_1/2; Boltzmann asymptote below eta=-30."""
    if eta < -30:
        return float(eta)
    integral = quad(lambda u: u*u*expit(eta-u*u), 0,
                    np.sqrt(max(eta, 0)+45), epsabs=1e-13, epsrel=1e-9,
                    points=[np.sqrt(eta)] if eta > 0 else None)[0]
    return np.log(4/np.sqrt(np.pi)*integral)


@dataclass(frozen=True)
class Parameters:
    T: float = 300
    Eg: float = 1.12
    Nd: float = 0
    Na: float = 0
    donor_depth: float = .045  # Ec-Ed
    acceptor_depth: float = .045  # Ea-Ev
    material: str = 'Si'

    @property
    def band_gap(self):
        return band_gap(self.material, self.T, self.Eg)

    @property
    def Nc300(self):
        return effective_dos(MATERIALS[self.material]['mn'])

    @property
    def Nv300(self):
        return effective_dos(MATERIALS[self.material]['mp'])

    def validate(self):
        if self.material not in MATERIALS:
            raise ValueError("Unknown material preset.")
        numeric = [v for v in vars(self).values() if isinstance(v, (int, float))]
        if not np.all(np.isfinite(numeric)):
            raise ValueError('Parameters must be finite.')
        if min(self.T, self.band_gap, self.Nc300, self.Nv300) <= 0:
            raise ValueError('Temperature, gap and DOS must be positive.')
        if min(self.Nd, self.Na) < 0:
            raise ValueError('Densities must be nonnegative.')
        if not all(0 <= d <= self.band_gap for d in (self.donor_depth, self.acceptor_depth)):
            raise ValueError('Dopant levels must lie in the gap.')


def _logcharges(ef, s):
    vt = KB*s.T
    ec, ev = s.band_gap/2, -s.band_gap/2
    nc, nv = s.Nc300*(s.T/300)**1.5, s.Nv300*(s.T/300)**1.5
    log_density = lambda d: np.log(d) if d > 0 else -np.inf
    log_sigmoid = lambda z: -np.logaddexp(0, -z)
    return dict(
        n=np.log(nc)+log_fermi_half((ef-ec)/vt),
        p=np.log(nv)+log_fermi_half((ev-ef)/vt),
        Nd_plus=log_density(s.Nd)+log_sigmoid((ec-s.donor_depth-ef)/vt-np.log(2)),
        Na_minus=log_density(s.Na)+log_sigmoid((ef-ev-s.acceptor_depth)/vt-np.log(4)))


def _balance(ef, s):
    q = _logcharges(ef, s)
    return logsumexp([q[k] for k in ('p','Nd_plus')])-logsumexp(
        [q[k] for k in ('n','Na_minus')])


def solve(s=Parameters()):
    """Self-consistent Ef: p + Nd+ = n + Na-. Log space avoids freeze-out underflow."""
    s.validate()
    width = max(2., 50*KB*s.T)
    lo, hi = -s.band_gap/2-width, s.band_gap/2+width
    for _ in range(30):
        if _balance(lo,s) > 0 and _balance(hi,s) < 0:
            break
        lo -= width
        hi += width
        width *= 2
    else:
        raise ValueError('Cannot bracket Fermi level.')
    ef = brentq(lambda e: _balance(e,s), lo, hi, xtol=1e-13)
    logs = _logcharges(ef,s)
    return dict(Ef=ef, **{k: np.exp(v) for k,v in logs.items()},
                log_n=logs['n'], log_p=logs['p'],
                residual=np.tanh(_balance(ef,s)/2),
                Ed=s.band_gap/2-s.donor_depth, Ea=-s.band_gap/2+s.acceptor_depth,
                Ec=s.band_gap/2, Ev=-s.band_gap/2,
                Nc=s.Nc300*(s.T/300)**1.5, Nv=s.Nv300*(s.T/300)**1.5)


def spectrum(energies, s, result):
    """n(E), p(E) in cm^-3 eV^-1, using the same DOS as the solver."""
    energies = np.asarray(energies)
    vt = KB*s.T
    gc = 2*result['Nc']/np.sqrt(np.pi)/vt**1.5*np.sqrt(np.maximum(energies-result['Ec'],0))
    gv = 2*result['Nv']/np.sqrt(np.pi)/vt**1.5*np.sqrt(np.maximum(result['Ev']-energies,0))
    return gc*expit((result['Ef']-energies)/vt), gv*expit((energies-result['Ef'])/vt)


def sweep(s, temperatures):
    return [solve(replace(s,T=float(t))) for t in temperatures]
```

```python
CARRIER_COLORS = {'n': '#dc2626', 'p': '#2563eb', 'Nd_plus': '#7c3aed', 'Na_minus': '#d97706', 'ni': '#059669'}

@lru_cache(maxsize=32)
def temperature_curves(s):
    ts = np.unique(np.r_[np.geomspace(5, 150, 65), np.linspace(150, 1200, 110)])
    return (ts, sweep(s, ts), sweep(replace(s, Nd=0, Na=0), ts))

@lru_cache(maxsize=32)
def doping_curves(s):
    ds = np.logspace(10, 20, 65)
    return (ds, [solve(replace(s, Nd=float(d))) for d in ds], [solve(replace(s, Na=float(d))) for d in ds])

def update(T, Eg, Nd, Na, Ed, Ea, axis, material='Si'):
    density = lambda x: 0 if x == 0 else 10.0 ** x
    s = Parameters(T=T, Eg=Eg, Nd=density(Nd), Na=density(Na), donor_depth=Ed, acceptor_depth=Ea, material=material)
    r = solve(s)
    Eg = s.band_gap
    ri = solve(replace(s, Nd=0, Na=0))
    summary = f"EF={r['Ef']:.6f} eV | Ei={ri['Ef']:.6f} eV | n={r['n']:.3e}, p={r['p']:.3e} cm⁻³"
    (ts, rs, ris) = temperature_curves(replace(s, T=300))
    x = ts if axis == 'T' else 1000 / ts
    temp = make_subplots(rows=1, cols=2, subplot_titles=['Electron concentration vs temperature', 'Fermi level vs temperature'])
    for key in ['n', 'p', 'Nd_plus', 'Na_minus']:
        temp.add_trace(go.Scatter(x=x, y=[q[key] if q[key] > 0 else None for q in rs], name=key, line={'color': CARRIER_COLORS[key]}), row=1, col=1)
    temp.add_trace(go.Scatter(x=x, y=[q['n'] if q['n'] > 0 else None for q in ris], name='ni', line={'dash': 'dot', 'color': CARRIER_COLORS['ni']}), row=1, col=1)
    if s.Nd:
        temp.add_hline(y=s.Nd, row=1, col=1, line_dash='dot', annotation_text='Nd')
    for (label, values) in [('EF', [q['Ef'] for q in rs]), ('Ei', [q['Ef'] for q in ris]), ('Ec', [q['Ec'] for q in rs]), ('Ev', [q['Ev'] for q in rs]), ('Ed', [q['Ed'] for q in rs]), ('Ea', [q['Ea'] for q in rs])]:
        temp.add_trace(go.Scatter(x=x, y=values, name=label), row=1, col=2)
    peak = max((max(q['n'], q['p']) for q in rs))
    temp.update_yaxes(type='log', range=[max(0, np.log10(peak) - 16), np.log10(peak) + 1], title_text='Concentration (cm⁻³, log scale)', row=1, col=1, **SCIENTIFIC_AXIS)
    temp.update_yaxes(title_text='Energy (eV)', row=1, col=2)
    temp.update_xaxes(title_text='T (K)' if axis == 'T' else '1000/T (K⁻¹)')
    temp.add_vline(x=T if axis == 'T' else 1000 / T, line_dash='dot')
    temp.update_layout(height=500)
    span = max(0.3, 20 * KB * T, abs(r['Ef']) - Eg / 2 + 0.3)
    edge = np.linspace(0, span, 1200) ** 2 / span
    energy = np.unique(np.r_[np.linspace(-Eg / 2 - span, Eg / 2 + span, 1000), Eg / 2 + edge, -Eg / 2 - edge])
    (ne, ph) = spectrum(energy, s, r)
    detail = make_subplots(rows=1, cols=3, shared_yaxes=True, subplot_titles=['Energy levels', 'Fermi–Dirac probability', 'Carrier density per energy'])
    levels = [('Ec', r['Ec']), ('Ev', r['Ev']), ('EF', r['Ef']), ('Ei', ri['Ef'])]
    # Show configured levels even when that dopant concentration is zero.
    levels.extend([('Ed', r['Ed']), ('Ea', r['Ea'])])
    for (label, e) in levels:
        detail.add_trace(go.Scatter(x=[0, 1], y=[e, e], name=label, mode='lines'), row=1, col=1)
    for (label, values) in [('f', expit((r['Ef'] - energy) / (KB * T))), ('1−f', expit((energy - r['Ef']) / (KB * T)))]:
        detail.add_trace(go.Scatter(x=values, y=energy, name=label, line={'color': CARRIER_COLORS['n' if label == 'f' else 'p']}), row=1, col=2)
    for (label, values) in [('gc f', ne), ('gv (1−f)', ph)]:
        detail.add_trace(go.Scatter(x=values, y=energy, name=label, line={'color': CARRIER_COLORS['n' if label == 'gc f' else 'p']}), row=1, col=3)
    detail.update_yaxes(title_text='Energy (eV)', row=1, col=1)
    detail.update_xaxes(title_text='cm⁻³ eV⁻¹', row=1, col=3, **SCIENTIFIC_AXIS)
    detail.update_layout(height=480)
    (ds, dr, ar) = doping_curves(s)
    doping = go.Figure()
    for (label, values) in [('Vary Nd (Na fixed)', dr), ('Vary Na (Nd fixed)', ar)]:
        doping.add_trace(go.Scatter(x=ds, y=[q['Ef'] for q in values], name=label))
    for (label, val) in [('Ec', Eg / 2), ('Ev', -Eg / 2), ('Ei', ri['Ef']), ('Ed', r['Ed']), ('Ea', r['Ea'])]:
        doping.add_hline(y=val, annotation_text=label, line_dash='dot')
    doping.update_layout(title=f'EF vs doping at {T:g} K; opposite dopant fixed', xaxis={'type': 'log', 'title': 'Varied concentration (cm⁻³)', **SCIENTIFIC_AXIS}, yaxis_title='Energy (eV)')
    return (summary, temp, detail, doping)
```

```python
@lru_cache(maxsize=32)
def equilibrium(T, Eg, Nd, Na, Ed, Ea, material='Si'):
    density = lambda value: 0 if value == 0 else 10.0 ** value
    s = Parameters(T=T, Eg=Eg, Nd=density(Nd), Na=density(Na), donor_depth=Ed, acceptor_depth=Ea, material=material)
    return (s, solve(s))

@lru_cache(maxsize=64)
def sampling_cdf(T, eta):
    x = np.linspace(0, np.sqrt(max(eta, 0) + 40), 2500) ** 2
    log_weight = -np.logaddexp(0, x - eta)
    weight = np.sqrt(x) * np.exp(log_weight - log_weight.max())
    cdf = np.r_[0, cumulative_trapezoid(weight, x)]
    cdf /= cdf[-1]
    return (x, cdf)

def sample_energies(s, r, carrier):
    log_density = r['log_n'] if carrier == 'n' else r['log_p']
    count = int(np.clip(18 * (log_density / np.log(10) - 5), 0, 300))
    if count == 0:
        return np.array([])
    eta = (r['Ef'] - r['Ec']) / (KB * s.T) if carrier == 'n' else (r['Ev'] - r['Ef']) / (KB * s.T)
    (x, cdf) = sampling_cdf(s.T, float(eta))
    kinetic = KB * s.T * np.interp(rng.random(count), cdf, x)
    return r['Ec'] + kinetic if carrier == 'n' else r['Ev'] - kinetic

def animate(frame, T, Eg, Nd, Na, Ed, Ea, bins, material='Si'):
    s,r=equilibrium(T,Eg,Nd,Na,Ed,Ea,material)
    Eg=s.band_gap
    electrons=sample_energies(s,r,'n')
    holes=sample_energies(s,r,'p')
    # Cover the entire energy support used by the inverse-CDF sampler.
    span=max(.15,40*KB*T+max(0,abs(r['Ef'])-Eg/2))
    low,high=-Eg/2-span,Eg/2+span
    fig=make_subplots(rows=1,cols=4,shared_yaxes=True,horizontal_spacing=.045,
        subplot_titles=['Representative particles','Fermi-Dirac occupation','Band density of states','Carrier density per energy'])
    for label,values,color,symbol in [('Electrons',electrons,CARRIER_COLORS['n'],'circle'),
                                      ('Holes',holes,CARRIER_COLORS['p'],'circle-open')]:
        fig.add_trace(go.Scatter(x=rng.random(len(values)),y=values,mode='markers',
            marker={'color':color,'size':6,'symbol':symbol},name=label,
            legendgroup=label),row=1,col=1)
    for label,level,color in [('Ec',r['Ec'],'#64748b'),('Ev',r['Ev'],'#64748b'),('EF',r['Ef'],'#059669')]:
        fig.add_hline(y=level,line_color=color,line_dash='dash',annotation_text=label,row=1,col=1)
    # Zero-density dopants still show their configured energy as a reference.
    for label,concentration,color,position in [
            ('Ed',s.Nd,'#7c3aed','bottom left'), ('Ea',s.Na,'#d97706','top left')]:
        fig.add_hline(y=r[label],line_dash='dot',line_color=color,
            opacity=1.0 if concentration else 0.55,
            annotation_text=f'{label} = {r[label]:.3f} eV' + (' (reference)' if not concentration else ''),
            annotation_position=position,row=1,col=1)
    edge_grid=np.linspace(0,np.sqrt(span),1400)**2
    energy=np.unique(np.r_[np.linspace(low,high,1600),r['Ec']+edge_grid,r['Ev']-edge_grid])
    for label,values in [('f(E)',expit((r['Ef']-energy)/(KB*T))),
                         ('1-f(E)',expit((energy-r['Ef'])/(KB*T)))]:
        fig.add_trace(go.Scatter(x=values,y=energy,name=label,
            line={'color': CARRIER_COLORS['n' if label=='f(E)' else 'p']}),row=1,col=2)
    # DOS is the number of available band states, before multiplying by occupancy.
    factor=2/np.sqrt(np.pi)/(KB*T)**1.5
    gc=factor*r['Nc']*np.sqrt(np.maximum(energy-r['Ec'],0))
    gv=factor*r['Nv']*np.sqrt(np.maximum(r['Ev']-energy,0))
    for label,values,color in [('gc(E)',gc,CARRIER_COLORS['n']),('gv(E)',gv,CARRIER_COLORS['p'])]:
        fig.add_trace(go.Scatter(x=values,y=energy,line={'color':color},name=label),row=1,col=3)
    for label,values,edge,direction,color in [('Electrons',electrons,r['Ec'],1,CARRIER_COLORS['n']),
                                            ('Holes',holes,r['Ev'],-1,CARRIER_COLORS['p'])]:
        edges=np.sort(np.linspace(edge,edge+direction*span,int(bins)+1))
        counts,_=np.histogram(values,bins=edges)
        # Use the same sampled particles, weighted by physical carrier density.
        density=r['n'] if direction==1 else r['p']
        heights=counts*density/(len(values)*np.diff(edges)) if len(values) else np.zeros_like(counts,dtype=float)
        fig.add_trace(go.Bar(x=heights,y=(edges[:-1]+edges[1:])/2,width=np.diff(edges),
            orientation='h',marker_color=color,opacity=.55,name=label+' sampled',
            legendgroup=label,customdata=np.c_[edges[:-1],edges[1:]],
            hovertemplate='Density: %{x:.3e} cm⁻³ eV⁻¹<br>Energy: %{customdata[0]:.4f} to %{customdata[1]:.4f} eV<extra>%{fullData.name}</extra>'),row=1,col=4)
        # Pointwise analytic density, independent of histogram bin width.
        # The grid contains Ec and Ev exactly; band DOS vanishes at both edges.
        expected=gc*expit((r['Ef']-energy)/(KB*T)) if direction==1 else gv*expit((energy-r['Ef'])/(KB*T))
        fig.add_trace(go.Scatter(x=expected,y=energy,
            line={'color':color,'dash':'dash'},name=label+' expected',legendgroup=label),row=1,col=4)
    fig.update_xaxes(range=[0,1],title_text='Representative position',row=1,col=1)
    fig.update_xaxes(type='linear',range=[0,1],title_text='Probability',row=1,col=2)
    fig.update_xaxes(title_text='States / (cm³ eV)',row=1,col=3,**SCIENTIFIC_AXIS)
    fig.update_xaxes(title_text='cm⁻³ eV⁻¹',autorange=False,row=1,col=4,**SCIENTIFIC_AXIS)
    fig.update_yaxes(range=[low,high],title_text='Energy (eV)',row=1,col=1)
    fig.update_layout(width=1700,height=780,autosize=False,template='plotly_white',barmode='overlay',
        margin={'t':55,'b':110},legend={'orientation':'h','y':-.18,'groupclick':'toggleitem'})
    summary=(f"Frame {frame} | EF={r['Ef']:.6f} eV | n={r['n']:.3e}, p={r['p']:.3e} cm⁻³\n"
             f"Relative charge imbalance={r['residual']:.2e}\n"
             f"Displayed samples: {len(electrons)} electrons, {len(holes)} holes. Histogram bars are weighted by physical carrier density.")
    masses=MATERIALS[material]
    summary=(f"Material: {material} | DOS masses: mn*/m0={masses['mn']:.4f}, mp*/m0={masses['mp']:.4f} | Eg={Eg:.3f} eV\n"
             f"Nc(T)={r['Nc']:.3e}, Nv(T)={r['Nv']:.3e} cm⁻³\n"+summary)
    return fig,summary


# Release old comms and callbacks when this cell is rerun.
if 'player' in globals():
    player.playing = False

def close_previous_widgets(widget, seen):
    if id(widget) in seen:
        return
    seen.add(id(widget))
    for child in getattr(widget, 'children', ()):
        close_previous_widgets(child, seen)
    widget.unobserve_all()
    widget.close()

_seen_widgets = set()
for _name in ('experiment_panel', 'player', 'animation_plot', 'sweep_plot'):
    if _name in globals():
        close_previous_widgets(globals()[_name], _seen_widgets)
del _seen_widgets


def widget_payload(value):
    """Avoid NumPy/binary-array properties in FigureWidget delta messages.

    Plotly's overlapping-property cleanup can test an ndarray's truth value
    when the frontend returns a nested property delta. Send ordinary lists
    consistently, both when adding traces and when updating existing ones.
    """
    if isinstance(value, np.ndarray):
        return widget_payload(value.tolist())
    if isinstance(value, np.generic):
        return value.item()
    if isinstance(value, dict):
        return {key: widget_payload(item) for key, item in value.items()}
    if isinstance(value, (list, tuple)):
        return [widget_payload(item) for item in value]
    return value

rng = np.random.default_rng()

def slider(label, low, high, value, step):
    return widgets.FloatSlider(description=label,min=low,max=high,value=value,step=step,
        continuous_update=False,style={'description_width':'200px'},
        layout=widgets.Layout(width='550px'))

material_change_in_progress=False
controls={
    'material':widgets.Dropdown(description='Material / DOS masses',
        options=[('Equal effective masses','Equal masses'),('Si','Si'),('GaAs','GaAs'),('Germanium (Ge)','Ge')],value='Si',
        style={'description_width':'200px'},layout=widgets.Layout(width='550px')),
    'T':slider('Temperature (K)',5,1200,300,5),
    'Eg':slider('Eg (eV)',.2,3,1.12,.01),
    'Nd':slider('log10 Nd; 0 = off',0,20,0,.25),
    'Na':slider('log10 Na; 0 = off',0,20,0,.25),
    'Ed':slider('Donor depth Ec - Ed (eV)',.001,.19,.045,.001),
    'Ea':slider('Acceptor depth Ea - Ev (eV)',.001,.19,.045,.001),
    'bins':widgets.IntSlider(description='Bins',min=10,max=80,value=35,continuous_update=False),
}
controls['Eg'].disabled = True
controls['Eg'].layout.display = 'none'

player=widgets.Play(value=0,min=0,max=1000000,step=1,interval=700,repeat=True,
                    description='Animate',show_repeat=True)
status=widgets.HTML()
summary_widget=widgets.HTML()
animation_plot=go.FigureWidget()
fit_button=widgets.Button(description='Fit axes to current settings')
axes_initialized=False
next_frame=widgets.Button(description='Next frame')
sweep_status=widgets.HTML()
last_sweep_parameters=None
sweep_plot=go.FigureWidget()
sweep_refresh=widgets.Button(description='Refresh temperature curves',
    layout=widgets.Layout(width='230px'))
# Preserve readable panel widths even in a narrow notebook output area.
animation_view=widgets.Box([animation_plot],layout=widgets.Layout(width='100%',overflow='auto'))
sweep_view=widgets.Box([sweep_plot],layout=widgets.Layout(width='100%',overflow='auto'))
temperature_section=widgets.Accordion(children=[widgets.VBox([
    widgets.HTML('Calculate n(T), p(T), ionized dopants, and EF(T) over 5–1200 K while holding doping and dopant depths fixed. Updates automatically when this section is open and a parameter slider is released.'),
    sweep_refresh,sweep_status,sweep_view])],selected_index=None)
temperature_section.set_title(0,'Temperature curves')


def fit_axes(data):
    # Keep all axes fixed until an explicit fit request.
    dos_peak=max(float(np.max(t.x)) for t in data.data if t.name in ('gc(E)','gv(E)'))
    density_peak=max([float(np.max(t.x)) for t in data.data if t.xaxis=='x4' and len(t.x)]+[1.0])
    with animation_plot.batch_update():
        animation_plot.update_yaxes(range=list(data.layout.yaxis.range),autorange=False)
        animation_plot.layout.xaxis.update(range=[0,1],autorange=False)
        animation_plot.layout.xaxis2.update(type='linear',range=[0,1],autorange=False)
        animation_plot.layout.xaxis3.update(range=[0,1.1*dos_peak],autorange=False)
        animation_plot.layout.xaxis4.update(range=[0,2*density_peak],autorange=False)


def render_frame(change=None, refit=False):
    global axes_initialized
    if material_change_in_progress:
        return
    try:
        args={key:widget.value for key,widget in controls.items()}
        data,summary=animate(player.value,**args)
        first=not axes_initialized
        with animation_plot.batch_update():
            # Retain existing traces whenever possible, including legend visibility.
            old_signature=[(trace.type,trace.name) for trace in animation_plot.data]
            new_signature=[(trace.type,trace.name) for trace in data.data]
            if old_signature!=new_signature:
                animation_plot.data=[]
                animation_plot.add_traces([widget_payload(trace.to_plotly_json()) for trace in data.data])
            else:
                for target,trace in zip(animation_plot.data,data.data):
                    props=widget_payload(trace.to_plotly_json())
                    props.pop('type',None)
                    target.update(props)
            if first:
                animation_plot.update_layout(widget_payload(data.layout.to_plotly_json()))
                animation_plot.update_layout(uirevision='fixed-axes',paper_bgcolor='white',
                    plot_bgcolor='white',font={'color':'#172033'})
            else:
                # Update band/dopant indicators without overwriting zoom or ranges.
                animation_plot.layout.shapes=data.layout.shapes
                animation_plot.layout.annotations=data.layout.annotations
            if first or refit:
                fit_axes(data)
        axes_initialized=True
        summary_widget.value='<pre>'+escape(summary)+'</pre>'
        status.value='Axes stay fixed during playback and parameter changes. Use Fit axes if data move out of view.'
        globals()['last_animation_summary']=summary
    except Exception as exc:
        player.playing=False
        status.value='<b>'+escape(str(exc))+'</b>'
        raise


def render_sweep(change=None, force=False):
    global last_sweep_parameters, sweep_plot
    if material_change_in_progress or temperature_section.selected_index is None:
        return
    args=tuple(controls[key].value for key in ['T','Eg','Nd','Na','Ed','Ea','material'])
    reopened = isinstance(change, dict) and change.get('owner') is temperature_section
    if args==last_sweep_parameters and not (force or reopened):
        return
    sweep_status.value='Calculating temperature curves...'
    try:
        s,_=equilibrium(*args)
        ts,rs,ris=temperature_curves(replace(s,T=300))
        temperature=make_subplots(rows=1,cols=2,
            subplot_titles=['Carrier concentration vs temperature','Fermi level vs temperature'])
        carrier_curves={key:np.array([r[key] for r in rs]) for key in ['n','p','Nd_plus','Na_minus']}
        carrier_curves['ni']=np.array([r['n'] for r in ris])
        for key,values in carrier_curves.items():
            temperature.add_trace(go.Scatter(
                x=ts,
                y=[float(value) if value>0 else None for value in values],
                name=key,
                line={'color': CARRIER_COLORS[key], 'dash': 'dot' if key=='ni' else 'solid'},
                hovertemplate='T=%{x:.1f} K'
                              '<br>Concentration=%{y:.3e} cm⁻³<extra>%{fullData.name}</extra>'),
                row=1,col=1)
        for label,values in [('EF',[r['Ef'] for r in rs]),('Ei',[r['Ef'] for r in ris]),
                             ('Ec',[q['Ec'] for q in rs]),('Ev',[q['Ev'] for q in rs]),
                             ('Ed',[q['Ed'] for q in rs]),
                             ('Ea',[q['Ea'] for q in rs])]:
            temperature.add_trace(go.Scatter(x=ts,y=values,name=label),row=1,col=2)
        peak=max(float(values.max()) for values in carrier_curves.values())
        log_floor=max(0,np.log10(peak)-16)
        temperature.update_yaxes(type='log',range=[log_floor,np.log10(peak)+1],
            autorange=False,title_text='Concentration (cm⁻³, log scale)',row=1,col=1,**SCIENTIFIC_AXIS)
        temperature.update_yaxes(tickmode='linear',tick0=0,dtick=1,tickformat='.0e',row=1,col=1)
        temperature.update_yaxes(title_text='Energy (eV)',autorange=True,row=1,col=2)
        temperature.update_xaxes(title_text='Temperature T (K)',type='linear',
            range=[float(ts.min()),float(ts.max())],autorange=False,row=1,col=1)
        temperature.update_xaxes(title_text='Temperature T (K)',type='linear',
            range=[float(ts.min()),float(ts.max())],autorange=False,row=1,col=2)
        temperature.add_vline(x=s.T,line_dash='dot',row=1,col=1)
        temperature.add_vline(x=s.T,line_dash='dot',row=1,col=2)
        # Initialize the frontend with complete axes instead of merging into a
        # FigureWidget that may retain an earlier linear-axis state.
        temperature.update_layout(width=1500,height=600,autosize=False,
            template='plotly_white',paper_bgcolor='white',font={'color':'#172033'})
        temperature.layout.yaxis.update(type='log',tickmode='linear',tick0=0,dtick=1,
            tickformat='.0e',hoverformat='.3e',
            title_text='Concentration (cm⁻³, log scale)')
        replacement=go.FigureWidget(widget_payload(temperature.to_plotly_json()))
        previous=sweep_plot
        sweep_plot=replacement
        sweep_view.children=(sweep_plot,)
        previous.close()
        last_sweep_parameters=args
        sweep_status.value=(
            'Left: concentration vs T (linear x, log y). '
            'Right: energy vs T. Dotted vertical lines mark the selected temperature.'
            + (' With Nd=Na=0, n=p=ni overlap; ionized dopants are zero.' if s.Nd==s.Na==0 else ''))
    except Exception as exc:
        sweep_status.value=escape(str(exc))
        raise

def material_changed(change):
    global material_change_in_progress
    material_change_in_progress=True
    try:
        # Named materials calculate Eg(T); retain the custom slider setting.
        custom = change['new'] == 'Equal masses'
        controls['Eg'].disabled = not custom
        controls['Eg'].layout.display = '' if custom else 'none'
    finally:
        material_change_in_progress=False
    render_frame()
    render_sweep()

controls['material'].observe(material_changed,names='value')
for key,widget in controls.items():
    if key!='material': widget.observe(render_frame,names='value')
player.observe(render_frame,names='value')
next_frame.on_click(lambda button: setattr(player,'value',player.value+1))
for key in ['T','Eg','Nd','Na','Ed','Ea','material']:
    if key!='material': controls[key].observe(render_sweep,names='value')
temperature_section.observe(render_sweep,names='selected_index')
sweep_refresh.on_click(lambda button: render_sweep(force=True))
fit_button.on_click(lambda button: render_frame(refit=True))
experiment_panel=widgets.VBox([
    widgets.HTML('<b>Carrier statistics — v2</b><br>Press Play to animate. Controls update the equilibrium distribution. Each frame is an independent sample, not a physical trajectory.'),
    widgets.HBox([player,next_frame,fit_button]),*controls.values(),status,summary_widget,animation_view,
    widgets.HTML('Ed and Ea are always shown in eV relative to midgap. At zero dopant concentration, their faint dotted lines are reference levels only. Inputs specify Ec - Ed and Ea - Ev in eV.'),
    widgets.HTML('Si, Ge, and GaAs use a temperature-dependent band gap Eg(T). The Eg slider is available only for Equal effective masses. Dopant depths remain measured from their band edges. Use Fit axes if curves are outside the view.'),
    widgets.HTML('Dots are representative samples; histogram bars are weighted by physical carrier density. Dashed curves show gc(E)f(E) and gv(E)[1-f(E)], which vanish at the band edges. DOS is shown separately.'),
    temperature_section])
display(experiment_panel)
render_frame()
```
