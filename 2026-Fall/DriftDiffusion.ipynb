## Basic Theory

### Drift–diffusion equations (module `drift_diffusion`, Neamen Ch. 5)

$$J_n=e\mu_nn\mathcal{E}+eD_n\frac{dn}{dx},\qquad J_p=e\mu_pp\mathcal{E}-eD_p\frac{dp}{dx}\qquad(5.35),\qquad \frac{D_n}{\mu_n}=\frac{D_p}{\mu_p}=\frac{kT}{e}\qquad(5.47)$$

The densities come from the `statistics` module: in equilibrium $n_0$ and $p_0$ follow from charge neutrality with complete ionization, (4.60) and (4.62), and the Boltzmann relations $n=n_ie^{(E_F-E_{Fi})/kT}$, $p=n_ie^{(E_{Fi}-E_F)/kT}$ give the potential $\psi=-E_{Fi}/e$. In a uniform bar the densities are uniform, diffusion vanishes and the drift current is $J=e(\mu_nn+\mu_pp)\mathcal{E}=\sigma\mathcal{E}$ (5.19), with $\rho=1/\sigma$ (5.20). In a graded bar the carriers diffuse away from the heavily doped side until the induced field balances diffusion; with quasi-neutrality $n\approx N_d(x)$ (5.39) and $J_n=0$ (5.43) this gives $\mathcal{E}=-(kT/e)(1/N_d)(dN_d/dx)$ (5.42).

### Textbook assumptions

Every model of this lab, and every density of the simulation, rests on the assumptions below. The last column says how the simulation treats each one.

| Assumption | Where Neamen uses it | This simulation |
| --- | --- | --- |
| **Boltzmann approximation** $f_F(E)\approx e^{-(E-E_F)/kT}$ (3.80): nondegenerate semiconductor | $n_0$, $p_0$, $n_i$ (Ch. 4); $E_F-E_{Fi}$ in (5.40); Einstein relation (5.47) | Kept (`statistics`). Fails when $E_F$ approaches a band edge (Si above about $10^{18}$–$10^{19}$ cm⁻³) |
| **Complete ionization** $N_d^+=N_d$, $N_a^-=N_a$ (Sec. 4.4.2) | $n_0\approx N_d$ (4.60) | Kept at every temperature; no freeze-out (compare the Fermi–Dirac notebook) |
| **Charge neutrality** $n_0+N_a^-=p_0+N_d^+$ (4.56) | $n_0$, $p_0$ (4.60), (4.62) | Kept for the equilibrium densities |
| **Low-field drift** $v_d=\mu\mathcal{E}$ (5.4) with a **constant mobility** (Secs. 5.1.2–5.1.4) | $J_{drf}=\sigma\mathcal{E}$ (5.19) | Kept: $\mu$ independent of field, doping and $T$; no velocity saturation (Sec. 5.1.4). Neamen notes that $\mu(T)$ from lattice scattering, $\mu_L\propto T^{-3/2}$ (5.15), is the main temperature dependence of $D$ (p. 179) |
| **Quasi-neutrality** $n\approx N_d(x)$ (5.39) | Induced field (5.42), derivation of the Einstein relation (5.44)–(5.47) | Imposed: the majority density equals $N(x)$ (the Poisson equation is not solved in this lab; see Lab 6) |
| **Thermal equilibrium** $J_n=J_p=0$ (5.43) | (5.42), (5.47) | Exact at $V=0$; under bias the field becomes $\mathcal{E}_{bi}+J/\sigma$ |
| **Einstein relation** $D/\mu=kT/e$ (5.47) | Diffusion coefficients (Table 5.2) | Used as an input ($D=\mu kT/e$); 4-MC obtains it from scattering instead |

**Simulation model.** One dimension; steady state; one lattice temperature; constant $E_g$ and $\varepsilon_r$ with $n_i(T)=n_i(300)(T/300)^{3/2}e^{-E_g/2k(1/T-1/300)}$; ohmic contacts (equilibrium densities, no contact resistance); no generation or recombination; uniform grid with Scharfetter–Gummel currents. **Monte Carlo (4-MC):** constant conductivity mass $m^*=0.26\,m_0$, collision probability $\Delta t/\tau$ with $\tau=\mu m^*/q$ independent of energy, velocities redrawn from the Maxwell distribution after each collision, two-dimensional motion, no recombination; valid at low fields.

### Analytic models used for comparison

| Structure / condition | Quantity | Analytic model [1] |
| --- | --- | --- |
| Uniform bar, $V\ne0$ | Field, current, conductivity | $\mathcal{E}=V/L$, $J=\sigma\mathcal{E}$, $\sigma=e(\mu_nn_0+\mu_pp_0)$ (5.19), $\rho=1/\sigma$ (5.20) |

<summary><strong>Advanced Theory</strong></summary>

<h3>Grid and units</h3>

<p>The device is sampled on a uniform grid $x_i=i\,\Delta x$. Internally lengths are in cm, potentials in V, energies in eV and concentrations in cm⁻³. With <strong>Grid points = 0</strong> the bar uses 1001 points; the cases preset 201 (4-A) and 401 (4-B) points.</p>

<h3>Scharfetter–Gummel currents (module <code>drift_diffusion</code>)</h3>

<p>With $\Delta_i=(\psi_{i+1}-\psi_i)/V_t$ and the Bernoulli function $B(z)=z/(e^z-1)$, the currents between nodes $i$ and $i+1$ are [2, 3]</p>

$$J_{n,i+1/2}=\frac{qD_n}{\Delta x}\left[B(\Delta_i)\,n_{i+1}-B(-\Delta_i)\,n_i\right],\qquad J_{p,i+1/2}=\frac{qD_p}{\Delta x}\left[B(\Delta_i)\,p_i-B(-\Delta_i)\,p_{i+1}\right].$$

<p>The expression is exact for a constant field between nodes and gives zero current for Boltzmann-distributed densities in equilibrium. The drift part is evaluated from the average density and the midpoint field; diffusion is the remainder.</p>

<h3>Quasi-neutral model</h3>

<p>The densities are the charge-neutral equilibrium densities $n_0(x)$, $p_0(x)$ of the <code>statistics</code> module, and $\psi_{eq}(x)=V_t\sinh^{-1}[(N_d-N_a)/2n_i]$. Under bias the current is uniform, $J=V\big/\int dx/\sigma(x)$, and the potential becomes $\psi=\psi_{eq}+V-\int_0^x J/\sigma\,dx'$, i.e. $\mathcal{E}=\mathcal{E}_{bi}+J/\sigma$. The densities do not change with bias (no injection in Ch. 5).</p>

<h3>Monte Carlo (4-MC)</h3>

<p>Electrons start at the origin with Maxwell velocities, $\langle v_x^2\rangle=kT/m^*$. Between collisions an electron accelerates with $a=-q\mathcal{E}/m^*$; in each time step $\Delta t\le\tau/50$ it scatters with probability $\Delta t/\tau$ and its velocity is redrawn. $\mu$ is extracted from the slope of $\langle x\rangle(t)$ and $D$ from the slope of $\sigma_y^2(t)=2Dt$ (across the field, where drift does not contribute), both for $t>5\tau$. The histogram of the distribution uses 200 bins over the plotted range. The flux panel bins the electrons along $x$ in 40 bins that follow the cloud, $-\mu\mathcal{E}t\pm4\sqrt{2Dt}$: the particle flux per electron is $F_{MC}=\sum v_x/(N\Delta x)$ from the velocities, the drift term is $-\mu\mathcal{E}P$ and the diffusion term $-D\,dP/dx$, with $P$ the normalized histogram and $dP/dx$ from central differences. Row <code>flux_resid</code> is the rms of $F_{MC}-(F_{drift}+F_{diff})$ over the bins with $P>5\,\%$ of its maximum, divided by $\max\lvert F_{MC}\rvert$, at the last frame; with 20000 electrons it is 0.05–0.1 (statistical noise). The peak flux falls as $t^{-1/2}$ while the cloud spreads, so the $y$ axis of the flux panel zooms in smoothly: in every frame it spans the drift term, the diffusion term and their sum for a Gaussian with the measured mean and width of the cloud (the width is limited to $\sqrt{2Dt}$), widened by three standard deviations of the counting noise and by 10 %. It does not use the noisy Monte Carlo values, so it does not jitter from frame to frame. The $y$ axis of the mean position spans at least $\pm\sigma(t_{max})/4$, a quarter of the final cloud width, so that the statistical scatter $\sigma/\sqrt{N}$ of $\langle x\rangle$ is seen on the scale of the cloud.</p>

# (1) Calculation

Modules of `semi1d` used by this lab, then the parts written in this notebook: units, materials and device; the lab model (how the modules are coupled) and its analysis; runners and case presets.

```python
###############################################################################
# Imports: modules of semi1d (folder one level above this notebook)
###############################################################################

import os
import re
import sys
sys.path.insert(0, os.path.abspath('..'))

import numpy as np
from dataclasses import dataclass, field
from types import SimpleNamespace
from scipy import constants as sc_const

from semi1d import statistics          # Ch. 3-4: n_i(T), equilibrium densities, quasi-Fermi levels
from semi1d import drift_diffusion     # Ch. 5:   drift and diffusion currents, Einstein relation, induced field
```

```python
###############################################################################
# Units, Materials and Device
###############################################################################
# Internal units: length [cm], potential [V], energy [eV], time [s], density [cm^-3]; positions shown in [um].

Q = sc_const.e                          # Elementary charge [C]
EPS0 = sc_const.epsilon_0 * 1e-2        # Vacuum permittivity [F/cm]
UM2CM, US2S = 1e-4, 1e-6


@dataclass(frozen=True)
class Material:
    name: str
    eps_r: float        # Relative permittivity
    Eg: float           # Band gap [eV]
    ni300: float        # Intrinsic carrier concentration at 300 K [cm^-3]
    mun: float          # Electron mobility [cm^2/V s]
    mup: float          # Hole mobility [cm^2/V s]


# 300 K values: Neamen (2012), Table 5.1 (mobility) and Appendix B.4
MATERIALS = {
    'Si':   Material('Si',   11.7, 1.12, 1.5e10, 1350.0,  480.0),
    'Ge':   Material('Ge',   16.0, 0.66, 2.4e13, 3900.0, 1900.0),
    'GaAs': Material('GaAs', 13.1, 1.42, 1.8e6,  8500.0,  400.0),
}


@dataclass
class GridConfig:
    points_per_debye: float = 8.0       # Auto grid: dx <= L_D(min) / points_per_debye
    npts_min: int = 401                 # Auto grid: lower limit of grid points
    npts_max: int = 20001               # Auto grid: upper limit of grid points
    npts_uniform: int = 1001            # Auto grid for a uniform bar (no built-in space charge)


GRID_CFG = GridConfig()


class Device:
    """
    1D device: grid x [um] (measured from the left contact), doping N_a(x), N_d(x) [cm^-3], material and
    transport parameters (n_i(T) from semi1d.statistics, D from the Einstein relation in semi1d.drift_diffusion).
    x0_um is the plotting origin (junction or device center); srh=False removes SRH recombination (R = 0).
    bc_left / bc_right: 'ohmic' contact (equilibrium concentrations) or 'free' surface (no carrier flux).
    """

    def __init__(self, x_um, Na, Nd, material='Si', T=300.0, ni300=None, mun=None, mup=None,
                 taun=1e-7, taup=1e-7, srh=True, x0_um=None, bc_left='ohmic', bc_right='ohmic'):
        mat = MATERIALS[material] if isinstance(material, str) else material
        self.material = mat
        self.x_um = np.asarray(x_um, dtype=float)
        self.x = self.x_um * UM2CM                                  # [cm]
        self.N = len(self.x)
        self.dx = self.x[1] - self.x[0]
        self.Na = np.broadcast_to(np.asarray(Na, dtype=float), self.x.shape).copy()
        self.Nd = np.broadcast_to(np.asarray(Nd, dtype=float), self.x.shape).copy()
        self.C = self.Nd - self.Na                                  # Net doping
        self.T = float(T)
        self.Vt = statistics.thermal_voltage(self.T)
        self.Eg = mat.Eg
        self.ni = statistics.intrinsic_density(self.T, mat.Eg, mat.ni300 if ni300 is None else float(ni300))
        self.mun = mat.mun if mun is None else float(mun)
        self.mup = mat.mup if mup is None else float(mup)
        self.Dn, self.Dp = drift_diffusion.einstein(self.mun, self.Vt), drift_diffusion.einstein(self.mup, self.Vt)
        self.taun, self.taup = float(taun), float(taup)
        self.srh = bool(srh)
        self.tau_eff_n = self.taun if self.srh else np.inf      # Used by the analytic models
        self.tau_eff_p = self.taup if self.srh else np.inf
        self.eps = mat.eps_r * EPS0
        self.x0_um = 0.5 * self.x_um[-1] if x0_um is None else float(x0_um)
        if bc_left not in ('ohmic', 'free') or bc_right not in ('ohmic', 'free'):
            raise ValueError("Ends must be 'ohmic' (contact) or 'free' (no surface recombination).")
        self.bc_left, self.bc_right = bc_left, bc_right


def debye_length_um(N, eps, Vt):
    """Extrinsic Debye length (eps kT / e^2 N)^{1/2} [um] (used to choose the grid)."""
    return np.sqrt(eps * Vt / (Q * N)) / UM2CM


def grid_points(uniform, L_um, Nmax, eps, Vt, npts=None):
    """Grid points: npts, or automatic (1001 for a uniform bar; otherwise dx <= L_D(Nmax)/8, 401-20001)."""
    if npts is None:
        if uniform:
            return GRID_CFG.npts_uniform
        LD = debye_length_um(Nmax, eps, Vt)
        npts = int(np.clip(int(np.ceil(GRID_CFG.points_per_debye * L_um / LD)) + 1, GRID_CFG.npts_min,
                           GRID_CFG.npts_max))
    if int(npts) < 11:
        raise ValueError("Use at least 11 grid points.")
    return int(npts)


def check_positive(**values):
    bad = [k for k, v in values.items() if not np.isfinite(v) or v <= 0]
    if bad:
        raise ValueError(f"Lengths, temperature, lifetimes and doping concentrations must be positive: {bad}")


def print_device(dev, structure_name, say, origin='device center'):
    """Device and grid block of the log."""
    LD_min = debye_length_um(float(np.max(np.maximum(dev.Na, dev.Nd))), dev.eps, dev.Vt)
    say('(0) Device and grid')
    say(f'structure \t = {structure_name}, {dev.material.name}, T = {dev.T:g} K')
    say(f'device length \t = {dev.x_um[-1]:.3f} µm, number of grid = {dev.N}, dx = {dev.dx / UM2CM * 1e3:.3f} nm')
    say(f'x range \t = {-dev.x0_um:.3f} to {dev.x_um[-1] - dev.x0_um:.3f} µm (x = 0 at the {origin})')
    say(f'Eg, ni, kT/e \t = {dev.Eg:.3g} eV, {dev.ni:.4e} cm^-3, {dev.Vt:.5f} V')
    say(f'μn, μp \t = {dev.mun:g}, {dev.mup:g} cm^2/V s  (Dn, Dp = {dev.Dn:.4g}, {dev.Dp:.4g} cm^2/s)')
    return LD_min


###############################################################################
# Comparison Table and Results
###############################################################################

ITER_ROWS = ('iters',)        # Printed first in every table


def _is_num(v):
    return isinstance(v, (int, float, np.integer, np.floating)) and not isinstance(v, bool) and np.isfinite(v)


def plain_sub(s):
    """Subscripts in printed text, where they cannot be typeset: 'E_g', 'μ_n', 'N_{a}' -> 'Eg', 'μn', 'Na'."""
    return re.sub(r'(?<=[^\W\d_])_(\{[^}]*\}|[A-Za-z0-9]+)', lambda m: m.group(1).strip('{}'), s) if isinstance(s, str) else s


def row(key, name, sim, ref=None, model=''):
    """One line of the comparison table. diff = (sim - ref)/|ref| when both are numbers. The key is the name used
    by res.value(key); subscripts in the quantity and the model are written without underscores (plain_sub)."""
    sim = float(sim) if _is_num(sim) else sim
    ref = float(ref) if _is_num(ref) else ref
    diff = (sim - ref) / abs(ref) if _is_num(sim) and _is_num(ref) and ref != 0 else None
    return dict(key=key, name=plain_sub(name), sim=sim, ref=ref, diff=diff, model=plain_sub(model))


def _fmt(v, width=12):
    if isinstance(v, str):
        s = v
    elif not _is_num(v):                          # None or NaN
        s = '—'
    elif isinstance(v, (int, np.integer)):
        s = f'{v:d}'
    else:
        s = f'{v:.4g}'
    return f'{s:>{width}s}'


def select_rows(rows, keys=None):
    """Rows to print: iteration rows first, then the requested keys in the given order (None = all rows)."""
    if keys is None:
        return rows
    by_key = {r['key']: r for r in rows}
    out = [by_key[k] for k in ITER_ROWS if k in by_key]
    out += [by_key[k] for k in keys if k in by_key and k not in ITER_ROWS]
    missing = [k for k in keys if k not in by_key]
    if missing:
        print(f'(not applicable for these parameters: {", ".join(missing)})')
    return out


def print_table(rows):
    """Print the comparison table: Key | Quantity | Simulation | Analytic | Diff. | Model"""
    print(f"{'Key':<14s} {'Quantity':<46s} {'Simulation':>12s} {'Analytic':>12s} {'Diff.':>8s}  Model")
    print('-' * 118)
    for r in rows:
        diff = f"{100 * r['diff']:+.1f}%" if r['diff'] is not None else '—'
        print(f"{r['key']:<14s} {r['name']:<46s} {_fmt(r['sim'])} {_fmt(r['ref'])} {diff:>8s}  {r['model']}")


@dataclass
class SimulationResult:
    """Result of a case: res.table (comparison rows), res.data (arrays), res.value(key)."""
    params: dict
    device: object
    states: list
    table: list
    data: dict = field(default_factory=dict)
    case: str = None
    panels: list = None

    def row(self, key):
        for r in self.table:
            if r['key'] == key:
                return r
        raise KeyError(f"No table row '{key}'. Available keys: {[r['key'] for r in self.table]}")

    def value(self, key, which='sim'):
        """Table value by key; which = 'sim', 'ref' or 'diff'."""
        return self.row(key)[which]


###############################################################################
# Case Runner
###############################################################################
# CASES = {case_id: dict(title, summary, params, ranges, fixed, runner, rows, cols, panels, schematic)}

def to_float(text, name='value'):
    try:
        return float(str(text).strip())
    except ValueError:
        raise ValueError(f"'{name}' must be a number such as 1e16; got {text!r}.") from None


def run_case(case_id, verbose=True, **overrides):
    """Run a case, e.g. run_case(list(CASES)[0], T=350.0). Parameters not given keep their presets."""
    case = CASES[case_id]
    bad = set(overrides) - set(case['params'])
    if bad:
        raise KeyError(f"Case {case_id} has no parameter {sorted(bad)}; available: {list(case['params'])}")
    kw = {**case.get('fixed', {}), **case_kwargs({**case['params'], **overrides})}
    if verbose:
        print(f"Case {case_id}: {case['title']}\n")
    if 'cols' in case:
        kw['cols'] = case['cols']
    res = case['runner'](verbose=verbose, rows=case.get('rows'), **kw)
    res.case, res.panels = case_id, case.get('panels')
    return res


def list_cases():
    """Print the case IDs, titles and parameter presets."""
    for cid, case in CASES.items():
        ps = ', '.join(f'{k}={v:g}' if isinstance(v, float) else f'{k}={v!r}' for k, v in case['params'].items())
        print(f"{cid}  {case['title']}\n      run_case('{cid}', {ps})")
```

```python
###############################################################################
# Lab 4 Devices: uniform bar and graded doping
###############################################################################

STRUCTURES = {0: 'Uniform bar', 1: 'Linearly graded doping', 2: 'Exponentially graded doping'}


def doping_profile(structure, x_um, doping_type='n', N=1e16, N0=1e16, NL=2e15, decay_um=1.0):
    """
    Doping on the grid: 0 uniform N; 1 linear N(0) + [N(L) - N(0)] x/L; 2 exponential N(0) exp(-x/lambda).
    output: N_a(x), N_d(x) [cm^-3] (donors for n-type, acceptors for p-type)
    """
    x_um = np.asarray(x_um, dtype=float)
    L = x_um[-1]
    if structure == 0:
        prof = np.full_like(x_um, N)
    elif structure == 1:
        prof = N0 + (NL - N0) * x_um / L
    elif structure == 2:
        prof = N0 * np.exp(-x_um / decay_um)
    else:
        raise ValueError("Unknown structure.")
    zero = np.zeros_like(x_um)
    return (prof, zero) if doping_type == 'p' else (zero, prof)


def build_lab4_device(structure=0, doping_type='n', N=1e16, N0=1e16, NL=2e15, decay_um=1.0, length_um=10.0,
                      material='Si', T=300.0, npts=None):
    """Bar of length L on a uniform grid; x = 0 of the plots at the center."""
    values = dict(length_um=length_um, T=T)
    values.update({0: dict(N=N), 1: dict(N0=N0, NL=NL), 2: dict(N0=N0, decay_um=decay_um)}[structure])
    check_positive(**values)
    mat = MATERIALS[material]
    profile = dict(doping_type=doping_type, N=N, N0=N0, NL=NL, decay_um=decay_um)
    Na_p, Nd_p = doping_profile(structure, np.linspace(0.0, length_um, 2001), **profile)
    npts = grid_points(True, length_um, float(np.max(np.maximum(Na_p, Nd_p))), mat.eps_r * EPS0,
                       statistics.thermal_voltage(T), npts)
    x = np.linspace(0.0, length_um, npts)
    Na_x, Nd_x = doping_profile(structure, x, **profile)
    return Device(x, Na_x, Nd_x, material=mat, T=T, srh=False)


###############################################################################
# Coupling of the Modules (quasi-neutral bar, Neamen Sec. 5.3)
###############################################################################
# Densities: charge-neutral equilibrium n0(x), p0(x) of semi1d.statistics (quasi-neutrality, (5.39)); potential
# psi_eq(x) of the neutral bar. Under bias the current is uniform, J = V / integral dx/sigma(x), and the field becomes
# E_bi(x) + J/sigma(x) (Ohm's law for the majority carriers). Currents: semi1d.drift_diffusion.

class DriftSolver:
    """Currents of a bar under bias V (left contact; right contact grounded), quasi-neutral densities."""

    def __init__(self, dev):
        self.d = dev
        self.psi_eq = statistics.neutral_potential(dev.C, dev.ni, dev.Vt)
        self.n_eq, self.p_eq = statistics.boltzmann_densities(self.psi_eq, dev.ni, dev.Vt)

    def solve(self, V=0.0):
        d = self.d
        n, p = self.n_eq.copy(), self.p_eq.copy()
        sig = d.mun * 0.5 * (n[1:] + n[:-1]) + d.mup * 0.5 * (p[1:] + p[:-1])         # sigma/q at midpoints
        J = Q * V / (np.sum(1.0 / sig) * d.dx)                                         # J = V / integral dx/sigma
        psi = self.psi_eq + V - np.concatenate([[0.0], np.cumsum(J / (Q * sig)) * d.dx])
        return self.pack(V, psi, n, p)

    def pack(self, V, psi, n, p):
        d = self.d
        J = drift_diffusion.currents(d, psi, n, p)
        bands = statistics.band_diagram(psi, n, p, d.ni, d.Vt, d.Eg)
        xh = 0.5 * (d.x_um[1:] + d.x_um[:-1])
        return SimpleNamespace(
            V=V, x=d.x_um, xh=xh, xc=d.x_um - d.x0_um, xhc=xh - d.x0_um, psi=psi, n=n, p=p,
            E=np.gradient(-psi, d.x), **bands, **vars(J), J=float(np.mean(J.Jt)), dev=d)


###############################################################################
# Lab 4 Analysis
###############################################################################

def analyze_drift(st, s, p):
    """
    Compare one operating point of a uniform bar under bias, or of a graded bar, with the Ch. 5 models.

    output:
        list    rows    Comparison-table rows
        dict    ov      Analytic overlays for plotting {panel: [(x, y, label, dash), ...]}
    """
    d = st.dev
    x, L = st.x, st.x[-1]
    V = st.V
    structure = p['structure']
    rows, ov = [], {k: [] for k in ('E', 'rho')}
    i = d.N // 2
    if structure == 0:                     # Uniform bar: E = V/L, J = sigma E
        n0, p0 = s.n_eq[i], s.p_eq[i]
        ntype = p['doping_type'] == 'n'
        if V != 0:
            E_app = V / (L * UM2CM)
            sig = drift_diffusion.conductivity(n0, p0, d.mun, d.mup)
            maj_drift = st.Jn_drift[i] if ntype else st.Jp_drift[i]
            rows += [row('E_center', 'Electric field at the center [V/cm]', st.E[i], E_app, 'E = V/L'),
                     row('J', 'Current density J [A/cm²]', st.J, sig * E_app, 'J = σE (5.19)'),
                     row('sigma', 'Conductivity σ = J/(V/L) [S/cm]', st.J / E_app, sig, 'σ = e(μn n0 + μp p0) (5.19)'),
                     row('resistivity', 'Resistivity 1/σ [Ω·cm]', E_app / st.J if st.J else np.nan, 1 / sig,
                         'ρ = 1/σ (5.20)'),
                     row('drift_share', 'Majority drift share of J at the center [%]',
                         100 * maj_drift / st.J if st.J else np.nan, 100.0, 'uniform doping: no diffusion')]
            ov['E'].append((x, np.full(d.N, E_app), 'V/L', 'dash'))
        return rows, ov
    # Graded doping: majority-carrier conduction, E(x) = E_bi(x) + J/sigma(x) with J = V / integral dx/sigma(x)
    ptype = p['doping_type'] == 'p'
    if structure == 1:
        N0, NL = p['N0'], p['NL']
        Nx = N0 + (NL - N0) * x / L
        dNdx = np.full(d.N, (NL - N0) / (L * UM2CM))
    else:
        Nx = p['N0'] * np.exp(-x / p['decay_um'])
        dNdx = -Nx / (p['decay_um'] * UM2CM)
    E_bi = drift_diffusion.induced_field(Nx, dNdx, d.Vt, p['doping_type'])
    sig = Q * (d.mup if ptype else d.mun) * Nx
    J_an = V / np.trapezoid(1.0 / sig, d.x)
    E_an = E_bi + J_an / sig
    model = '(5.42)' if V == 0 else '(5.42) + J/σ(x)'
    rows.append(row('J', 'Current density J [A/cm²]', st.J, J_an, 'J = V / ∫dx/σ(x)'))
    for frac in (0.1, 0.5, 0.9):
        k = int(np.argmin(np.abs(x - frac * L)))
        rows.append(row(f'E_{int(frac * 100)}', f'E at x = {x[k] - d.x0_um:.3g} µm [V/cm]', st.E[k], E_an[k], model))
    ov['E'].append((x, E_an, 'built-in field' if V == 0 else 'built-in + J/σ(x)', 'dash'))
    rows += [row('Jn_drift_max', 'max |J_n drift| [A/cm²]', np.max(np.abs(st.Jn_drift))),
             row('Jn_diff_max', 'max |J_n diffusion| [A/cm²]', np.max(np.abs(st.Jn_diff))),
             row('Jp_drift_max', 'max |J_p drift| [A/cm²]', np.max(np.abs(st.Jp_drift))),
             row('Jp_diff_max', 'max |J_p diffusion| [A/cm²]', np.max(np.abs(st.Jp_diff))),
             row('Jt_max', 'max |J_n + J_p| [A/cm²]', np.max(np.abs(st.Jt)), '0' if V == 0 else None,
                 'equilibrium (5.43)' if V == 0 else '')]
    return rows, ov
```

```python
###############################################################################
# Lab 4 Monte Carlo: particle picture of drift and diffusion
###############################################################################
# Electrons released at the origin with thermal velocities: free flight in the field between collisions, and a
# collision with probability dt/tau in each time step after which the velocity is redrawn from the thermal
# (Maxwell) distribution; tau = mu m*/q. No recombination, no velocity saturation (low fields).

M0 = 9.109e-31          # Electron rest mass [kg]
Q_SI = 1.602e-19        # Elementary charge [C]


def monte_carlo(E_Vcm, mu_cm2, t_max_s, frame_s, kT_eV=0.0259, m_eff=0.26 * M0, num_particles=20000,
                num_tracers=3, seed=None):
    """
    Electrons released at the origin with thermal velocities: free flight under the field between collisions,
    and a collision with probability dt/tau in each time step, after which the velocity is redrawn from the
    thermal (Maxwell) distribution. tau = mu m*/q; no recombination and no velocity saturation.

    input:
        float   E_Vcm           Electric field along +x [V/cm]
        float   mu_cm2          Low-field mobility [cm^2/V s] (sets tau = mu m*/q)
        float   t_max_s         Simulated time [s]
        float   frame_s         Time between stored frames [s]
        float   kT_eV           Thermal energy [eV]
        float   m_eff           Conductivity effective mass [kg] (Si electrons: 0.26 m0)
    output:
        SimpleNamespace         times [s], X, Y [m], VX [m/s] (frames x particles), TX, TY [m] (time steps x tracers), tau,
                                v_th, dt, n_sub, mfp, coll_per_particle
    """
    E_SI, mu_SI = E_Vcm * 1e2, mu_cm2 * 1e-4
    tau = mu_SI * m_eff / Q_SI                      # Mean time between collisions [s], mu = q tau/m*
    v_th = np.sqrt(kT_eV * Q_SI / m_eff)            # 1D thermal velocity [m/s], <v_x^2> = kT/m*
    a = -Q_SI * E_SI / m_eff                        # Acceleration of electrons [m/s^2]
    n_sub = int(np.ceil(frame_s / (tau / 50)))      # Sub-steps per frame, dt <= tau/50
    dt = frame_s / n_sub
    n_frm = int(round(t_max_s / frame_s))
    p_col = dt / tau                                # Collision probability per step

    rng = np.random.default_rng(seed)
    x, y = np.zeros(num_particles), np.zeros(num_particles)
    vx, vy = rng.normal(0, v_th, num_particles), rng.normal(0, v_th, num_particles)
    times = np.arange(n_frm + 1) * frame_s
    X, Y = np.zeros((n_frm + 1, num_particles)), np.zeros((n_frm + 1, num_particles))
    VX = np.zeros((n_frm + 1, num_particles))
    VX[0] = vx
    trace_x, trace_y = [np.zeros(num_tracers)], [np.zeros(num_tracers)]
    n_coll = 0
    for f in range(1, n_frm + 1):
        for _ in range(n_sub):
            x += vx * dt + 0.5 * a * dt ** 2
            y += vy * dt
            vx += a * dt
            hit = rng.random(num_particles) < p_col              # Scattering event
            n_hit = np.count_nonzero(hit)
            vx[hit] = rng.normal(0, v_th, n_hit)                 # Velocity randomized
            vy[hit] = rng.normal(0, v_th, n_hit)
            n_coll += n_hit
            trace_x.append(x[:num_tracers].copy())
            trace_y.append(y[:num_tracers].copy())
        X[f], Y[f], VX[f] = x, y, vx
    return SimpleNamespace(times=times, X=X, Y=Y, VX=VX, TX=np.array(trace_x), TY=np.array(trace_y), tau=tau, v_th=v_th,
                           dt=dt, n_sub=n_sub, mfp=np.sqrt(2) * v_th * tau,   # 2D mean free path estimate
                           coll_per_particle=n_coll / num_particles)


def drift_diffusion_packet(x, t, v_d, D):
    """Analytic packet of the drift-diffusion equation without recombination (Haynes-Shockley form):
    P(x, t) = exp[-(x - v_d t)^2 / 4Dt] / (4 pi D t)^{1/2}."""
    return np.exp(-(x - v_d * t) ** 2 / (4 * D * t)) / np.sqrt(4 * np.pi * D * t)


def run_montecarlo(E_field=2000.0, mun=1350.0, t_max_ps=30.0, frame_ps=0.3, T=300.0, num_particles=20000,
                   seed=None, nbins=40, verbose=True, rows=None, cols=None):
    """
    Monte Carlo drift and diffusion (monte_carlo above) and comparison with the drift-diffusion picture:
    <x> = -mu E t, sigma^2 = 2Dt, D/mu = kT/q, and the particle flux F = sum v_x / (N dx) compared with its drift
    term -mu E P(x) and diffusion term -D dP/dx (P = density per electron; electron current J_n = -q F N).

    input:
        float   E_field         Electric field along +x [V/cm]
        float   mun             Electron mobility [cm^2/V s] (sets the collision time tau = mu m*/q)
        float   t_max_ps        Simulated time [ps]
        float   frame_ps        Time between animation frames [ps]
        float   T               Temperature [K] (thermal velocity, kT = 0.0259 eV x T/300)
        int     num_particles   Number of simulated electrons
        int     nbins           Number of bins of the density and flux profiles
    output:
        SimulationResult        res, res.data: times, X, Y, TX, TY, x_mean, var_x, var_y, flux profiles, ...
    """
    if E_field < 0 or mun <= 0 or t_max_ps <= 0 or frame_ps <= 0 or frame_ps > t_max_ps:
        raise ValueError("Use E >= 0, mobility > 0 and 0 < frame interval <= simulation time.")
    say = print if verbose else (lambda *a, **k: None)
    kT = statistics.thermal_voltage(T)                        # [eV] (kT/q = 0.0259 V x T/300)
    say('(0) Model')
    say(f'electrons \t = {num_particles}, m* = 0.26 m0 (Si), T = {T:g} K, kT = {kT:.4f} eV')
    say(f'field, mobility = {E_field:g} V/cm, {mun:g} cm^2/V s')
    say('\n(1) Free flight + scattering (collision probability dt/tau per time step) ...')
    mc = monte_carlo(E_field, mun, t_max_ps * 1e-12, frame_ps * 1e-12, kT_eV=kT, num_particles=num_particles,
                     seed=seed)
    times, X, Y = mc.times, mc.X, mc.Y
    say(f'\ttau = {mc.tau * 1e12:.3f} ps, time step = {mc.dt * 1e15:.2f} fs, {len(times) - 1} frames, '
        f'{mc.coll_per_particle:.0f} collisions per electron')
    say('Done!!')

    # Drift-diffusion picture (SI units) and values extracted from the particle cloud
    D_th = drift_diffusion.einstein(mun, kT) * 1e-4           # Einstein relation D = mu kT/q [m^2/s]
    vd_th = -mun * 1e-4 * E_field * 1e2                       # Electrons drift against E [m/s]
    x_mean, var_x, var_y = X.mean(axis=1), X.var(axis=1), Y.var(axis=1)
    late = times > 5 * mc.tau                                 # Diffusive regime, t >> tau
    vd_fit = np.polyfit(times[late], x_mean[late], 1)[0]
    D_fit = np.polyfit(times[late], var_y[late], 1)[0] / 2    # Transverse spread (no drift along y)
    Dx_fit = np.polyfit(times[late], var_x[late], 1)[0] / 2
    mu_fit = -vd_fit / (E_field * 1e2) * 1e4 if E_field > 0 else np.nan

    # Particle flux per electron in each bin: from the velocities (MC) and from the drift-diffusion terms.
    # The bins follow the cloud, -mu E t +- 4 (2Dt)^1/2 (frame 0: all electrons at the origin, no profile).
    centers, P, F_mc, F_diff = (np.full((len(times), nbins), np.nan) for _ in range(4))
    for k in range(1, len(times)):
        half = 4 * np.sqrt(2 * D_th * times[k])
        edges = np.linspace(vd_th * times[k] - half, vd_th * times[k] + half, nbins + 1)
        centers[k], dxb = 0.5 * (edges[1:] + edges[:-1]), edges[1] - edges[0]
        P[k] = np.histogram(X[k], edges)[0] / (num_particles * dxb)                       # [1/m]
        F_mc[k] = np.histogram(X[k], edges, weights=mc.VX[k])[0] / (num_particles * dxb)   # [1/s]
        F_diff[k] = -D_th * np.gradient(P[k], centers[k])     # Diffusion term (Fick's law) [1/s]
    F_drift = vd_th * P                                       # Drift term (electrons move against E) [1/s]
    core = P[-1] > 0.05 * P[-1].max()                         # Bins inside the cloud at the last frame
    F_sum = F_drift[-1] + F_diff[-1]
    flux_resid = np.sqrt(np.mean((F_mc[-1][core] - F_sum[core]) ** 2)) / np.max(np.abs(F_mc[-1][core]))
    table = [row('tau', 'Mean time between collisions τ = μm*/q [ps]', mc.tau * 1e12, None, 'input mobility'),
             row('mfp', 'Mean free path ≈ √2 v_th τ [nm]', mc.mfp * 1e9, None, 'v_th = (kT/m*)^1/2'),
             row('mu', 'Mobility from ⟨x⟩ = −μEt [cm²/V s]', mu_fit, mun if E_field > 0 else None,
                 'drift velocity v_d = −μE (5.7)' if E_field > 0 else 'E = 0: no drift'),
             row('D_y', 'Diffusion coefficient from σ_y² = 2Dt [cm²/s]', D_fit * 1e4, D_th * 1e4,
                 'D = μkT/q (Einstein relation (5.47))'),
             row('D_x', 'Diffusion coefficient from σ_x² = 2Dt [cm²/s]', Dx_fit * 1e4, D_th * 1e4,
                 'same D along the field'),
             row('D_over_mu', 'D/μ from the cloud [V]', D_fit / (mu_fit * 1e-4) if E_field > 0 else None, kT,
                 'kT/q (Einstein relation)'),
             row('flux_resid', 'Flux: rms(MC − drift − diffusion)/max|MC|, last frame', flux_resid, '≈ noise',
                 'J = drift + diffusion (5.35)')]
    data = dict(times=times, X=X, Y=Y, TX=mc.TX, TY=mc.TY, x_mean=x_mean, var_x=var_x, var_y=var_y, vd_th=vd_th,
                flux_x=centers, flux_mc=F_mc, flux_drift=F_drift, flux_diff=F_diff,
                D_th=D_th, tau=mc.tau, n_sub=mc.n_sub, mfp=mc.mfp, coll_per_particle=mc.coll_per_particle,
                E=E_field, mu=mun, T=T)
    params = dict(analysis='montecarlo', E_field=E_field, mun=mun, t_max_ps=t_max_ps, frame_ps=frame_ps, T=T,
                  num_particles=num_particles, nbins=nbins)
    res = SimulationResult(params, None, [], table, data)
    if verbose:
        print('\n(2) Comparison with the drift-diffusion picture (late times, t > 5τ)')
        print_table(select_rows(res.table, rows))
    return res


###############################################################################
# Lab 4 Runner and Cases
###############################################################################

def run_drift(structure=0, doping_type='n', N=1e16, N0=1e16, NL=2e15, decay_um=1.0, length_um=10.0, V=0.0,
              material='Si', T=300.0, npts=None, verbose=True, rows=None):
    """
    Drift and diffusion currents of a uniform bar (structure 0) or a graded bar (1 linear, 2 exponential).

    input:
        float   V               Bias of the left contact [V] (right contact grounded)
    output:
        SimulationResult        res (res.data['state'], res.data['overlays'])
    """
    params = dict(structure=structure, doping_type=doping_type, N=N, N0=N0, NL=NL, decay_um=decay_um,
                  length_um=length_um, V=V, material=material, T=T, npts=npts, analysis='point', G=0.0, inj=0.0)
    say = print if verbose else (lambda *a, **k: None)
    dev = build_lab4_device(structure, doping_type, N, N0, NL, decay_um, length_um, material, T, npts)
    print_device(dev, STRUCTURES[structure], say)
    say('model \t\t = quasi-neutral densities, n = Nd(x) or p = Na(x) (Neamen (5.39)); uniform current')
    say('\n(1) Evaluating the drift and diffusion currents ...')
    s = DriftSolver(dev)
    st = s.solve(V)
    say(f'\tV = {V:+.3f} V    : J = {st.J:+.4e} A/cm^2')
    table, ov = analyze_drift(st, s, params)
    res = SimulationResult(params, dev, [st], table, dict(state=st, overlays=ov))
    say('Done!!')
    if verbose:
        print('\n(2) Comparison with analytic models')
        print_table(select_rows(res.table, rows))
    return res


def case_kwargs(p):
    """Translate case parameters into runner arguments (doping profile, automatic grid)."""
    kw = dict(p)
    if 'profile' in kw:                                       # Graded doping: exponential or linear
        profile = kw.pop('profile')
        if profile not in ('exponential', 'linear'):
            raise ValueError("Use profile = 'exponential' or 'linear'.")
        kw['structure'] = 2 if profile == 'exponential' else 1
    if kw.get('npts') == 0:
        kw['npts'] = None                                     # Automatic grid
    return kw


CASES = {
    '4-MC': dict(title='Monte Carlo: scattered electrons drift and diffuse', runner=run_montecarlo,
                 summary='Free flight + random scattering: the cloud drifts with −μE and spreads as 2Dt; the particle '
                         'current equals the drift term plus the diffusion term; D/μ = kT/q (Einstein relation).',
                 params=dict(E_field=2000.0, mun=1350.0, t_max_ps=30.0, frame_ps=0.3, T=300.0),
                 ranges=dict(mun=(200.0, 10000.0, 50.0)),
                 rows=None, panels=None, schematic=lambda v: schematic_montecarlo()),
    '4-A': dict(title='Uniform bar under bias', runner=run_drift, fixed=dict(structure=0),
                summary='Drift only: E = V/L and J = σE in a uniformly doped bar.',
                params=dict(doping_type='n', N=1e16, length_um=10.0, V=0.1, material='Si', T=300.0, npts=201),
                ranges=dict(length_um=(1.0, 50.0, 0.5), V=(-2.0, 2.0, 0.01)),
                rows=['E_center', 'J', 'sigma', 'resistivity', 'drift_share'],
                panels=['band', 'E', 'current_n', 'current_p', 'density'],
                schematic=lambda v: schematic_bar(v.get('doping_type', 'n'))),
    '4-B': dict(title='Graded doping (exponential or linear)', runner=run_drift,
                summary='Induced field E = ∓(kT/e)(1/N)(dN/dx) of a quasi-neutral graded bar; under bias '
                        'E(x) = Ebi(x) + J/σ(x).',
                params=dict(profile='exponential', doping_type='n', N0=1e16, NL=2e15, decay_um=1.0, length_um=3.0,
                            V=0.0, material='Si', T=300.0, npts=401),
                ranges=dict(decay_um=(0.05, 500.0, 0.05), length_um=(0.5, 10.0, 0.1), V=(-2.0, 2.0, 0.01)),
                rows=['J', 'E_10', 'E_50', 'E_90', 'Jn_drift_max', 'Jn_diff_max', 'Jp_drift_max', 'Jp_diff_max',
                      'Jt_max'],
                panels=['band', 'E', 'current_n', 'current_p', 'density'],
                schematic=lambda v: schematic_graded(v.get('doping_type', 'n'), v.get('profile', 'exponential'))),
}
```

```python
import contextlib
import html
import io
import time
import traceback

import ipywidgets as widgets
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from IPython.display import HTML, display

###############################################################################
# Plot Settings and Figure Builder
###############################################################################

C_E, C_H, C_T, C_A = '#1f77b4', '#d62728', '#222222', '#7f7f7f'   # Electrons, holes, total, analytic
FIG_WIDTH = 1150
PANEL_HEIGHT = 250         # Height of one panel [px]
ROW_GAP = 105              # Vertical gap between panel rows (axis title + panel title) [px]
NMAX_PLOT = 600            # Maximum points per curve sent to the browser
NMAX_FRAME = 300           # Maximum points per curve in each bias-sweep frame


def _sub(s):
    """Subscripts in Plotly text: 'E_g', 'μ_n', 'N_{a}' -> 'E<sub>g</sub>', 'μ<sub>n</sub>', 'N<sub>a</sub>'."""
    if not isinstance(s, str):
        return s
    return re.sub(r'(?<=[^\W\d_])_(\{[^}]*\}|[A-Za-z0-9]+)', lambda m: f"<sub>{m.group(1).strip('{}')}</sub>", s)


def _trace(x, y, name, color=None, dash='solid', mode='lines', width=2):
    """Trace specification."""
    return dict(x=np.asarray(x, dtype=float), y=np.asarray(y, dtype=float), name=name, color=color,
                dash=dash, mode=mode, width=width)


def _panel(title, xlabel, ylabel, traces, ylog=False, xrange=None, yrange=None, vlines=(), xlog=False, decades=None):
    """Panel specification; decades limits a log axis to that many decades below the maximum."""
    return dict(title=title, xlabel=xlabel, ylabel=ylabel, traces=traces, ylog=ylog,
                xrange=xrange, yrange=yrange, vlines=list(vlines), xlog=xlog, decades=decades)


def _thin(x, y, ylog, nmax=NMAX_PLOT):
    """Reduce points along the arc length of the curve, keeping steep regions dense."""
    if len(x) <= nmax:
        return x, y
    yy = np.log10(np.where(y > 0, y, np.nan)) if ylog else y.astype(float)
    ok = np.isfinite(yy)
    if ok.sum() < 2:
        idx = np.linspace(0, len(x) - 1, nmax).astype(int)
        return x[idx], y[idx]
    yy = np.where(ok, yy, np.nanmin(yy[ok]))
    xs = (x - x.min()) / max(np.ptp(x), 1e-30)
    ys = (yy - yy.min()) / max(np.ptp(yy), 1e-30)
    s = np.concatenate([[0.0], np.cumsum(np.hypot(np.diff(xs), np.diff(ys)))])
    idx = np.unique(np.concatenate([np.searchsorted(s, np.linspace(0, s[-1], nmax)),
                                    np.linspace(0, len(x) - 1, nmax // 4).astype(int), [len(x) - 1]]))
    idx = idx[idx < len(x)]
    return x[idx], y[idx]


def _scatter(t, ylog, legend, nmax=NMAX_PLOT, xrange=None):
    ylog = ylog or t.get('thin_log', False)
    x, y = t['x'], (np.where(t['y'] > 0, t['y'], np.nan) if ylog else t['y'])
    if xrange is not None and len(x) > 2:
        # Keep the zoomed window (plus one point on each side) before thinning.
        inside = np.flatnonzero((x >= xrange[0]) & (x <= xrange[1]))
        if inside.size:
            sl = slice(max(inside[0] - 1, 0), min(inside[-1] + 2, len(x)))
            x, y = x[sl], y[sl]
    x, y = _thin(x, y, ylog, t.get('nmax', nmax))
    extra = {k: t[k] for k in ('visible', 'legendgroup', 'showlegend', 'fill', 'fillcolor', 'opacity') if k in t}
    line = dict(color=t['color'], dash=t['dash'], width=t['width'])
    if 'shape' in t:
        line['shape'] = t['shape']
    return go.Scatter(x=x.astype(np.float32), y=y.astype(np.float32), name=_sub(t['name']), mode=t['mode'], line=line,
                      marker=dict(size=t.get('marker_size', 7), color=t['color']), legend=legend, **extra)


def _log_range(values, decades=30):
    v = np.concatenate([np.asarray(a, dtype=float).ravel() for a in values])
    v = v[np.isfinite(v) & (v > 0)]
    if v.size == 0:
        return None
    hi = np.log10(v.max())
    lo = max(np.log10(v.min()), hi - decades)
    if hi - lo < 1.0:                                   # Nearly constant: two decades around the values
        c = 0.5 * (hi + lo)
        return [c - 1.0, c + 1.0]
    pad = 0.05 * max(hi - lo, 1.0)
    return [lo - pad, hi + pad]


def _lin_range(values, symmetric=False):
    v = np.concatenate([np.asarray(a, dtype=float).ravel() for a in values])
    v = v[np.isfinite(v)]
    if v.size == 0:
        return None
    lo, hi = (-np.max(np.abs(v)), np.max(np.abs(v))) if symmetric else (v.min(), v.max())
    if hi - lo <= 1e-6 * max(abs(lo), abs(hi)):          # Nearly constant: include zero
        lo, hi = min(lo, 0.0), max(hi, 0.0)
        if hi == lo:
            return [-1.0, 1.0]
    pad = 0.05 * (hi - lo)
    return [lo - pad, hi + pad]


def build_figure(panels, title, ncols=2, bottom=60):
    """
    Build one Plotly figure with one legend per panel (placed to the right of the panel).

    output:
        go.Figure   fig
        list        trace_ids   Trace indices of each panel
    """
    nrows = (len(panels) + ncols - 1) // ncols
    plot_height = nrows * PANEL_HEIGHT + (nrows - 1) * ROW_GAP
    fig = make_subplots(rows=nrows, cols=ncols, subplot_titles=[_sub(pn['title']) if pn else ' ' for pn in panels],
                        horizontal_spacing=0.24, vertical_spacing=ROW_GAP / plot_height if nrows > 1 else 0.0)
    trace_ids = []
    for k, pn in enumerate(panels):
        r, c = k // ncols + 1, k % ncols + 1
        if pn is None:                              # Blank position (keeps the following panels aligned)
            fig.update_xaxes(visible=False, row=r, col=c)
            fig.update_yaxes(visible=False, row=r, col=c)
            trace_ids.append([])
            continue
        legend = 'legend' if k == 0 else f'legend{k + 1}'
        ids = []
        for t in pn['traces']:
            fig.add_trace(_scatter(t, pn['ylog'], legend, xrange=pn['xrange']), row=r, col=c)
            ids.append(len(fig.data) - 1)
        trace_ids.append(ids)
        sub = fig.get_subplot(r, c)
        fig.update_layout({legend: dict(x=sub.xaxis.domain[1] + 0.01, y=sub.yaxis.domain[1], xanchor='left',
                                        yanchor='top', font=dict(size=10), bgcolor='rgba(255,255,255,0.85)')})
        yrange = pn['yrange']
        if yrange is None:                          # Limit log axes; avoid zoomed-in axes on constant curves
            vals = [t['y'] for t in pn['traces']]
            if pn['ylog']:
                auto = _log_range(vals, decades=pn.get('decades') or 30)
                if auto is not None and (pn.get('decades') or auto[1] - auto[0] <= 2.0):
                    yrange = auto
            elif vals:
                auto = _lin_range(vals)
                v = np.concatenate([np.asarray(a, dtype=float).ravel() for a in vals])
                v = v[np.isfinite(v)]
                if v.size and np.ptp(v) <= 1e-6 * np.max(np.abs(v)):
                    yrange = auto
        fig.update_xaxes(title_text=_sub(pn['xlabel']), range=pn['xrange'], row=r, col=c)
        fig.update_yaxes(title_text=_sub(pn['ylabel']), range=yrange, row=r, col=c)
        if 'cm⁻³' in pn['ylabel'] or 'A/cm²' in pn['ylabel']:   # 1e16 -> 10^16 instead of SI prefixes (T, p)
            fig.update_yaxes(exponentformat='power', row=r, col=c)
        if pn['ylog']:
            fig.update_yaxes(type='log', exponentformat='power', row=r, col=c)
        if pn.get('xlog'):
            fig.update_xaxes(type='log', exponentformat='power', row=r, col=c)
        for xv in pn['vlines']:
            fig.add_vline(x=xv, line=dict(color='gray', dash='dot', width=1), row=r, col=c)
    fig.update_layout(width=FIG_WIDTH, height=plot_height + 90 + bottom, template='plotly_white', title=_sub(title),
                      margin=dict(l=80, r=200, t=90, b=bottom))
    fig.update_annotations(font_size=13)
    return fig, trace_ids


FIGURE_SINK = None          # Set by the launcher to collect figures instead of displaying them


def show_figure(fig):
    if FIGURE_SINK is not None:
        FIGURE_SINK(fig)
    else:
        display(HTML(fig.to_html(full_html=False, include_plotlyjs='inline', auto_play=False)))


PLOTLY_JS_SENT = {'done': False}     # The launcher embeds plotly.js in its first figure only


def figure_light_html(fig):
    """Figure HTML without plotly.js: draws with the plotly.js already in the page, or loads it from the
    plotly CDN if the page does not have it (for example after reloading the window)."""
    html_fig = fig.to_html(full_html=False, include_plotlyjs=False, auto_play=False)
    cdn = re.search(r'src="([^"]+)"', go.Figure().to_html(full_html=False, include_plotlyjs='cdn')).group(1)
    i0 = html_fig.index('<script>') + len('<script>')
    i1 = html_fig.rindex('</script>')
    wrapped = ("(function () { var draw = function () {" + html_fig[i0:i1] + "};"
               " if (window.Plotly) { draw(); return; }"
               " var s = document.createElement('script'); s.src = '" + cdn + "'; s.onload = draw;"
               " document.head.appendChild(s); })();")
    return html_fig[:i0] + wrapped + html_fig[i1:]


def launcher_figure_html(fig):
    """
    Figure HTML for the launcher: (html shown now, html kept for showing again).
    The first figure after launch() embeds plotly.js (about 4.7 MB); later figures carry only their data.
    """
    light = figure_light_html(fig)
    if not PLOTLY_JS_SENT['done']:
        PLOTLY_JS_SENT['done'] = True
        return fig.to_html(full_html=False, include_plotlyjs='inline', auto_play=False), light
    return light, light


def _overlay_traces(ov, key, x0=0.0):
    """Analytic overlays (computed on the device grid) shifted to the plotting origin x0."""
    return [_trace(np.asarray(x) - x0, y, name, C_A, dash, width=1.5) for x, y, name, dash in ov.get(key, [])]


def _rgba(hex_color, alpha):
    h = hex_color.lstrip('#')
    return f'rgba({int(h[0:2], 16)},{int(h[2:4], 16)},{int(h[4:6], 16)},{alpha})'


###############################################################################
# Device Schematics (inline SVG shown next to the case parameters)
###############################################################################

# Drawings with symbols only: the left contact is the bias terminal V, the right contact is grounded.

SCH_W, SCH_H = 480, 225
SCH_FONT = "font-family='Helvetica, Arial, sans-serif'"
SCH_COLORS = dict(p='#d6e4fb', n='#fbd5d5', p_dark='#3b6fd8', n_dark='#d24b4b', edge='#555555', wire='#333333',
                  contact='#444444', light='#e08a00', text='#222222', note='#666666')     # p: blue, n: red
_X0, _X1, _Y0, _Y1 = 110, 370, 95, 140          # Device block [px]


def _markup(s, size):
    """'N_{a}' and 'e^{-s/λ}' -> SVG tspans with subscripts and superscripts."""
    small = round(0.72 * size)
    s = re.sub(r'_\{([^}]*)\}', lambda m: f"<tspan baseline-shift='sub' font-size='{small}'>{m.group(1)}</tspan>", s)
    return re.sub(r'\^\{([^}]*)\}', lambda m: f"<tspan baseline-shift='super' font-size='{small}'>{m.group(1)}</tspan>", s)


def _text(x, y, s, size=13, anchor='middle', color=None, weight='normal', italic=False, bg=False):
    color = color or SCH_COLORS['text']
    style = " font-style='italic'" if italic else ''
    out = ''
    if bg:                                       # White box behind the label (hides lines underneath)
        w = 0.62 * size * len(re.sub(r'[_^]\{([^}]*)\}', r'\1', s)) + 6
        x_left = x - w / 2 if anchor == 'middle' else (x - 3 if anchor == 'start' else x - w + 3)
        out += f"<rect x='{x_left:.1f}' y='{y - size + 1:.1f}' width='{w:.1f}' height='{size + 4}' fill='white'/>"
    return out + (f"<text x='{x}' y='{y}' text-anchor='{anchor}' font-size='{size}' fill='{color}' "
                  f"font-weight='{weight}'{style} {SCH_FONT}>{_markup(s, size)}</text>")


def _line(x0, y0, x1, y1, color=None, width=1.5, dash=None):
    d = f" stroke-dasharray='{dash}'" if dash else ''
    return (f"<line x1='{x0}' y1='{y0}' x2='{x1}' y2='{y1}' stroke='{color or SCH_COLORS['wire']}' "
            f"stroke-width='{width}'{d}/>")


def _arrowhead(x, y, angle, size=7, color=None):
    """Filled arrowhead with its tip at (x, y), pointing along angle [rad]."""
    c, s = np.cos(angle), np.sin(angle)
    pts = [(x, y), (x - size * c + 0.5 * size * s, y - size * s - 0.5 * size * c),
           (x - size * c - 0.5 * size * s, y - size * s + 0.5 * size * c)]
    return (f"<polygon points='{' '.join(f'{px:.1f},{py:.1f}' for px, py in pts)}' "
            f"fill='{color or SCH_COLORS['wire']}'/>")


def _dimension(x0, x1, y, label):
    """Double-headed length arrow with a label underneath."""
    return (_line(x0 + 1, y, x1 - 1, y, width=1) + _arrowhead(x0, y, np.pi) + _arrowhead(x1, y, 0.0)
            + _line(x0, y - 6, x0, y + 6, width=1) + _line(x1, y - 6, x1, y + 6, width=1)
            + _text((x0 + x1) / 2, y + 18, label, bg=True))


def _wavy_arrow(x0, y0, x1, y1, color=None, waves=3, amp=4.0):
    """Light ray: sine-wave path from (x0, y0) to (x1, y1) with an arrowhead."""
    color = color or SCH_COLORS['light']
    L = np.hypot(x1 - x0, y1 - y0)
    ux, uy = (x1 - x0) / L, (y1 - y0) / L
    t = np.linspace(0, L - 8, 60)
    off = amp * np.sin(2 * np.pi * waves * t / (L - 8))
    xs, ys = x0 + ux * t - uy * off, y0 + uy * t + ux * off
    path = 'M ' + ' L '.join(f'{a:.1f},{b:.1f}' for a, b in zip(xs, ys))
    return (f"<path d='{path}' fill='none' stroke='{color}' stroke-width='1.8'/>"
            + _arrowhead(x1, y1, np.arctan2(uy, ux), size=8, color=color))


def _ground(x, y):
    return (_line(x, y, x + 16, y) + _line(x + 16, y, x + 16, y + 26)
            + _line(x + 4, y + 26, x + 28, y + 26, width=2) + _line(x + 8, y + 31, x + 24, y + 31, width=2)
            + _line(x + 12, y + 36, x + 20, y + 36, width=2) + _text(x + 16, y + 52, 'ground', size=11,
                                                                       color=SCH_COLORS['note']))


def _contacts():
    """Thin ohmic-contact plates at both ends of the device."""
    return (f"<rect x='{_X0 - 6}' y='{_Y0 - 4}' width='6' height='{_Y1 - _Y0 + 8}' fill='{SCH_COLORS['contact']}'/>"
            f"<rect x='{_X1}' y='{_Y0 - 4}' width='6' height='{_Y1 - _Y0 + 8}' fill='{SCH_COLORS['contact']}'/>")


def _circuit():
    """Bias terminal V on the left contact and ground on the right contact."""
    ym = (_Y0 + _Y1) / 2
    return (_line(46, ym, _X0 - 6, ym)
            + f"<circle cx='40' cy='{ym}' r='5.5' fill='white' stroke='{SCH_COLORS['wire']}' stroke-width='1.5'/>"
            + _text(40, ym - 14, 'V', weight='bold') + _ground(_X1 + 6, ym))


def _block(x0, x1, fill, label_lines, y0=_Y0, y1=_Y1):
    out = f"<rect x='{x0}' y='{y0}' width='{x1 - x0}' height='{y1 - y0}' fill='{fill}' stroke='{SCH_COLORS['edge']}'/>"
    yc = (y0 + y1) / 2
    for k, s in enumerate(label_lines):
        out += _text((x0 + x1) / 2, yc + 5 + (k - (len(label_lines) - 1) / 2) * 16, s,
                     weight='bold' if k == 0 else 'normal')
    return out


def _origin(x, label='x = 0', through=False):
    """Dashed mark of x = 0; through=True crosses the device (junction), otherwise it starts below the block."""
    y_top = _Y0 - 8 if through else _Y1
    return _line(x, y_top, x, 196, color='#888888', width=1, dash='4,3') + _text(x, 210, label, size=12, bg=True)


def _svg(body, gradient='', top=0):
    """White card with the drawing, shifted up by 'top' px when nothing is drawn above the device."""
    h = SCH_H - top
    return (f"<svg xmlns='http://www.w3.org/2000/svg' width='{SCH_W}' height='{h}' viewBox='0 0 {SCH_W} {h}'>"
            f"<defs>{gradient}</defs>"
            f"<rect x='0.5' y='0.5' width='{SCH_W - 1}' height='{h - 1}' rx='8' fill='white' stroke='#d0d0d0'/>"
            f"<g transform='translate(0,{-top})'>{body}</g></svg>")


def _bar(dtype, label2='N'):
    """Uniformly doped bar: n-type red, p-type blue."""
    return _block(_X0, _X1, SCH_COLORS[dtype], [f'{dtype}-type', label2])


def _pn(biased):
    xm = (_X0 + _X1) / 2
    body = (_block(_X0, xm, SCH_COLORS['p'], ['p-type', 'N_{a}']) + _block(xm, _X1, SCH_COLORS['n'], ['n-type', 'N_{d}'])
            + _contacts() + (_circuit() if biased else '') + _origin(xm, through=True)
            + _dimension(_X0, xm, 160, 'L_{p}') + _dimension(xm, _X1, 160, 'L_{n}'))
    return _svg(body, top=45)


def _glow(gid, width_frac=0.35):
    """Orange shading that fades from the left end of the bar (excess carriers near the left end)."""
    grad = (f"<linearGradient id='{gid}' x1='0' x2='1' y1='0' y2='0'>"
            f"<stop offset='0' stop-color='{SCH_COLORS['light']}' stop-opacity='0.75'/>"
            f"<stop offset='1' stop-color='{SCH_COLORS['light']}' stop-opacity='0'/></linearGradient>")
    rect = (f"<rect x='{_X0}' y='{_Y0}' width='{width_frac * (_X1 - _X0):.0f}' height='{_Y1 - _Y0}' "
            f"fill='url(#{gid})'/>")
    return grad, rect


def _open_bar(dtype, label2='N'):
    """Bar cut at both ends (the sample continues: no contacts)."""
    fill, edge = SCH_COLORS[dtype], SCH_COLORS['edge']
    zig = lambda x, s: ' '.join(f'{x + s * (5 if k % 2 else 0)},{_Y0 + k * (_Y1 - _Y0) / 6:.1f}' for k in range(7))
    body = (f"<rect x='{_X0}' y='{_Y0}' width='{_X1 - _X0}' height='{_Y1 - _Y0}' fill='{fill}'/>"
            + _line(_X0, _Y0, _X1, _Y0, edge, 1) + _line(_X0, _Y1, _X1, _Y1, edge, 1)
            + f"<polyline points='{zig(_X0, -1)}' fill='none' stroke='{edge}' stroke-width='1.2'/>"
            + f"<polyline points='{zig(_X1, 1)}' fill='none' stroke='{edge}' stroke-width='1.2'/>")
    ym = (_Y0 + _Y1) / 2 + 5
    body += (_text(_X0 - 26, ym, '⋯', size=18, color=SCH_COLORS['note'])
             + _text(_X1 + 26, ym, '⋯', size=18, color=SCH_COLORS['note']))
    yc = (_Y0 + _Y1) / 2
    for k, s in enumerate([f'{dtype}-type', label2]):
        body += _text((_X0 + _X1) / 2, yc + 5 + (k - 0.5) * 16, s, weight='bold' if k == 0 else 'normal')
    return body


_XM = (_X0 + _X1) / 2


###############################################################################
# Lab 4 Panels and Figures
###############################################################################

def band_panel(st, title=None):
    title = title or f'Energy band diagram (E_g = {st.dev.Eg:g} eV)'
    return _panel(title, 'x [µm]', 'Energy [eV]', [
        _trace(st.xc, st.Ec, 'E_c', C_T), _trace(st.xc, st.Ev, 'E_v', C_T),
        _trace(st.xc, st.Ei, 'E_Fi', C_A, 'dot', width=1.5),
        _trace(st.xc, st.Fn, 'E_Fn', C_E, 'dash'), _trace(st.xc, st.Fp, 'E_Fp', C_H, 'dash')])


def density_panel(st, title='Carrier and doping concentrations'):
    d = st.dev
    return _panel(title, 'x [µm]', 'Concentration [cm⁻³]', [
        _trace(st.xc, st.n, 'n', C_E), _trace(st.xc, st.p, 'p', C_H),
        _trace(st.xc, d.Nd, 'N_d', C_E, 'dot', width=1.5), _trace(st.xc, d.Na, 'N_a', C_H, 'dot', width=1.5)], ylog=True)


def field_panel(st, ov, xr=None, title='Electric field'):
    return _panel(title, 'x [µm]', 'E [V/cm]', [_trace(st.xc, st.E, 'simulation', C_T)] + _overlay_traces(ov, 'E', st.dev.x0_um),
                  xrange=xr)


def carrier_current_panel(st, carrier):
    """Drift, diffusion and their sum for one carrier, on its own axis. Each component is drawn twice: signed
    (linear view, visible) and as |J| (log view, hidden; negative parts pale); the figure buttons switch views."""
    if carrier == 'n':
        drift, diff, color, name, title = st.Jn_drift, st.Jn_diff, C_E, 'J_n', 'Electron current'
    else:
        drift, diff, color, name, title = st.Jp_drift, st.Jp_diff, C_H, 'J_p', 'Hole current'
    comps = [(drift + diff, f'{name} (total)', _rgba(color, 0.35), 'solid', 6),
             (drift, 'drift', color, 'solid', 2), (diff, 'diffusion', color, 'dash', 2)]
    traces = []
    for y, label, col, dash, width in comps:
        lin = _trace(st.xhc, y, label, col, dash, width=width)
        lin['group'] = 'lin'
        pos = _trace(st.xhc, np.where(y > 0, y, np.nan), label, col, dash, width=width)
        neg = _trace(st.xhc, np.where(y < 0, -y, np.nan), label, _rgba(color, 0.3) if width < 5 else _rgba(color, 0.15),
                     dash, width=width)
        for t in (pos, neg):
            t.update(group='log', visible=False, legendgroup=f'{name} {label}', thin_log=True)
        neg['showlegend'] = False
        traces += [lin, pos, neg]
    return _panel(title, 'x [µm]', 'J [A/cm²]', traces)


def _add_current_toggle(fig, panels, trace_ids):
    """Buttons that switch the carrier-current panels between signed J (linear) and |J| (log)."""
    keys = [k for k, pn in panels if pn is not None and pn.get('current_toggle')]
    if not keys:
        return
    lin_vis, log_vis = [True] * len(fig.data), [True] * len(fig.data)
    lin_lay, log_lay, last = {}, {}, None
    for k, (key, pn) in enumerate(panels):
        if pn is None:
            continue
        for t, i in zip(pn['traces'], trace_ids[k]):
            if t.get('group') == 'lin':
                log_vis[i] = False
            elif t.get('group') == 'log':
                lin_vis[i] = False
        if pn.get('current_toggle'):
            sub = fig.get_subplot(k // 2 + 1, k % 2 + 1)
            last = sub
            ya = sub.yaxis.plotly_name
            lin_lay.update({f'{ya}.type': 'linear', f'{ya}.title.text': 'J [A/cm²]', f'{ya}.autorange': True})
            log_lay.update({f'{ya}.type': 'log', f'{ya}.title.text': '|J| [A/cm²] (pale: J < 0)',
                            f'{ya}.autorange': True, f'{ya}.exponentformat': 'power'})
    # Buttons below the legend, to the right of the last current panel
    fig.update_layout(updatemenus=list(fig.layout.updatemenus or ()) + [dict(
        type='buttons', direction='down', showactive=True, active=0, x=last.xaxis.domain[1] + 0.01, xanchor='left',
        y=last.yaxis.domain[0], yanchor='bottom', pad=dict(l=4), font=dict(size=11),
        buttons=[dict(label='Linear', method='update', args=[{'visible': lin_vis}, lin_lay]),
                 dict(label='Log |J|', method='update', args=[{'visible': log_vis}, log_lay])])])


def _run_title(res):
    p, d = res.params, res.device
    title = f"[{res.case}] " if res.case else ''
    title += f"{STRUCTURES[p['structure']]} | {d.material.name}, T = {d.T:g} K"
    if p['structure'] == 3:
        title += f" | N_a = {p['Na']:.3g}, N_d = {p['Nd']:.3g} cm⁻³"
    return title


def plot_montecarlo(res, panels=None, num_plot=800):
    """Monte Carlo drift-diffusion demonstration: particle cloud, distribution along the field, mean position
    against time and the particle flux split into drift and diffusion terms; one animation frame per stored time
    (slider and Play)."""
    d = res.data
    nm, ps = 1e9, 1e12
    times, X, Y, TX, TY = d['times'], d['X'], d['Y'], d['TX'], d['TY']
    vd, D, t_max = d['vd_th'], d['D_th'], d['times'][-1]
    sig_max = np.sqrt(2 * D * t_max)
    x_lo, x_hi = min(0.0, vd * t_max) * nm - 3.5 * sig_max * nm, max(0.0, vd * t_max) * nm + 3.5 * sig_max * nm
    y_lim = 3.5 * sig_max * nm
    bins = np.linspace(x_lo, x_hi, 201)                       # 200 bins: at t = 0 the cloud is a narrow spike
    x_grid = np.linspace(x_lo, x_hi, 300)
    t_ref = max(0.3 * t_max, 5 * d['tau'])
    p_max = 1.3 * drift_diffusion_packet(0.0, t_ref, 0.0, D) / nm
    stride = max(1, (TX.shape[0] - 1) // 400)                 # Tracer points sent to the browser
    tracer_colors = ['#2ca02c', '#ff7f0e', '#9467bd']
    keys = ['cloud', 'hist', 'mean', 'flux'] if panels is None else panels
    # Mean position: the y axis spans at least +-sigma(t_max)/4, so the statistical scatter sigma/N^1/2 of <x> is
    # seen on the scale of the cloud (without a field <x> stays at 0)
    m_data = _lin_range([vd * times * nm, d['x_mean'] * nm, [0.0]])
    m_half = max(0.5 * (m_data[1] - m_data[0]), 0.25 * sig_max * nm)
    m_range = [0.5 * (m_data[0] + m_data[1]) - m_half, 0.5 * (m_data[0] + m_data[1]) + m_half]
    # Flux per electron [1/ps] on bins that follow the cloud. The y axis changes smoothly from frame to frame (the
    # peak flux falls as t^-1/2 while the cloud spreads): it spans the drift term v_d P, the diffusion term
    # D P (x - <x>)/s^2 and their sum for a Gaussian P of width s = min(sigma_x, (2Dt)^1/2) (sigma_x: measured;
    # the smaller width gives the higher peak; at high fields sigma_x exceeds (2Dt)^1/2) around <x>, widened
    # by three standard deviations of the counting noise of the bins (velocity spread (D/tau + v_d^2)^1/2, or
    # D/(2^1/2 dx) for the gradient of the diffusion term) and by 10 %.
    fx = d['flux_x'] * nm
    fl = {k: d[f'flux_{k}'] * 1e-12 for k in ('mc', 'drift', 'diff')}
    f_ranges = [None] * len(times)
    for k in range(1, len(times)):
        s_x, dxb = min(np.sqrt(d['var_x'][k]), np.sqrt(2 * D * times[k])), d['flux_x'][k][1] - d['flux_x'][k][0]
        u = np.linspace(-4.0, 4.0, 401)
        Pg = np.exp(-u ** 2 / 2) / (np.sqrt(2 * np.pi) * s_x)
        noise = 3 * np.sqrt(Pg / (res.params['num_particles'] * dxb)) * max(np.sqrt(D / d['tau'] + vd ** 2),
                                                                           D / (np.sqrt(2) * dxb))
        F = np.concatenate([Pg * vd, Pg * D * u / s_x, Pg * (vd + D * u / s_x)])
        f_lo, f_hi = min((F - np.tile(noise, 3)).min(), 0.0), max((F + np.tile(noise, 3)).max(), 0.0)
        f_ranges[k] = [(f_lo - 0.1 * (f_hi - f_lo)) * 1e-12, (f_hi + 0.1 * (f_hi - f_lo)) * 1e-12]
    f_ranges[0] = f_ranges[1]

    def frame_panels(k):
        t = times[k]
        h, _ = np.histogram(X[k] * nm, bins=bins, density=True)
        idx = np.unique(np.r_[np.arange(0, k * d['n_sub'] + 1, stride), k * d['n_sub']])
        cloud = dict(_trace(X[k, :num_plot] * nm, Y[k, :num_plot] * nm, 'electrons', C_E, mode='markers'),
                     marker_size=3, opacity=0.45, nmax=num_plot)
        tracers = [dict(_trace(TX[idx, i] * nm, TY[idx, i] * nm, f'path {i + 1}', c, width=1.2), nmax=1000)
                   for i, c in enumerate(tracer_colors[:TX.shape[1]])]
        theory = drift_diffusion_packet(x_grid / nm, t, vd, D) / nm if t > 0 else np.full(x_grid.size, np.nan)
        return {
            'cloud': _panel('Electrons: free flight + scattering', 'x [nm]', 'y [nm]', [
                cloud, *tracers, _trace([vd * t * nm] * 2, [-y_lim, y_lim], '−μEt (theory)', C_H, 'dash', width=1.5),
                _trace([0.0], [0.0], 'release point', C_T, mode='markers')], xrange=[x_lo, x_hi], yrange=[-y_lim, y_lim]),
            'hist': _panel('Distribution along the field', 'x [nm]', 'P(x) [1/nm]', [
                dict(_trace(bins, np.r_[h, h[-1]], 'Monte Carlo', C_E, width=1), shape='hv', fill='tozeroy',
                     fillcolor=_rgba(C_E, 0.3)),
                _trace(x_grid, theory, 'drift–diffusion theory', C_H)], xrange=[x_lo, x_hi], yrange=[0, p_max]),
            'mean': _panel('Mean position', 'Time [ps]', '⟨x⟩ [nm]', [
                _trace(times * ps, vd * times * nm, '−μEt', C_H, 'dash', width=1.5),
                _trace(times[:k + 1] * ps, d['x_mean'][:k + 1] * nm, '⟨x⟩ (MC)', C_E)], xrange=[0, t_max * ps],
                yrange=m_range),
            'flux': _panel('Electron flux: MC vs drift + diffusion', 'x [nm]', 'Flux per electron [1/ps]', [
                dict(_trace(fx[k], fl['drift'][k] + fl['diff'][k], 'drift + diffusion', _rgba(C_T, 0.35), width=6)),
                _trace(fx[k], fl['drift'][k], 'drift −μE·n', C_E),
                _trace(fx[k], fl['diff'][k], 'diffusion −D dn/dx', C_H, 'dash'),
                _trace(fx[k], fl['mc'][k], 'MC: Σ v_x', C_T, mode='markers')], xrange=[x_lo, x_hi], yrange=f_ranges[k])}

    static = {'cloud': (5,), 'mean': (0,)}                       # Traces that do not change with time
    first = frame_panels(0)
    title = (f"[{res.case}] " if res.case else '') + \
        (f"Monte Carlo drift–diffusion | E = {d['E']:g} V/cm, μ = {d['mu']:g} cm²/V·s, T = {d['T']:g} K<br>"
         f"<sup>τ = {d['tau'] * ps:.3f} ps, mean free path ≈ {d['mfp'] * nm:.0f} nm, "
         f"{d['coll_per_particle']:.0f} collisions per electron</sup>")
    fig, ids = build_figure([first[k] for k in keys], title, bottom=170)
    if 'cloud' in keys:                                          # Equal x and y scales in real space
        sub = fig.get_subplot(keys.index('cloud') // 2 + 1, keys.index('cloud') % 2 + 1)
        fig.update_yaxes(scaleanchor=sub.xaxis.plotly_name.replace('axis', ''), scaleratio=1, constrain='domain',
                         row=keys.index('cloud') // 2 + 1, col=keys.index('cloud') % 2 + 1)
    dynamic = [i for j, key in enumerate(keys) for m, i in enumerate(ids[j]) if m not in static.get(key, ())]
    flux_axis = None
    if 'flux' in keys:                                           # y axis of the flux panel, rescaled per frame
        flux_axis = fig.get_subplot(keys.index('flux') // 2 + 1, keys.index('flux') % 2 + 1).yaxis.plotly_name
    frames = []
    for k in range(len(times)):
        pk = frame_panels(k)
        traces = [_scatter(t, pk[key]['ylog'], None, NMAX_FRAME, pk[key]['xrange'])
                  for key in keys for m, t in enumerate(pk[key]['traces']) if m not in static.get(key, ())]
        layout = {flux_axis: {'range': f_ranges[k]}} if flux_axis else None
        frames.append(go.Frame(data=traces, traces=dynamic, name=str(k), layout=layout))
    fig.frames = frames
    steps = [{"args": [[str(k)], {"mode": "immediate", "frame": {"duration": 0, "redraw": True},
                                  "transition": {"duration": 0}}],
              "label": f"{times[k] * ps:.1f}", "method": "animate"} for k in range(len(times))]
    fig.update_layout(
        updatemenus=[{"type": "buttons", "direction": "left", "showactive": True, "x": 0.0, "y": 0.0,
                      "xanchor": "left", "yanchor": "top", "pad": {"t": 85},
                      "buttons": [
                          {"label": "Play", "method": "animate",
                           "args": [None, {"fromcurrent": True, "frame": {"duration": 80, "redraw": True},
                                           "transition": {"duration": 0}}]},
                          {"label": "Pause", "method": "animate",
                           "args": [[None], {"mode": "immediate", "frame": {"duration": 0, "redraw": False},
                                             "transition": {"duration": 0}}]}]}],
        sliders=[{"active": 0, "x": 0.12, "y": 0.0, "yanchor": "top", "len": 0.78, "pad": {"t": 70},
                  "currentvalue": {"prefix": "Time [ps]: "}, "steps": steps}])
    show_figure(fig)


def plot_operating_point(res, panels=None):
    st, ov = res.data['state'], res.data['overlays']
    builders = {
        'band': lambda: band_panel(st), 'density': lambda: density_panel(st), 'E': lambda: field_panel(st, ov),
        'current_n': lambda: dict(carrier_current_panel(st, 'n'), current_toggle=True),
        'current_p': lambda: dict(carrier_current_panel(st, 'p'), current_toggle=True)}
    keys = list(panels or ['band', 'E', 'current_n', 'current_p', 'density'])
    built = [builders[k]() if k else None for k in keys]
    fig, ids = build_figure(built, f'{_run_title(res)} | V = {st.V:+.3f} V')
    _add_current_toggle(fig, list(zip(keys, built)), ids)
    show_figure(fig)


def plot_result(res, panels=None):
    """Plot a result; panels = list of panel keys (default: the case's panels)."""
    panels = panels if panels is not None else res.panels
    print('\n(3) Visualizing ...')
    if res.params['analysis'] == 'montecarlo':
        print('\tBlue = Monte Carlo electrons, red = drift-diffusion theory; use Play or the slider to step in time.')
        plot_montecarlo(res, panels)
    else:
        print('\tBlue = electrons, Red = holes, Black = total / simulation, Gray = analytic model')
        plot_operating_point(res, panels)
    print('Done!!')


###############################################################################
# Lab 4 Device Schematics
###############################################################################

def schematic_bar(dtype='n', biased=True):
    """Uniformly doped bar between two ohmic contacts; V on the left contact and ground on the right."""
    body = _bar(dtype) + _contacts() + (_circuit() if biased else '') + _origin(_XM) + _dimension(_X0, _X1, 160, 'L')
    return _svg(body, top=45)


def schematic_graded(dtype='n', profile='exponential'):
    """Graded doping (exponential or linear) between two contacts, biased on the left."""
    linear = profile == 'linear'
    dark, light = (SCH_COLORS['n_dark'], '#fdf0f0') if dtype == 'n' else (SCH_COLORS['p_dark'], '#eef3fd')
    gid = f'grad{dtype}{int(linear)}'
    grad = (f"<linearGradient id='{gid}' x1='0' x2='1' y1='0' y2='0'>"
            f"<stop offset='0' stop-color='{dark}'/><stop offset='1' stop-color='{light}'/></linearGradient>")
    N = 'N_{d}' if dtype == 'n' else 'N_{a}'
    law = f'{N}(s) = N(0) + [N(L) − N(0)] s/L' if linear else f'{N}(s) = N(0) e^{{−s/λ}}'
    body = (f"<rect x='{_X0}' y='{_Y0}' width='{_X1 - _X0}' height='{_Y1 - _Y0}' fill='url(#{gid})' "
            f"stroke='{SCH_COLORS['edge']}'/>"
            + _text(_X0 + 12, (_Y0 + _Y1) / 2 + 5, f'{dtype}-type', anchor='start', color='white', weight='bold')
            + _text(_XM, _Y0 - 32, law, size=12) + _text(_XM, _Y0 - 15, 's: distance from the left contact', size=11,
                                                         color=SCH_COLORS['note'])
            + (_text(_X0 + 2, _Y0 - 4, 'N(0)', size=11, anchor='start')
               + _text(_X1 - 2, _Y0 - 4, 'N(L)', size=11, anchor='end') if linear else '')
            + _contacts() + _circuit() + _origin(_XM) + _dimension(_X0, _X1, 160, 'L'))
    return _svg(body, grad, top=40)


def schematic_montecarlo():
    """Electrons released at the origin: zigzag free flights between collisions in a field E along +x."""
    rng = np.random.default_rng(3)
    pts, x, y = [], 300.0, 110.0
    for _ in range(14):                                 # Zigzag path drifting to the left (electrons, E -> +x)
        pts.append((x, y))
        x += rng.uniform(-34, 14)
        y = float(np.clip(y + rng.uniform(-28, 28), 62, 168))
    path = 'M ' + ' L '.join(f'{a:.1f},{b:.1f}' for a, b in pts)
    dots = ''.join(f"<circle cx='{a:.1f}' cy='{b:.1f}' r='2.6' fill='{SCH_COLORS['p_dark']}'/>" for a, b in pts[1:])
    body = ("<rect x='40' y='45' width='400' height='140' rx='6' fill='#f7f9fc' stroke='#c8d0dc'/>"
            + f"<path d='{path}' fill='none' stroke='{SCH_COLORS['p_dark']}' stroke-width='1.6'/>" + dots
            + f"<circle cx='300' cy='110' r='5' fill='white' stroke='{SCH_COLORS['text']}' stroke-width='1.5'/>"
            + _text(300, 200, 'release point (t = 0)', size=11, color=SCH_COLORS['note'])
            + _line(80, 25, 180, 25, width=2) + _arrowhead(186, 25, 0.0, size=9) + _text(200, 30, 'E', weight='bold',
                                                                                         anchor='start')
            + _text(395, 30, 'e⁻ drift: −x', size=12, anchor='end'))
    return _svg(body)
```

```python
###############################################################################
# Interactive Launcher (case selector, parameters, device schematic, result area)
###############################################################################

widget_width = '560px'
widget_description_width = '260px'
widget_style = {'description_width': widget_description_width}


def make_widget_layout(display=''):
    return widgets.Layout(width=widget_width, display=display)


def make_dropdown(options, value, description):
    return widgets.Dropdown(options=options, value=value, description=description, style=widget_style,
                            layout=make_widget_layout())


def make_float_slider(value, min, max, step, description, readout_format='.2f'):
    return widgets.FloatSlider(value=value, min=min, max=max, step=step, description=description,
                               continuous_update=False, orientation='horizontal', readout=True,
                               readout_format=readout_format, style=widget_style, layout=make_widget_layout())


def make_text(value, description):
    """Text input for values in scientific notation (e.g. 1e16)."""
    return widgets.Text(value=value, description=description, continuous_update=False, style=widget_style,
                        layout=make_widget_layout())


def make_int_text(value, min, max, description):
    return widgets.BoundedIntText(value=value, min=min, max=max, description=description, style=widget_style,
                                  layout=make_widget_layout())


def set_widget_visible(widget, visible):
    widget.layout.display = '' if visible else 'none'


# Case parameters in display order: doping and geometry, bias and excitation, material and transport, numerics.
# kind: 'sci' (text, scientific notation), 'text', 'float' (min, max, step, format), 'int' (min, max), 'choice'
PARAM_SPECS = {
    'doping_type':      ('Doping type', 'choice', [('n-type', 'n'), ('p-type', 'p')]),
    'N':                ('Doping N [cm⁻³]', 'sci', None),
    'profile':          ('Doping profile', 'choice', [('Exponential  N(0) e^(−s/λ)', 'exponential'),
                                                      ('Linear  N(0) → N(L)', 'linear')]),
    'N0':               ('N(0) at the left contact [cm⁻³]', 'sci', None),
    'NL':               ('N(L) at the right contact [cm⁻³]', 'sci', None),
    'decay_um':         ('Decay length λ [µm]', 'float', (0.05, 50.0, 0.05, '.2f')),
    'N_min':            ('Lowest doping Na [cm⁻³]', 'sci', None),
    'N_max':            ('Highest doping Na [cm⁻³]', 'sci', None),
    'length_um':        ('Length L [µm]', 'float', (0.1, 500.0, 0.1, '.1f')),
    'ends':             ('Ends of the bar', 'choice', [('No contacts (infinite sample)', 'free'),
                                                       ('Ohmic contacts', 'ohmic')]),
    'set1':             ('Set 1: Na, Nd [cm⁻³]', 'text', None),
    'set2':             ('Set 2: Na, Nd [cm⁻³]', 'text', None),
    'set3':             ('Set 3: Na, Nd [cm⁻³]', 'text', None),
    'Na':               ('Na, p side [cm⁻³]', 'sci', None),
    'Nd':               ('Nd, n side [cm⁻³]', 'sci', None),
    'Lp_um':            ('p-side length Lp [µm]', 'float', (0.1, 200.0, 0.1, '.1f')),
    'Ln_um':            ('n-side length Ln [µm]', 'float', (0.1, 200.0, 0.1, '.1f')),
    'E_field':          ('Electric field E (along +x) [V/cm]', 'float', (0.0, 3000.0, 100.0, '.0f')),
    'V':                ('Bias V (left contact) [V]', 'float', (-20.0, 1.5, 0.01, '.2f')),
    'VR_max':           ('Maximum reverse bias VR [V]', 'float', (0.0, 50.0, 0.1, '.1f')),
    'VR_step':          ('VR step [V]', 'float', (0.05, 5.0, 0.05, '.2f')),
    'V_max':            ('Maximum forward bias [V]', 'float', (0.05, 1.5, 0.01, '.2f')),
    'V_step':           ('Forward bias step [V]', 'float', (0.01, 0.5, 0.01, '.2f')),
    'from_equilibrium': ('Initial guess at each bias', 'choice',
                         [('Previous bias (continuation)', False), ('Equilibrium', True)]),
    'G':                ('Generation rate G (G₀ at the surface) [cm⁻³s⁻¹]', 'sci', None),
    'gen_depth_um':     ('Absorption depth d [µm]', 'float', (0.1, 10.0, 0.1, '.1f')),
    'inj':              ('Injection δ(0) at the left contact [cm⁻³]', 'sci', None),
    'material':         ('Material (Eg; μn/μp)', 'choice',
                         [(f'{m}: Eg {MATERIALS[m].Eg:g} eV; μn/μp {MATERIALS[m].mun:g}/{MATERIALS[m].mup:g} cm²/V·s', m)
                          for m in ('Si', 'GaAs', 'Ge')]),
    'T':                ('Temperature [K]', 'float', (200.0, 500.0, 5.0, '.0f')),
    'ni300':            ('ni at 300 K [cm⁻³]', 'sci', None),
    'tau_us':           ('τn = τp [µs]', 'sci', None),
    'mun':              ('μn [cm²/V·s]', 'float', (1.0, 10000.0, 1.0, '.0f')),
    'mup':              ('μp [cm²/V·s]', 'float', (1.0, 10000.0, 1.0, '.0f')),
    't_max_ps':         ('Simulation time [ps]', 'float', (10.0, 40.0, 5.0, '.0f')),
    'frame_ps':         ('Frame interval [ps]', 'float', (0.2, 1.0, 0.1, '.1f')),
    'srh':              ('Recombination (SRH)', 'choice', [('On', True), ('Off', False)]),
    'thr':              ('Depletion edge |ρ|/(qN)', 'float', (0.05, 0.95, 0.05, '.2f')),
    'npts':             ('Grid points (0 = auto)', 'int', (0, 100001)),
}
# Parameters shown only for one value of another parameter: key -> (controlling key, value)
PARAM_WHEN = {'decay_um': ('profile', 'exponential'), 'NL': ('profile', 'linear')}


def make_param_widget(key):
    label, kind, spec = PARAM_SPECS[key]
    if kind == 'choice':
        return make_dropdown(spec, spec[0][1], label)
    if kind in ('sci', 'text'):
        return make_text('', label)
    if kind == 'int':
        return make_int_text(spec[0], spec[0], spec[1], label)
    lo, hi, step, fmt = spec
    return make_float_slider(lo, lo, hi, step, label, readout_format=fmt)


def to_widget_value(key, value):
    if PARAM_SPECS[key][1] == 'sci':
        return f'{value:g}'.replace('e+', 'e')
    return value


def from_widget_value(key, value):
    if PARAM_SPECS[key][1] == 'sci':
        return to_float(value, PARAM_SPECS[key][0])
    return value


UI = {}                                       # Widgets of the launcher: UI['case'], ['controls'], ['button'], ...
LAST = {}                                     # Last result: LAST['res'] (also LAST[case_id] for each case run)
RESULT = {'handle': None}                     # Result area below the launcher


def _log_html(text):
    return (f"<pre style='font-size:12px; line-height:1.35; margin:4px 0; overflow-x:auto'>"
            f"{html.escape(text)}</pre>") if text else ''


def launch(title=''):
    """
    Display the launcher of a lab: Case selector, case parameters, device schematic and Run Interact button,
    with the result area (log and figure) below. Selecting a case loads its presets; the calculation starts
    only with Run Interact. The result area is an ordinary output updated by display id: ipywidgets Output
    widgets are not used, because some front ends (VS Code) hold all kernel messages until an Output-widget
    update is acknowledged, which can stall the notebook.

    input:
        str         title       Heading of the launcher (cases: CASES of this notebook, run with run_case)
    """
    cases, run = CASES, run_case
    ids = list(cases)
    keys = [k for k in PARAM_SPECS if any(k in cases[c]['params'] for c in ids)]
    case_widget = widgets.Dropdown(options=[(f"{c}  {cases[c]['title']}", c) for c in ids], value=ids[0],
                                   description='Case', style={'description_width': '80px'},
                                   layout=widgets.Layout(width='680px'))
    case_info = widgets.HTML()
    controls = {k: make_param_widget(k) for k in keys}
    button = widgets.Button(description='Run Interact', button_style='primary', icon='play',
                            layout=widgets.Layout(width='180px', margin='12px 0 4px 0'))
    status = widgets.HTML(layout=widgets.Layout(margin='12px 0 4px 12px'))
    schematic = widgets.HTML(layout=widgets.Layout(margin='6px 0 0 24px'))      # Device drawing (SVG)
    busy = {'run': False, 't_end': 0.0}

    def apply_case(change=None):
        cid = case_widget.value
        case = cases[cid]
        case_info.value = (f"<div style='margin:2px 0 10px 86px;color:#444'><b>{cid}</b> — "
                           f"{html.escape(case['summary'])}</div>")
        for k, w in controls.items():
            set_widget_visible(w, k in case['params'])
            if k in case['params']:
                if PARAM_SPECS[k][1] == 'float':           # Slider range: case override or default
                    lo, hi, step = case.get('ranges', {}).get(k, PARAM_SPECS[k][2][:3])
                    if lo > w.max:
                        w.max, w.min = hi, lo
                    else:
                        w.min, w.max = lo, hi
                    w.step = step
                w.value = to_widget_value(k, case['params'][k])
        update_visibility()

    def update_visibility(change=None):
        params = cases[case_widget.value]['params']
        for k, (key, value) in PARAM_WHEN.items():
            if k in controls and k in params:
                set_widget_visible(controls[k], key not in controls or controls[key].value == value)
        update_schematic()

    def update_schematic(change=None):
        case = cases[case_widget.value]
        values = {k: controls[k].value for k in case['params'] if PARAM_SPECS[k][1] == 'choice'}
        schematic.value = case['schematic'](values) if case.get('schematic') else ''

    def on_run(_=None):
        # The printed log and the figure are collected into one HTML block that replaces the result area.
        global FIGURE_SINK
        if busy['run'] or time.time() - busy['t_end'] < 0.5:     # Running, or a click queued during the run
            return
        busy['run'], button.disabled = True, True
        status.value = '<i>Running ...</i>'
        shown, log = [], io.StringIO()

        def flush_log():
            shown.append(_log_html(log.getvalue()))
            log.seek(0)
            log.truncate(0)

        def add_figure(fig):
            flush_log()
            shown.append(launcher_figure_html(fig)[0])

        t0 = time.time()
        FIGURE_SINK = add_figure
        try:
            with contextlib.redirect_stdout(log), contextlib.redirect_stderr(log):
                try:
                    cid = case_widget.value
                    values = {k: from_widget_value(k, controls[k].value) for k in cases[cid]['params']}
                    res = run(cid, **values)
                    LAST['res'] = LAST[cid] = res
                    plot_result(res)
                    print(f"(Result stored in LAST['{cid}'] and LAST['res'])")
                except ValueError as err:
                    print(f'Input error: {err}')
                except Exception:
                    traceback.print_exc()
            flush_log()
            if RESULT['handle'] is not None:
                RESULT['handle'].update(HTML(''.join(shown)))
        finally:
            FIGURE_SINK = None
            busy['run'], button.disabled = False, False
            busy['t_end'] = time.time()
            status.value = f'Finished in {time.time() - t0:.1f} s'

    apply_case()
    case_widget.observe(apply_case, names='value')
    for k in ('doping_type', 'ends'):
        if k in controls:
            controls[k].observe(update_schematic, names='value')
    if 'profile' in controls:
        controls['profile'].observe(update_visibility, names='value')
    button.on_click(on_run)
    heading = widgets.HTML(f"<h3 style='margin:4px 0 8px 0'>{html.escape(title)}</h3>") if title else None
    params_box = widgets.VBox([*controls.values(), widgets.HBox([button, status])])
    body = widgets.HBox([params_box, schematic],                  # Drawing to the right of the parameters
                        layout=widgets.Layout(flex_flow='row wrap', align_items='flex-start'))
    panel = widgets.VBox([w for w in (heading, case_widget, case_info, body) if w is not None])
    UI.update(case=case_widget, controls=controls, button=button, panel=panel, run=on_run, schematic=schematic)
    PLOTLY_JS_SENT['done'] = False                            # The next figure embeds plotly.js
    display(panel)
    RESULT['handle'] = display(HTML("<p style='color:#777'><i>Select a case and click Run Interact.</i></p>"),
                               display_id=True)                # Result area, updated by Run Interact
```

# (3) Lab 4 — Drift and diffusion currents

Each carrier current is the sum of a drift term, driven by the electric field, and a diffusion term, driven by the concentration gradient [1, Ch. 5], (5.35):

$$J_n=\underbrace{q\mu_n n\,\mathcal{E}}_{\text{drift}}+\underbrace{qD_n\frac{dn}{dx}}_{\text{diffusion}},\qquad J_p=\underbrace{q\mu_p p\,\mathcal{E}}_{\text{drift}}-\underbrace{qD_p\frac{dp}{dx}}_{\text{diffusion}}$$

- In a uniformly doped bar under bias, the concentrations are uniform, diffusion vanishes and $J=\sigma\mathcal{E}$ with $\sigma=e(\mu_nn_0+\mu_pp_0)$ (5.19).
- With nonuniform doping at equilibrium, the majority current must vanish (5.43), so an induced field cancels diffusion. With quasi-neutrality, $n\approx N_d(x)$ (n-type) or $p\approx N_a(x)$ (p-type) (5.39),

$$\mathcal{E}_x=\mp\frac{kT}{e}\,\frac{1}{N(x)}\frac{dN(x)}{dx}\qquad(5.42),\quad -\ \text{n-type},\ +\ \text{p-type}.$$

  - **Exponential grading**, $N(s)=N(0)e^{-s/\lambda}$ ($s$ from the left contact): the field is uniform, $\lvert\mathcal{E}\rvert=(kT/e)/\lambda$.
  - **Linear grading**, $N(s)=N(0)+[N(L)-N(0)]\,s/L$: $\mathcal{E}(s)=\mp\frac{kT}{e}\frac{N(L)-N(0)}{L\,N(s)}$. The field is not uniform; it is largest where the doping is lowest.

  The field points so that it pushes majority carriers back toward the heavily doped side, so its sign flips between n- and p-type. Under bias, the bar carries a current $J$ that is constant in $x$; with the majority density equal to $N(x)$ the field becomes $\mathcal{E}(x)=\mathcal{E}_{bi}(x)+J/\sigma(x)$ with $\sigma=e\mu_{maj}N(x)$, and $V=J\int dx/\sigma(x)$. The field therefore grows where the doping is low.

  The simulation uses exactly this quasi-neutral picture (the Poisson equation is not solved in this lab), so its field and current agree with (5.42) and $\mathcal{E}_{bi}+J/\sigma$ to the discretization error ($\sim10^{-5}$); the case then shows how the two currents split into drift and diffusion. Strictly, the induced field needs a small space charge (Gauss's law, $\rho=\varepsilon_s\,d\mathcal{E}/dx$); Neamen notes that it is a small fraction of the impurity concentration (Sec. 5.3.1). Lab 6 solves the Poisson equation, where the space charge of the pn junction is essential.
- The simulation uses $D=\mu kT/e$ (5.47); the Einstein relation is an input of the drift–diffusion model, not a result. Case 4-MC obtains it from scattering instead.
- **Band gap and temperature.** The material menu and the band-diagram title show $E_g$. The **Temperature** control changes $kT/e$ and $n_i(T)=n_i(300)(T/300)^{3/2}e^{-E_g/2k(1/T-1/300)}$; $E_g$ and the mobilities stay at their 300 K values (see **Textbook assumptions**: Neamen attributes the main temperature dependence of transport to $\mu(T)$, which this model leaves out).
  - 4-B: the built-in field is proportional to $kT/e$ (259 V/cm at 300 K for $\lambda=1$ µm).
  - 4-A: $\sigma$ stays constant while $n_i\ll N$. In Ge with $N=10^{15}$ cm⁻³, intrinsic carriers take over above about 400 K and $\sigma$ rises 13-fold at 500 K.

### Monte Carlo picture of drift and diffusion (4-MC)

Drift and diffusion are averages over many carriers that scatter randomly. Case 4-MC follows 20000 electrons ($m^*=0.26\,m_0$, Si) released at the origin with thermal velocities. Between collisions each electron accelerates in the field (free flight); in every short time step it scatters with probability $\Delta t/\tau$, after which its velocity is drawn again from the thermal (Maxwell) distribution. The mean time between collisions follows from the mobility, $\tau=\mu m^*/q$ (0.2 ps for $\mu=1350$ cm²/V·s).

- **Drift:** after many collisions the cloud moves with the drift velocity $v_d=-\mu\mathcal{E}$, so $\langle x\rangle=-\mu\mathcal{E}t$; the slope of $\langle x\rangle(t)$ gives $\mu$ back.
- **Diffusion:** the cloud spreads as a Gaussian with $\sigma^2=2Dt$. Across the field ($y$) the spread comes from scattering alone, so its slope gives $D$; along the field ($x$) the same $D$ applies.
- **Einstein relation:** the extracted $D/\mu$ equals $kT/q$ (0.0259 V at 300 K) within the statistical error of a few percent. The relation follows from the scattering picture itself: the same collisions set both the response to the field ($\mu$) and the random walk ($D$).
- **Distribution along the field:** the histogram follows $P(x,t)=\exp[-(x+\mu\mathcal{E}t)^2/4Dt]/\sqrt{4\pi Dt}$, the drift–diffusion solution without recombination that also describes the Haynes–Shockley experiment.
- **Current = drift + diffusion:** the particle current is what the electrons actually carry, $\sum v_x$ over the electrons in each slice of $x$ (electron current $J_n=-e\sum v_x$ per volume). No current equation is used to compute it. After a few collision times it equals the sum of the two terms of (5.35), written per electron with the normalized density $P(x)$: the drift term $-\mu\mathcal{E}P$ ($J_{drf}=e\mu_nn\mathcal{E}$) and the diffusion term $-D\,dP/dx$ ($J_{dif}=eD_n\,dn/dx$). On the leading side of the moving cloud both terms push in the drift direction; on the trailing side diffusion opposes drift, so the net flux is smaller there. During the first few $\tau$ the electrons still move ballistically from the origin, so the two do not agree yet: with 20000 electrons the residual $\mathrm{rms}(F_{MC}-F_{drift}-F_{diff})/\max\lvert F_{MC}\rvert$ is about 0.3 at $1.5\tau$ and 0.13 at $4.5\tau$, and from about $7\tau$ on it stays at the statistical noise of 0.05–0.1 (row `flux_resid`, last frame).
- **Panels:** (1) electrons in real space with three zigzag paths (each corner is a collision) and the theoretical center $-\mu\mathcal{E}t$; (2) histogram along the field and the analytic packet; (3) $\langle x\rangle(t)$ against $-\mu\mathcal{E}t$; (4) particle flux per electron: MC (markers), drift term (solid), diffusion term (dashed) and their sum (thick pale line); its $y$ axis zooms in smoothly as the cloud spreads. The spreading $\sigma_y^2=2Dt$ is reported in the table (rows `D_y`, `D_x`). Use **Play** or the slider to step in time.
- The model has no recombination, no velocity saturation and a collision time independent of energy, so it holds at low fields. **Temperature** changes the thermal velocity and $kT$, and therefore $D$ but not $\mu$.
